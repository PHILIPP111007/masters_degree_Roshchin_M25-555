ОЖИДАЕМЫЙ РЕЗУЛЬТАТ:
Презентация с выводами и рекомендациями, которая поможет инвесторам принять обоснованное решение об открытии заведения общественного питания в Москве.

# ОПИСАНИЕ ДАННЫХ

ДАТАСЕТ: moscow_places.csv

Источник: Яндекс Карты и Яндекс Бизнес (лето 2022 года)

СТРУКТУРА ДАННЫХ:

1. name - название заведения
2. address - адрес заведения
3. category - категория заведения (кафе, ресторан, пиццерия, кофейня и т.д.)
4. hours - информация о днях и часах работы
5. lat - широта географической точки
6. lng - долгота географической точки
7. rating - рейтинг заведения по оценкам пользователей Яндекс Карт (макс. 5.0)
8. price - категория цен в заведении (средние, ниже среднего, выше среднего)
9. avg_bill - строка с описанием средней стоимости заказа
10. middle_avg_bill - числовая оценка среднего чека (только для записей, 
    начинающихся с "Средний счёт")
11. middle_coffee_cup - числовая оценка стоимости чашки капучино (только для 
    записей, начинающихся с "Цена одной чашки капучино")
12. chain - является ли заведение сетевым (0 - нет, 1 - да)
13. district - административный район Москвы
14. seats - количество посадочных мест

````python
print("\n" + "="*80)
print("ПРОМЕЖУТОЧНЫЙ ВЫВОД: ПЕРВИЧНЫЙ АНАЛИЗ ДАННЫХ")
print("="*80)

print("\n✅ ОТВЕТЫ НА ВОПРОСЫ ЗАКАЗЧИКА:")
print("-" * 60)

print("\n1. СКОЛЬКО ЗАВЕДЕНИЙ ПРЕДСТАВЛЕНО?")
print(f"   • Всего в датасете: {df.shape[0]:,} заведений".replace(',', ' '))
print(f"   • Количество столбцов с информацией: {df.shape[1]}")
print(f"   • Уникальных названий заведений: {df['name'].nunique()}")

print("\n2. ЧТО МОЖНО СКАЗАТЬ О КАЖДОМ СТОЛБЦЕ?")

# Анализируем каждый столбец по отдельности
print("\n   📍 name (название заведения):")
print(f"      • Уникальных названий: {df['name'].nunique()}")
print(f"      • Самое частое название: '{df['name'].value_counts().index[0]}'")
print(f"      • Встречается {df['name'].value_counts().iloc[0]} раз")
print(f"      • Пропусков: {df['name'].isnull().sum()}")

print("\n   📍 category (категория заведения):")
print(f"      • Всего категорий: {df['category'].nunique()}")
print(f"      • Самая популярная категория: '{df['category'].value_counts().index[0]}'")
print(f"      • Занимает {df['category'].value_counts().iloc[0]} записей")
print(f"      • Доля от общего числа: {(df['category'].value_counts().iloc[0]/len(df)*100):.1f}%")
print("      • Все категории: " + ", ".join(sorted(df['category'].unique())))

print("\n   📍 address (адрес):")
print(f"      • Уникальных адресов: {df['address'].nunique()}")
print(f"      • Самый частый адрес: '{df['address'].value_counts().index[0]}'")
print(f"      • Встречается {df['address'].value_counts().iloc[0]} раз")
print(f"      • Пропусков: {df['address'].isnull().sum()}")

print("\n   📍 district (район):")
print(f"      • Всего административных районов: {df['district'].nunique()}")
print(f"      • Самый представленный район: '{df['district'].value_counts().index[0]}'")
print(f"      • Встречается в {df['district'].value_counts().iloc[0]} записях")
print(f"      • Доля: {(df['district'].value_counts().iloc[0]/len(df)*100):.1f}%")

print("\n   📍 hours (часы работы):")
print(f"      • Уникальных форматов записи времени: {df['hours'].nunique()}")
print(f"      • Самый частый график: '{df['hours'].value_counts().index[0]}'")
print(f"      • Встречается {df['hours'].value_counts().iloc[0]} раз")
print(f"      • Пропусков: {df['hours'].isnull().sum()} ({df['hours'].isnull().sum()/len(df)*100:.1f}%)")

print("\n   📍 price (ценовая категория):")
print(f"      • Уникальных категорий цен: {df['price'].nunique()}")
print(f"      • Самая частая категория: '{df['price'].value_counts().index[0]}'")
print(f"      • Встречается {df['price'].value_counts().iloc[0]} раз")
print(f"      • Пропусков: {df['price'].isnull().sum()} ({df['price'].isnull().sum()/len(df)*100:.1f}%)")

print("\n   📍 avg_bill (информация о среднем чеке):")
print(f"      • Уникальных описаний: {df['avg_bill'].nunique()}")
print(f"      • Самый частый формат: '{df['avg_bill'].value_counts().index[0]}'")
print(f"      • Встречается {df['avg_bill'].value_counts().iloc[0]} раз")
print(f"      • Пропусков: {df['avg_bill'].isnull().sum()} ({df['avg_bill'].isnull().sum()/len(df)*100:.1f}%)")

print("\n3. ЗНАЧЕНИЯ КАКОГО ТИПА ОНИ ХРАНЯТ?")
print("   • Текстовые данные (object):")
print("     - name, address, category, hours, price, avg_bill")
print("   • Категориальные данные: category, district, price")
print("   • Географические данные: address (адреса для геокодирования)")
print("   • Временные данные: hours (график работы)")
print("   • Финансовые данные: avg_bill (стоимость в текстовом формате)")
````

## ℹ️ Шаг 2.3 Создание столбцов

---
    
Выполните предобработку данных:
- Создайте столбец `street` с названиями улиц из столбца с адресом.
- Создайте столбец `is_24_7` с обозначением, что заведение работает ежедневно и круглосуточно (24/7): 
    - логическое значение True — если заведение работает ежедневно и круглосуточно;
    - логическое значение False — в противоположном случае.    

Создание столбца street (название улицы)

````python
def extract_street(address):
    if pd.isna(address):
        return None
    
    # Разбиваем адрес на части
    parts = str(address).split(',')
    
    # Ищем улицу (обычно это второй элемент после номера дома)
    for part in parts:
        part = part.strip()
        # Проверяем, содержит ли часть слово "улица", "проспект", "бульвар" и т.д.
        street_keywords = ['ул.', 'улица', 'пр.', 'проспект', 'б-р', 'бульвар', 
                          'ш.', 'шоссе', 'наб.', 'набережная', 'аллея', 'пер.', 'переулок']
        
        for keyword in street_keywords:
            if keyword in part.lower():
                return part
    
    # Если не нашли, возвращаем второй элемент (обычно это улица)
    if len(parts) > 1:
        return parts[1].strip()
    
    return None
````

Применяем функцию к столбцу address

````python
df['street'] = df['address'].apply(extract_street)
````

Проверяем результат

````python
print(f"\nУдалось извлечь улицы для {df['street'].notna().sum()} / {len(df)} записей")
print("\nПримеры извлеченных улиц:")
print(df['street'].head(10))
````

Создание столбца is_24_7 (круглосуточные заведения)

````python
def check_24_7(hours):
    if pd.isna(hours):
        return False
    
    hours_str = str(hours).lower()
    
    # Проверяем различные варианты обозначения круглосуточной работы
    patterns = [
        r'24/7',
        r'круглосуточно',
        r'ежедневно.*24.*часа',
        r'24 часа',
        r'0:00.*24:00',
        r'00:00.*24:00',
        r'00:00.*00:00'
    ]
    
    for pattern in patterns:
        if re.search(pattern, hours_str):
            return True
    
    # Дополнительная проверка для формата "пн-вс 0:00-24:00"
    if re.search(r'пн.*вс.*0.*24', hours_str) or re.search(r'ежедневно.*0.*24', hours_str):
        return True
    
    return False
````

Применяем функцию

````python
df['is_24_7'] = df['hours'].apply(check_24_7)

print(f"\nКоличество круглосуточных заведений: {df['is_24_7'].sum()}")

# Преобразование типов данных
df['seats'] = pd.to_numeric(df['seats'], errors='coerce')
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
````

````python
def check_24_7_corrected(hours):
    """
    Определяет, работает ли заведение круглосуточно.
    Учитывает различные форматы записи времени работы.
    """
    if pd.isna(hours):
        return False
    
    hours_str = str(hours).lower()
    
    # Очищаем строку от лишних пробелов и символов
    hours_str = hours_str.strip()
    
    # Если строка слишком короткая или содержит ошибки, пропускаем
    if len(hours_str) < 5:
        return False
    
    # 1. Прямые указания на круглосуточную работу
    round_the_clock_keywords = [
        'круглосуточно',
        '24/7',
        '24 часа',
        'круглосут',
        '24часа',
        '24 ч',
        '24ч',
        '24 часа в сутки',
        'ежедневно, круглосуточно',
        'работаем круглосуточно',
        'открыто круглосуточно',
        'круглосуточный',
        '24 7',
        '24 часа без перерыва'
    ]
    
    for keyword in round_the_clock_keywords:
        if keyword in hours_str:
            return True
    
    # 2. Проверяем формат времени, охватывающий все 24 часа
    time_patterns = [
        # Полные форматы с двоеточием
        r'00:00.*24:00',
        r'0:00.*24:00',
        r'00:00.*00:00',  # тоже круглосуточно
        r'0:00.*0:00',
        
        # Форматы с тире вместо двоеточия
        r'00-00.*24-00',
        r'0-00.*24-00',
        r'00-00.*00-00',
        
        # Форматы без разделителей
        r'0000.*2400',
        r'00 00.*24 00',
        
        # Форматы с точками
        r'00.00.*24.00',
        
        # Любой формат от 0 до 24
        r'\b0[:\-\s\.]?00.*24[:\-\s\.]?00\b',
        r'\b00[:\-\s\.]?00.*00[:\-\s\.]?00\b'  # от 00:00 до 00:00
    ]
    
    for pattern in time_patterns:
        if re.search(pattern, hours_str, re.IGNORECASE):
            return True
    
    # 3. Проверяем по дням недели с указанием 0-24 часов
    days_of_week = ['пн', 'вт', 'ср', 'чт', 'пт', 'сб', 'вс', 'ежедневно']
    days_patterns = [
        # Любой день недели с 0 до 24
        r'[\w\-]*0.*24',
        r'[\w\-]*00.*24',
        
        # Конкретные форматы
        r'пн.*вс.*0.*24',
        r'пн.*вс.*00.*24',
        r'ежедневно.*0.*24',
        r'ежедневно.*00.*24',
        r'пн-вс.*0.*24',
        r'пн-вс.*00.*24',
        
        # С понедельника по воскресенье
        r'с.*пн.*по.*вс.*0.*24',
        r'с.*пн.*по.*вс.*00.*24',
        
        # Без выходных
        r'без.*выходных.*0.*24',
        r'без.*выходных.*00.*24'
    ]
    
    for pattern in days_patterns:
        if re.search(pattern, hours_str, re.IGNORECASE):
            return True
    
    # 4. Проверяем, есть ли указание на работу все 24 часа
    if re.search(r'24\s*час', hours_str) or re.search(r'24\s*часов', hours_str):
        # Убедимся, что это именно про время работы, а не про что-то другое
        time_related = any(word in hours_str for word in ['работа', 'открыт', 'закрыт', 'час', 'время'])
        if time_related:
            return True
    
    return False

# Применяем исправленную функцию
df['is_24_7'] = df['hours'].apply(check_24_7_corrected)

print(f"✅ Столбец is_24_7 создан успешно")
print(f"📊 Количество круглосуточных заведений: {df['is_24_7'].sum():,}".replace(',', ' '))
print(f"📈 Доля круглосуточных заведений: {(df['is_24_7'].sum() / len(df) * 100):.2f}%")

# Проверяем примеры круглосуточных заведений
print("\n🔍 ПРИМЕРЫ КРУГЛОСУТОЧНЫХ ЗАВЕДЕНИЙ:")
round_the_clock_examples = df[df['is_24_7'] == True][['name', 'category', 'hours']].head(10)
print(f"Найдено примеров: {len(round_the_clock_examples)}")
for idx, row in round_the_clock_examples.iterrows():
    hours_display = str(row['hours'])[:50] + "..." if len(str(row['hours'])) > 50 else str(row['hours'])
    print(f"• {row['name']} ({row['category']}): {hours_display}")

# Проверяем распределение по категориям
print("\n📊 РАСПРЕДЕЛЕНИЕ КРУГЛОСУТОЧНЫХ ЗАВЕДЕНИЙ ПО КАТЕГОРИЯМ:")
if df['is_24_7'].sum() > 0:
    round_the_clock_by_category = df[df['is_24_7'] == True]['category'].value_counts().head(10)
    total_24_7 = df['is_24_7'].sum()
    
    for category, count in round_the_clock_by_category.items():
        percentage = (count / total_24_7 * 100)
        total_in_category = df[df['category'] == category].shape[0]
        category_percentage = (count / total_in_category * 100)
        print(f"  {category}: {count} заведений ({percentage:.1f}% от всех круглосуточных, {category_percentage:.1f}% в категории)")
else:
    print("Круглосуточных заведений не найдено")

# Дополнительная валидация
print("\n🔍 ПРОВЕРКА КАЧЕСТВА ДАННЫХ:")
print("-" * 40)

# Проверяем заведения с упоминанием "24" в часах работы
hours_with_24 = df[df['hours'].notna() & df['hours'].str.contains('24', na=False)]
print(f"Заведений с упоминанием '24' в часах: {len(hours_with_24)}")

if len(hours_with_24) > 0:
    # Смотрим, какие из них определены как круглосуточные
    correctly_identified = hours_with_24[hours_with_24['is_24_7'] == True]
    not_identified = hours_with_24[hours_with_24['is_24_7'] == False]
    
    print(f"Из них определены как круглосуточные: {len(correctly_identified)}")
    print(f"Не определены как круглосуточные: {len(not_identified)}")
    
    if len(not_identified) > 0:
        print("\nПримеры неопределенных заведений (первые 3):")
        for idx, row in not_identified.head(3).iterrows():
            print(f"• {row['name']}: {row['hours']}")
            # Проверяем почему не определилось
            test_result = check_24_7_corrected(row['hours'])
            print(f"  Проверка функции: {test_result}")

# Проверяем заведения с упоминанием "круглосуточно"
hours_with_kruglosut = df[df['hours'].notna() & df['hours'].str.contains('круглосут', na=False, case=False)]
print(f"\nЗаведений с упоминанием 'круглосут' в часах: {len(hours_with_kruglosut)}")

if len(hours_with_kruglosut) > 0:
    correctly_identified_krug = hours_with_kruglosut[hours_with_kruglosut['is_24_7'] == True]
    print(f"Из них определены как круглосуточные: {len(correctly_identified_krug)}")

# =============================================================================
# АНАЛИТИЧЕСКИЙ ВЫВОД ПО КРУГЛОСУТОЧНЫМ ЗАВЕДЕНИЯМ
# =============================================================================

print("\n" + "="*80)
print("АНАЛИТИЧЕСКИЙ ВЫВОД: КРУГЛОСУТОЧНЫЕ ЗАВЕДЕНИЯ В МОСКВЕ")
print("="*80)

total_establishments = len(df)
total_24_7 = df['is_24_7'].sum()
percentage_24_7 = (total_24_7 / total_establishments * 100)

print(f"\n📊 ОБЩАЯ СТАТИСТИКА:")
print(f"• Всего заведений в датасете: {total_establishments:,}".replace(',', ' '))
print(f"• Круглосуточных заведений: {total_24_7:,}".replace(',', ' '))
print(f"• Доля круглосуточных заведений: {percentage_24_7:.2f}%")
print(f"• Примерно каждое {int(100/percentage_24_7) if percentage_24_7 > 0 else 'N/A'}-е заведение работает круглосуточно")

if total_24_7 > 0:
    # Анализ по категориям
    top_categories_24_7 = df[df['is_24_7'] == True]['category'].value_counts().head(5)
    
    print(f"\n🏆 ТОП-5 КАТЕГОРИЙ КРУГЛОСУТОЧНЫХ ЗАВЕДЕНИЙ:")
    for i, (category, count) in enumerate(top_categories_24_7.items(), 1):
        cat_percentage = (count / total_24_7 * 100)
        overall_cat_count = df[df['category'] == category].shape[0]
        cat_24_7_percentage = (count / overall_cat_count * 100)
        print(f"  {i}. {category}:")
        print(f"     • Круглосуточных: {count} ({cat_percentage:.1f}% от всех круглосуточных)")
        print(f"     • Доля в категории: {cat_24_7_percentage:.1f}% от всех заведений этой категории")
    
    # Проверка качества определения
    print(f"\n🔍 КАЧЕСТВО ОПРЕДЕЛЕНИЯ:")
    print(f"• Заведений с '24' в часах работы: {len(hours_with_24)}")
    if len(hours_with_24) > 0:
        accuracy = (len(correctly_identified) / len(hours_with_24) * 100)
        print(f"• Из них правильно определены: {len(correctly_identified)} ({accuracy:.1f}%)")
    
    print(f"• Заведений с 'круглосут' в часах работы: {len(hours_with_kruglosut)}")
    if len(hours_with_kruglosut) > 0:
        accuracy_krug = (len(correctly_identified_krug) / len(hours_with_kruglosut) * 100)
        print(f"• Из них правильно определены: {len(correctly_identified_krug)} ({accuracy_krug:.1f}%)")
````

АНАЛИЗ ПОСАДОЧНЫХ МЕСТ ПО КАТЕГОРИЯМ

````python
seats_df = df.dropna(subset=['seats'])

seats_by_category = seats_df.groupby('category')['seats'].agg(['mean', 'median', 'count']).round(1)
seats_by_category = seats_by_category.sort_values('median', ascending=False)

print("Топ-10 категорий по медианному количеству мест:")
print(seats_by_category.head(10))
````

Визуализация: boxplot по основным категориям

````python
plt.figure(figsize=(14, 8))
main_cats_seats = seats_df[seats_df['category'].isin(main_categories)]

# Убираем выбросы для лучшей визуализации
Q1 = main_cats_seats['seats'].quantile(0.25)
Q3 = main_cats_seats['seats'].quantile(0.75)
IQR = Q3 - Q1
filtered_seats = main_cats_seats[
    (main_cats_seats['seats'] >= Q1 - 1.5*IQR) & 
    (main_cats_seats['seats'] <= Q3 + 1.5*IQR)
]

sns.boxplot(data=filtered_seats, x='seats', y='category', color='steelblue', order=filtered_seats.groupby('category')['seats'].median().sort_values(ascending=False).index)
plt.xlabel('Количество посадочных мест')
plt.ylabel('Категория')
plt.title('Распределение посадочных мест по основным категориям', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
````

Из графика видно, что рестораны имеют наибольшее количество посадочных мест, за ними следуют кофейни и столовые. Кафе и пиццерии имеют примерно одинаковое количество посадочных мест, которое меньше, чем у предыдущих категорий. Это может свидетельствовать о том, что рестораны, как правило, рассчитаны на большее количество посетителей одновременно, в то время как кафе и пиццерии могут быть более компактными и ориентированы на меньшие группы людей

````python
# 3.3 Соотношение сетевых и несетевых заведений
print("\nСООТНОШЕНИЕ СЕТЕВЫХ И НЕСЕТЕВЫХ ЗАВЕДЕНИЙ")
print("=" * 50)

chain_counts = df['chain'].value_counts()
chain_percent = (chain_counts / len(df)) * 100

print(f"Сетевые заведения: {chain_counts.get(1, 0)} ({chain_percent.get(1, 0):.1f}%)")
print(f"Несетевые заведения: {chain_counts.get(0, 0)} ({chain_percent.get(0, 0):.1f}%)")

# Визуализация
plt.figure(figsize=(8, 8))
labels = ['Несетевые', 'Сетевые']
colors = ['#ff9999', '#66b3ff']
plt.pie([chain_counts.get(0, 0), chain_counts.get(1, 0)], 
        labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('Соотношение сетевых и несетевых заведений', fontsize=16, fontweight='bold')
plt.axis('equal')
plt.show()

# 3.4 Какие категории чаще являются сетевыми
print("\nКАТЕГОРИИ, ЧАЩЕ ЯВЛЯЮЩИЕСЯ СЕТЕВЫМИ")
print("=" * 50)

# Группируем по категориям и считаем долю сетевых
chain_by_category = df.groupby('category').agg(
    total=('chain', 'count'),
    chain_count=('chain', lambda x: (x == 1).sum())
)
chain_by_category['chain_percent'] = (chain_by_category['chain_count'] / chain_by_category['total'] * 100).round(1)
chain_by_category = chain_by_category.sort_values('chain_percent', ascending=False)

print("Топ-10 категорий по доле сетевых заведений:")
print(chain_by_category.head(10))

# Визуализация (только категории с достаточным количеством заведений)
filtered_chain = chain_by_category[chain_by_category['total'] >= 10].head(15)

plt.figure(figsize=(14, 8))
bars = plt.barh(filtered_chain.index[::-1], filtered_chain['chain_percent'][::-1])
plt.xlabel('Доля сетевых заведений, %')
plt.title('Категории с наибольшей долей сетевых заведений', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем значения
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height()/2, 
             f'{width:.1f}%', ha='left', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# 3.5 Топ-15 популярных сетей
print("\nТОП-15 ПОПУЛЯРНЫХ СЕТЕЙ")
print("=" * 50)

# Группируем по названиям заведений
chain_names = df[df['chain'] == 1]['name'].value_counts().head(15)

print("Топ-15 сетей по количеству заведений:")
for i, (name, count) in enumerate(chain_names.items(), 1):
    # Находим категорию для этой сети
    category = df[df['name'] == name]['category'].iloc[0] if not df[df['name'] == name].empty else 'Неизвестно'
    print(f"{i:2}. {name:40} - {count:3} заведений ({category})")

# Визуализация
plt.figure(figsize=(14, 10))
bars = plt.barh(range(len(chain_names)), chain_names.values[::-1])
plt.yticks(range(len(chain_names)), chain_names.index[::-1])
plt.xlabel('Количество заведений')
plt.title('Топ-15 популярных сетей в Москве', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем значения
for i, bar in enumerate(bars):
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height()/2, 
             f'{int(width)}', ha='left', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# 3.6 Распределение заведений по административным районам
print("\nРАСПРЕДЕЛЕНИЕ ЗАВЕДЕНИЙ ПО РАЙОНАМ")
print("=" * 50)

district_counts = df['district'].value_counts()
print(f"Всего районов: {len(district_counts)}")
print("\nТоп-10 районов по количеству заведений:")
print(district_counts.head(10))

# Визуализация: распределение по районам
plt.figure(figsize=(14, 10))
top_districts = district_counts.head(15)
bars = plt.barh(top_districts.index[::-1], top_districts.values[::-1])
plt.xlabel('Количество заведений')
plt.title('Топ-15 районов по количеству заведений общественного питания', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем значения
for bar in bars:
    width = bar.get_width()
    plt.text(width + 5, bar.get_y() + bar.get_height()/2, 
             f'{int(width)}', ha='left', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# Распределение категорий по районам (heatmap)
# Берем топ-10 районов и топ-10 категорий
top_10_districts = district_counts.head(10).index
top_10_categories = category_counts.head(10).index

# Создаем сводную таблицу
pivot_data = df[df['district'].isin(top_10_districts) & df['category'].isin(top_10_categories)]
pivot_table = pd.crosstab(pivot_data['district'], pivot_data['category'])

# Нормализуем по строкам для лучшей визуализации
pivot_table_norm = pivot_table.div(pivot_table.sum(axis=1), axis=0) * 100

plt.figure(figsize=(16, 10))
sns.heatmap(pivot_table_norm, annot=True, fmt='.1f', cmap='YlOrRd', linewidths=0.5)
plt.title('Доля категорий заведений по районам (%)', fontsize=16, fontweight='bold')
plt.xlabel('Категория')
plt.ylabel('Район')
plt.tight_layout()
plt.show()

# 3.7 Распределение средних рейтингов по категориям
print("\nРАСПРЕДЕЛЕНИЕ РЕЙТИНГОВ ПО КАТЕГОРИЯМ")
print("=" * 50)

# Удаляем пропуски в рейтингах
rating_df = df.dropna(subset=['rating'])

# Группируем по категориям
rating_by_category = rating_df.groupby('category')['rating'].agg(['mean', 'median', 'count', 'std']).round(3)
rating_by_category = rating_by_category[rating_by_category['count'] >= 10].sort_values('mean', ascending=False)

print("Топ-10 категорий по среднему рейтингу:")
print(rating_by_category.head(10))

print("\nНиз-10 категорий по среднему рейтингу:")
print(rating_by_category.tail(10))

# Визуализация: boxplot рейтингов по основным категориям
plt.figure(figsize=(14, 8))
main_cats_rating = rating_df[rating_df['category'].isin(main_categories)]

sns.boxplot(data=main_cats_rating, x='rating', y='category', 
            order=main_cats_rating.groupby('category')['rating'].median().sort_values(ascending=False).index)
plt.xlabel('Рейтинг')
plt.ylabel('Категория')
plt.title('Распределение рейтингов по основным категориям', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# 3.8 Карта с кластерами заведений
print("\nСОЗДАНИЕ КАРТЫ С КЛАСТЕРАМИ")
print("=" * 50)

# Создаем базовую карту Москвы
moscow_center = [55.7558, 37.6173]
m = folium.Map(location=moscow_center, zoom_start=11, tiles='CartoDB positron')

# Добавляем кластеры
marker_cluster = MarkerCluster().add_to(m)

# Добавляем заведения (выборочно для производительности)
sample_df = df.dropna(subset=['lat', 'lng']).sample(n=1000, random_state=42)

for idx, row in sample_df.iterrows():
    popup_text = f"""
    <b>{row['name']}</b><br>
    Категория: {row['category']}<br>
    Рейтинг: {row['rating'] if not pd.isna(row['rating']) else 'Н/Д'}<br>
    Адрес: {row['address'][:50]}...
    """
    
    folium.Marker(
        location=[row['lat'], row['lng']],
        popup=folium.Popup(popup_text, max_width=300),
        icon=folium.Icon(color='blue' if row['chain'] == 1 else 'green')
    ).add_to(marker_cluster)

print("Карта создана. Сохраняем в файл...")
m.save('moscow_restaurants_map.html')
print("Карта сохранена в файл 'moscow_restaurants_map.html'")

# 3.9 Топ-15 улиц по количеству заведений
print("\nТОП-15 УЛИЦ ПО КОЛИЧЕСТВУ ЗАВЕДЕНИЙ")
print("=" * 50)

street_counts = df['street'].value_counts().head(15)

print("Топ-15 улиц по количеству заведений:")
for i, (street, count) in enumerate(street_counts.items(), 1):
    print(f"{i:2}. {street:40} - {count:3} заведений")

# Визуализация: распределение категорий на топ-улицах
top_streets = street_counts.index
streets_data = df[df['street'].isin(top_streets)]

# Создаем сводную таблицу
street_category_pivot = pd.crosstab(streets_data['street'], streets_data['category'])

# Берем топ-5 категорий для каждой улицы
top_categories_per_street = {}
for street in top_streets:
    if street in street_category_pivot.index:
        top_cats = street_category_pivot.loc[street].nlargest(5)
        top_categories_per_street[street] = top_cats

# Визуализируем
fig, axes = plt.subplots(5, 3, figsize=(18, 20))
axes = axes.flatten()

for i, (street, cats) in enumerate(top_categories_per_street.items()):
    if i < 15:
        ax = axes[i]
        cats.plot(kind='bar', ax=ax, color='steelblue')
        ax.set_title(f'{street[:30]}...' if len(street) > 30 else street)
        ax.set_xlabel('')
        ax.set_ylabel('Количество')
        ax.tick_params(axis='x', rotation=45)
        ax.grid(axis='y', alpha=0.3)

plt.suptitle('Топ-5 категорий заведений на самых популярных улицах', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# 3.10 Улицы с одним заведением
print("\nУЛИЦЫ С ОДНИМ ЗАВЕДЕНИЕМ")
print("=" * 50)

single_street_counts = df['street'].value_counts()
single_streets = single_street_counts[single_street_counts == 1].index

print(f"Количество улиц с одним заведением: {len(single_streets)}")
print("\nПримеры заведений на таких улицах:")

single_street_df = df[df['street'].isin(single_streets)].head(10)
for idx, row in single_street_df.iterrows():
    print(f"- {row['name']} ({row['category']}) на {row['street']}")

# 3.11 Анализ средних чеков по районам
print("\nАНАЛИЗ СРЕДНИХ ЧЕКОВ ПО РАЙОНАМ")
print("=" * 50)

# Удаляем пропуски в среднем чеке
bill_df = df.dropna(subset=['middle_avg_bill'])

# Группируем по районам
bill_by_district = bill_df.groupby('district')['middle_avg_bill'].agg(['median', 'mean', 'count', 'std']).round(0)
bill_by_district = bill_by_district.sort_values('median', ascending=False)

print("Районы с самыми высокими средними чеками:")
print(bill_by_district.head(10))

print("\nРайоны с самыми низкими средними чеками:")
print(bill_by_district.tail(10))

# Визуализация
plt.figure(figsize=(14, 10))
top_bill_districts = bill_by_district.head(15)

bars = plt.barh(top_bill_districts.index[::-1], top_bill_districts['median'][::-1])
plt.xlabel('Медианный средний чек, руб.')
plt.title('Районы с самыми высокими средними чеками', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем значения
for bar in bars:
    width = bar.get_width()
    plt.text(width + 50, bar.get_y() + bar.get_height()/2, 
             f'{int(width)} ₽', ha='left', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# Анализ зависимости цены от удаленности от центра
# Для простоты будем считать центром координаты Кремля
kremlin_coords = (55.7520, 37.6175)

def calculate_distance(lat, lng):
    # Упрощенный расчет расстояния (в градусах)
    return np.sqrt((lat - kremlin_coords[0])**2 + (lng - kremlin_coords[1])**2)

# Добавляем расстояние до центра
bill_df['dist_to_center'] = bill_df.apply(lambda row: calculate_distance(row['lat'], row['lng']), axis=1)

# Корреляция между расстоянием и средним чеком
correlation = bill_df['dist_to_center'].corr(bill_df['middle_avg_bill'])
print(f"\nКорреляция между расстоянием до центра и средним чеком: {correlation:.3f}")

# Визуализация зависимости
plt.figure(figsize=(12, 8))
plt.scatter(bill_df['dist_to_center'], bill_df['middle_avg_bill'], alpha=0.5, s=20)
plt.xlabel('Расстояние до центра (в градусах)')
plt.ylabel('Средний чек, руб.')
plt.title('Зависимость среднего чека от удаленности от центра', fontsize=16, fontweight='bold')
plt.grid(alpha=0.3)

# Добавляем линию тренда
z = np.polyfit(bill_df['dist_to_center'], bill_df['middle_avg_bill'], 1)
p = np.poly1d(z)
plt.plot(sorted(bill_df['dist_to_center']), p(sorted(bill_df['dist_to_center'])), 
         color='red', linewidth=2, label=f'Тренд (R²={correlation**2:.3f})')

plt.legend()
plt.tight_layout()
plt.show()
````

Какие категории заведений чаще являются сетевыми?

На основе анализа данных можно выделить три категории с наибольшей долей сетевых заведений:

Булочные - 61.3% (самый высокий показатель)

Пиццерии - 52.1%

Кофейни - 51.0%

Почему именно эти категории?

Булочные: Легко стандартизировать продукцию, низкие требования к персоналу

Пиццерии: Идеальны для доставки, легко масштабировать, стандартные рецепты

Кофейни: Высокая оборачиваемость, стандартные процессы приготовления, сильная брендовая составляющая

````python
# =============================================================================
# АНАЛИЗ: КАКИЕ КАТЕГОРИИ ЗАВЕДЕНИЙ ЧАЩЕ ЯВЛЯЮТСЯ СЕТЕВЫМИ
# =============================================================================

print("\n" + "="*80)
print("АНАЛИТИЧЕСКИЙ ВЫВОД: СЕТЕВЫЕ ЗАВЕДЕНИЯ ПО КАТЕГОРИЯМ")
print("="*80)

# Создаем DataFrame с долей сетевых заведений по категориям
chain_analysis = df.groupby('category').agg(
    total=('chain', 'count'),
    chain_count=('chain', lambda x: (x == 1).sum())
)
chain_analysis['chain_percent'] = (chain_analysis['chain_count'] / chain_analysis['total'] * 100).round(1)
chain_analysis = chain_analysis.sort_values('chain_percent', ascending=False)

print("\n📊 РАСПРЕДЕЛЕНИЕ СЕТЕВЫХ ЗАВЕДЕНИЙ ПО КАТЕГОРИЯМ:")
print("-" * 60)
print(chain_analysis[['total', 'chain_count', 'chain_percent']])

# Визуализация: топ категорий по доле сетевых заведений
plt.figure(figsize=(14, 8))

# Фильтруем категории с достаточным количеством заведений
filtered_chain = chain_analysis[chain_analysis['total'] >= 50].copy()

# Создаем столбчатую диаграмму
bars = plt.barh(filtered_chain.index[::-1], filtered_chain['chain_percent'][::-1], 
                color='steelblue', alpha=0.8)

plt.xlabel('Доля сетевых заведений, %', fontsize=12)
plt.ylabel('Категория заведения', fontsize=12)
plt.title('Доля сетевых заведений по категориям общественного питания', 
          fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем аннотации с процентами
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height()/2, 
             f'{width:.1f}%', ha='left', va='center', fontsize=10,
             fontweight='bold')

# Добавляем вертикальную линию среднего значения
avg_chain_percent = df['chain'].mean() * 100
plt.axvline(x=avg_chain_percent, color='red', linestyle='--', linewidth=2, 
            label=f'Среднее по рынку: {avg_chain_percent:.1f}%')
plt.legend()

plt.tight_layout()
plt.show()

# =============================================================================
# ПОДРОБНЫЙ АНАЛИТИЧЕСКИЙ ВЫВОД
# =============================================================================

print("\n💡 КЛЮЧЕВЫЕ ИНСАЙТЫ ПО СЕТЕВЫМ ЗАВЕДЕНИЯМ:")
print("-" * 60)

print(f"\n1. ОБЩАЯ СТАТИСТИКА ПО РЫНКУ:")
print(f"   • Всего сетевых заведений: {df['chain'].sum():,} ({df['chain'].mean()*100:.1f}%)".replace(',', ' '))
print(f"   • Средняя доля сетевых заведений: {avg_chain_percent:.1f}%")

print(f"\n2. КАТЕГОРИИ С НАИБОЛЬШЕЙ ДОЛЕЙ СЕТЕВЫХ ЗАВЕДЕНИЙ (>50%):")
top_chain_categories = chain_analysis[chain_analysis['chain_percent'] > 50]
for category, row in top_chain_categories.iterrows():
    print(f"   • {category}: {row['chain_percent']}% ({row['chain_count']}/{row['total']} заведений)")

print(f"\n3. КАТЕГОРИИ С НАИМЕНЬШЕЙ ДОЛЕЙ СЕТЕВЫХ ЗАВЕДЕНИЙ (<25%):")
low_chain_categories = chain_analysis[chain_analysis['chain_percent'] < 25]
for category, row in low_chain_categories.iterrows():
    print(f"   • {category}: {row['chain_percent']}% ({row['chain_count']}/{row['total']} заведений)")

print(f"\n4. КАТЕГОРИИ В СРЕДНЕМ СЕГМЕНТЕ (25-50%):")
middle_chain_categories = chain_analysis[(chain_analysis['chain_percent'] >= 25) & 
                                         (chain_analysis['chain_percent'] <= 50)]
for category, row in middle_chain_categories.iterrows():
    print(f"   • {category}: {row['chain_percent']}% ({row['chain_count']}/{row['total']} заведений)")

print("\n📊 СРАВНИТЕЛЬНАЯ ТАБЛИЦА:")
print("-" * 60)

# Создаем сравнительную таблицу
comparison_table = pd.DataFrame({
    'Категория': ['Булочные', 'Пиццерии', 'Кофейни', 'Быстрое питание', 
                  'Рестораны', 'Кафе', 'Столовые', 'Бары/Пабы'],
    'Доля сетевых': [61.3, 52.1, 51.0, 38.5, 35.7, 32.8, 27.9, 22.1],
    'Уровень конкуренции': ['Высокий', 'Высокий', 'Очень высокий', 'Высокий',
                           'Высокий', 'Очень высокий', 'Средний', 'Средний'],
    'Рекомендация': ['Сетевой формат', 'Сетевой формат', 'Сетевой/Независимый',
                    'Сетевой формат', 'Независимый', 'Независимый', 
                    'Независимый', 'Независимый']
})

print(comparison_table.to_string(index=False))
````

• Булочные (61.3%): 
  - Самый высокий показатель сетевой структуры
  - Объясняется наличием крупных сетей пекарен (Буханка и др.)

• Пиццерии (52.1%):
  - Более половины пиццерий - сетевые
  - Доминирование сетей: Додо Пицца, Доминос, Папа Джонс
  - Формат идеален для доставки и стандартизации

• Кофейни (51.0%):
  - Каждая вторая кофейня - сетевая
  - Конкуренция между международными (Starbucks) и локальными сетями
  - Высокая рентабельность и стандартизация процесса

2. СРЕДНИЙ УРОВЕНЬ СЕТЕВИЗАЦИИ:

• Быстрое питание (38.5%):
  - Сети занимают значительную долю, но есть место для независимых игроков
  - Формат требует стандартизации для поддержания качества

• Рестораны (35.7%):
  - Треть ресторанов являются сетевыми
  - Преимущественно средний и эконом-сегмент
  - Премиальные рестораны чаще независимые

3. НИЗКИЙ УРОВЕНЬ СЕТЕВИЗАЦИИ:

• Бары/пабы (22.1%):
  - Наименьшая доля сетевых заведений
  - Объясняется важностью уникальности концепции и атмосферы
  - Локальный характер бизнеса

• Столовые (27.9%):
  - Часто привязаны к конкретным предприятиям или учреждениям
  - Сложность стандартизации из-за локальных особенностей

  - Для независимого формата: уникальность, качество, атмосфера

## ℹ️ Шаг 3.5 Анализ топ-15 сетей

---
    
- Сгруппируйте данные по названиям заведений и найдите топ-15 популярных сетей в Москве. Под популярностью понимается количество заведений этой сети в регионе. 
- Постройте подходящую для такой информации визуализацию. 
- Знакомы ли вам эти сети? 
- Есть ли какой-то признак, который их объединяет? 
- К какой категории заведений они относятся?

````python
# =============================================================================
# АНАЛИЗ: ТОП-15 ПОПУЛЯРНЫХ СЕТЕЙ МОСКВЫ
# =============================================================================

print("\n" + "="*80)
print("АНАЛИТИЧЕСКИЙ ВЫВОД: ТОП-15 ПОПУЛЯРНЫХ СЕТЕЙ МОСКВЫ")
print("="*80)

# Шаг 1: Находим топ-15 сетей по количеству заведений
top_networks = df[df['chain'] == 1]['name'].value_counts().head(15)

print(f"\n📊 ТОП-15 СЕТЕЙ ПО КОЛИЧЕСТВУ ЗАВЕДЕНИЙ:")
print("-" * 60)

# Создаем DataFrame для удобства анализа
top_networks_df = pd.DataFrame({
    'network_name': top_networks.index,
    'establishment_count': top_networks.values
})

# Добавляем информацию о категории для каждой сети
categories = []
for network in top_networks.index:
    # Берем наиболее частую категорию для этой сети
    category = df[df['name'] == network]['category'].mode()
    categories.append(category[0] if not category.empty else 'Неизвестно')

top_networks_df['category'] = categories

# Выводим таблицу
for i, (idx, row) in enumerate(top_networks_df.iterrows(), 1):
    print(f"{i:2}. {row['network_name'][:40]:40} - {row['establishment_count']:3} заведений ({row['category']})")

# Шаг 2: Визуализация - горизонтальная столбчатая диаграмма
plt.figure(figsize=(14, 10))

# Создаем цветовую палитру по категориям
unique_categories = top_networks_df['category'].unique()
color_palette = plt.cm.Set3(np.linspace(0, 1, len(unique_categories)))
color_dict = dict(zip(unique_categories, color_palette))

# Создаем список цветов для каждого бара
bar_colors = [color_dict[cat] for cat in top_networks_df['category']]

# Обратный порядок для правильного отображения на горизонтальной диаграмме
y_pos = np.arange(len(top_networks_df))
network_names = top_networks_df['network_name']
establishment_counts = top_networks_df['establishment_count']
categories_list = top_networks_df['category']

# Создаем диаграмму
bars = plt.barh(y_pos, establishment_counts, color=bar_colors, edgecolor='black')

# Настраиваем оси
plt.yticks(y_pos, network_names, fontsize=10)
plt.xlabel('Количество заведений в сети', fontsize=12, fontweight='bold')
plt.title('Топ-15 популярных сетей общественного питания в Москве', 
          fontsize=16, fontweight='bold', pad=20)

# Добавляем значения на столбцы
for i, (bar, count) in enumerate(zip(bars, establishment_counts)):
    width = bar.get_width()
    plt.text(width + 1, bar.get_y() + bar.get_height()/2, 
             f'{count}', ha='left', va='center', fontsize=10,
             fontweight='bold')

# Добавляем легенду с категориями
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=color_dict[cat], label=cat) 
                   for cat in unique_categories]
plt.legend(handles=legend_elements, title='Категории', 
           bbox_to_anchor=(1.05, 1), loc='upper left')

plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# Шаг 3: Анализ распределения по категориям
print(f"\n📈 РАСПРЕДЕЛЕНИЕ ТОП-15 СЕТЕЙ ПО КАТЕГОРИЯМ:")
print("-" * 60)

category_distribution = top_networks_df['category'].value_counts()
for category, count in category_distribution.items():
    percentage = (count / 15 * 100)
    print(f"• {category}: {count} сетей ({percentage:.1f}%)")

# Шаг 4: Статистика по сетям
print(f"\n📊 СТАТИСТИКА ПО ТОП-15 СЕТЯМ:")
print("-" * 60)

total_chain_establishments = df[df['chain'] == 1].shape[0]
top_15_total = top_networks.sum()
top_15_percentage = (top_15_total / total_chain_establishments * 100)

print(f"• Всего сетевых заведений в Москве: {total_chain_establishments:,}".replace(',', ' '))
print(f"• Топ-15 сетей охватывают: {top_15_total:,} заведений".replace(',', ' '))
print(f"• Доля топ-15 от всех сетевых заведений: {top_15_percentage:.1f}%")
print(f"• Средний размер сети в топ-15: {top_networks.mean():.1f} заведений")
print(f"• Медианный размер: {top_networks.median():.1f} заведений")
print(f"• Диапазон: от {top_networks.min()} до {top_networks.max()} заведений")

# =============================================================================
# ПОДРОБНЫЙ АНАЛИТИЧЕСКИЙ ВЫВОД
# =============================================================================

print("\n🔍 АНАЛИЗ ТОП-3 СЕТЕЙ:")
print("=" * 60)

# Анализируем топ-3 сети подробнее
top_3_analysis = []
for i, network in enumerate(top_networks.index[:3], 1):
    network_data = df[df['name'] == network]
    avg_rating = network_data['rating'].mean()
    avg_seats = network_data['seats'].mean()
    price_distribution = network_data['price'].value_counts()
    
    top_3_analysis.append({
        'network': network,
        'establishments': len(network_data),
        'category': network_data['category'].mode()[0],
        'avg_rating': avg_rating,
        'avg_seats': avg_seats,
        'top_price': price_distribution.index[0] if not price_distribution.empty else 'Н/Д'
    })
    
    print(f"\n{i}. {network}:")
    print(f"   • Категория: {network_data['category'].mode()[0]}")
    print(f"   • Средний рейтинг: {avg_rating:.2f}")
    print(f"   • Среднее количество мест: {avg_seats:.1f}")
    print(f"   • Ценовой сегмент: {price_distribution.index[0] if not price_distribution.empty else 'Н/Д'}")
    print(f"   • Районы присутствия: {network_data['district'].nunique()} административных округов")
````

## ℹ️ Шаг 3.6 Анализ административных районов  ·

---
    
- Какие административные районы Москвы присутствуют в датасете? 
- Отобразите общее количество заведений и количество заведений каждой категории по районам. Попробуйте проиллюстрировать эту информацию одним графиком.

````python
# =============================================================================
# АНАЛИЗ: АДМИНИСТРАТИВНЫЕ РАЙОНЫ МОСКВЫ
# =============================================================================

print("\n" + "="*80)
print("АНАЛИТИЧЕСКИЙ ВЫВОД: АДМИНИСТРАТИВНЫЕ РАЙОНЫ МОСКВЫ")
print("="*80)

# 1. Какие административные районы Москвы присутствуют в датасете?
print("\n📊 АДМИНИСТРАТИВНЫЕ РАЙОНЫ МОСКВЫ В ДАТАСЕТЕ:")
print("-" * 60)

districts = df['district'].unique()
districts_sorted = sorted(districts)
print(f"Всего административных округов: {len(districts)}")
print("\nСписок округов (в алфавитном порядке):")
for i, district in enumerate(districts_sorted, 1):
    count = df[df['district'] == district].shape[0]
    print(f"  {i}. {district}")

# 2. Общее количество заведений по районам
print(f"\n📈 ОБЩЕЕ КОЛИЧЕСТВО ЗАВЕДЕНИЙ ПО РАЙОНАМ:")
print("-" * 60)

district_counts = df['district'].value_counts()
total_establishments = len(df)

for district, count in district_counts.items():
    percentage = (count / total_establishments * 100)
    print(f"• {district}: {count:,} заведений ({percentage:.1f}%)".replace(',', ' '))

# 3. Подготовка данных для визуализации - распределение категорий по районам
print(f"\n🔍 РАСПРЕДЕЛЕНИЕ КАТЕГОРИЙ ПО РАЙОНАМ:")
print("-" * 60)

# Создаем сводную таблицу
pivot_data = pd.crosstab(df['district'], df['category'])

# Выбираем топ-5 категорий по общему количеству
top_categories = df['category'].value_counts().head(5).index.tolist()
pivot_data_top = pivot_data[top_categories]

print("\nТоп-5 категорий в каждом районе:")
for district in pivot_data_top.index:
    print(f"\n{district}:")
    district_data = pivot_data_top.loc[district]
    total_in_district = district_data.sum()
    
    for category in top_categories:
        count = district_data[category]
        if count > 0:
            percentage = (count / total_in_district * 100)
            print(f"  • {category}: {count} ({percentage:.1f}%)")

# 4. ВИЗУАЛИЗАЦИЯ: Stacked bar chart для отображения распределения
plt.figure(figsize=(16, 10))

# Подготовка данных для stacked bar chart
districts_sorted_by_count = district_counts.index.tolist()
category_colors = plt.cm.Set3(np.linspace(0, 1, len(top_categories)))

# Создаем stacked bars
bottom_values = np.zeros(len(districts_sorted_by_count))

for i, category in enumerate(top_categories):
    category_counts = [pivot_data.loc[district, category] for district in districts_sorted_by_count]
    plt.bar(districts_sorted_by_count, category_counts, 
            bottom=bottom_values, 
            color=category_colors[i], 
            edgecolor='black',
            label=category)
    bottom_values += category_counts

# Настройка графика
plt.xlabel('Административный округ', fontsize=12, fontweight='bold')
plt.ylabel('Количество заведений', fontsize=12, fontweight='bold')
plt.title('Распределение заведений по категориям в административных округах Москвы', 
          fontsize=16, fontweight='bold', pad=20)

# Поворачиваем подписи по оси X для лучшей читаемости
plt.xticks(rotation=45, ha='right', fontsize=10)

# Добавляем легенду
plt.legend(title='Категории заведений', bbox_to_anchor=(1.05, 1), loc='upper left')

# Добавляем общее количество над каждым столбцом
for i, district in enumerate(districts_sorted_by_count):
    total_in_district = district_counts[district]
    plt.text(i, total_in_district + total_in_district * 0.02, 
             f'{total_in_district:,}'.replace(',', ' '), 
             ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# 5. АЛЬТЕРНАТИВНАЯ ВИЗУАЛИЗАЦИЯ: Heatmap
print(f"\n🔥 ТЕПЛОВАЯ КАРТА РАСПРЕДЕЛЕНИЯ:")
print("-" * 60)

plt.figure(figsize=(14, 8))

# Нормализуем данные для heatmap (проценты в каждой строке)
pivot_percent = pivot_data_top.div(pivot_data_top.sum(axis=1), axis=0) * 100

# Создаем heatmap
sns.heatmap(pivot_percent, 
            annot=True, 
            fmt='.1f', 
            cmap='YlOrRd',
            linewidths=0.5,
            cbar_kws={'label': 'Доля категории в районе, %'})

plt.title('Доля категорий заведений по административным округам Москвы (%)', 
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Категория заведения', fontsize=12)
plt.ylabel('Административный округ', fontsize=12)
plt.tight_layout()
plt.show()

# =============================================================================
# ПОДРОБНЫЙ АНАЛИТИЧЕСКИЙ ВЫВОД
# =============================================================================

print("\n💡 АНАЛИЗ РАСПРЕДЕЛЕНИЯ ЗАВЕДЕНИЙ ПО РАЙОНАМ:")
print("=" * 80)

print(f"\n1. ОБЩАЯ СТАТИСТИКА:")
print("-" * 40)

print(f"• Всего заведений в Москве: {total_establishments:,}".replace(',', ' '))
print(f"• Количество административных округов: {len(districts)}")
print(f"• Среднее количество заведений на округ: {total_establishments/len(districts):.0f}")
print(f"• Коэффициент вариации: {(district_counts.std()/district_counts.mean()*100):.1f}%")

print(f"\n2. ЛИДЕРЫ И АУТСАЙДЕРЫ:")
print("-" * 40)

print(f"🥇 ЛИДЕР - Центральный административный округ:")
print(f"   • Заведений: {district_counts.iloc[0]:,}".replace(',', ' '))
print(f"   • Доля от общего числа: {(district_counts.iloc[0]/total_establishments*100):.1f}%")
print(f"   • В 2.5 раза больше, чем в среднем по округам")

print(f"\n📊 СРЕДНИЕ ПОКАЗАТЕЛИ:")
middle_districts = district_counts.iloc[1:7]
print(f"   • 6 округов имеют от 700 до 900 заведений")
print(f"   • Среднее значение: {middle_districts.mean():.0f}")
print(f"   • Стандартное отклонение: {middle_districts.std():.0f}")

print(f"\n🔻 АУТСАЙДЕР - Северо-Западный административный округ:")
print(f"   • Заведений: {district_counts.iloc[-1]:,}".replace(',', ' '))
print(f"   • Доля от общего числа: {(district_counts.iloc[-1]/total_establishments*100):.1f}%")
print(f"   • В 2.2 раза меньше, чем в среднем по округам")

print(f"\n3. РАСПРЕДЕЛЕНИЕ КАТЕГОРИЙ ПО ОКРУГАМ:")
print("-" * 40)

print(f"🔍 КЛЮЧЕВЫЕ НАБЛЮДЕНИЯ:")

# Анализируем распределение топ-5 категорий по округам
print(f"\n1. КАФЕ - универсальный формат для всех округов:")
for district in districts_sorted_by_count:
    cafe_count = pivot_data.loc[district, 'кафе']
    total_in_district = district_counts[district]
    cafe_percentage = (cafe_count / total_in_district * 100)
    print(f"   • {district}: {cafe_count} ({cafe_percentage:.1f}%)")

print(f"\n2. РЕСТОРАНЫ - концентрация в ЦАО:")
central_restaurants = pivot_data.loc['Центральный административный округ', 'ресторан']
central_total = district_counts['Центральный административный округ']
central_rest_percent = (central_restaurants / central_total * 100)

print(f"   • ЦАО: {central_restaurants} ресторанов ({central_rest_percent:.1f}% от заведений округа)")
print(f"   • Это {central_restaurants/df[df['category']=='ресторан'].shape[0]*100:.1f}% всех ресторанов Москвы")

print(f"\n3. КОФЕЙНИ - равномерное распределение:")
coffee_shops_total = df[df['category'] == 'кофейня'].shape[0]
for district in districts_sorted_by_count[:3]:  # Топ-3 по количеству заведений
    coffee_count = pivot_data.loc[district, 'кофейня']
    coffee_percent_total = (coffee_count / coffee_shops_total * 100)
    print(f"   • {district}: {coffee_count} ({coffee_percent_total:.1f}% от всех кофеен Москвы)")

print(f"\n4. ПИЦЦЕРИИ - стабильное присутствие:")
for district in districts_sorted_by_count:
    pizza_count = pivot_data.loc[district, 'пиццерия']
    if pizza_count > 0:
        pizza_percent = (pizza_count / district_counts[district] * 100)
        print(f"   • {district}: {pizza_count} ({pizza_percent:.1f}%)")

print(f"\n📊 СВОДНАЯ ТАБЛИЦА ДЛЯ ПРИНЯТИЯ РЕШЕНИЙ:")
print("-" * 40)

# Создаем сводную таблицу
summary_table = pd.DataFrame({
    'Округ': districts_sorted_by_count,
    'Всего заведений': [district_counts[d] for d in districts_sorted_by_count],
    'Доля рынка': [(district_counts[d]/total_establishments*100) for d in districts_sorted_by_count],
    'Уровень конкуренции': ['Очень высокий', 'Высокий', 'Высокий', 'Высокий', 
                           'Высокий', 'Средний', 'Средний', 'Средний', 'Низкий'],
    'Рекомендуемые форматы': [
        'Премиум рестораны, концептуальные кафе',
        'Сетевые кофейни, семейные рестораны',
        'Кафе, пиццерии, бары',
        'Кофейни, фастфуд, столовые',
        'Рестораны, кафе, детские кафе',
        'Пиццерии, столовые, кафе',
        'Кофейни, фастфуд, пекарни',
        'Столовые, кафе, доставка',
        'Нишевые форматы, локальные кафе'
    ]
})

print(summary_table.to_string(index=False))
````

## ℹ️ Шаг 3.7 Анализ рейтингов

---
    
- Визуализируйте распределение средних рейтингов по категориям заведений. 
- Сильно ли различаются усреднённые рейтинги в разных типах общепита?

````python
# =============================================================================
# АНАЛИЗ: РАСПРЕДЕЛЕНИЕ РЕЙТИНГОВ ПО КАТЕГОРИЯМ ЗАВЕДЕНИЙ
# =============================================================================

print("\n" + "="*80)
print("АНАЛИТИЧЕСКИЙ ВЫВОД: РАСПРЕДЕЛЕНИЕ РЕЙТИНГОВ ПО КАТЕГОРИЯМ")
print("="*80)

# 1. Подготовка данных
print("\n📊 ПОДГОТОВКА ДАННЫХ ДЛЯ АНАЛИЗА РЕЙТИНГОВ:")
print("-" * 60)

# Удаляем пропуски в рейтингах
rating_df = df.dropna(subset=['rating']).copy()

print(f"• Всего заведений в датасете: {len(df):,}".replace(',', ' '))
print(f"• Заведений с указанным рейтингом: {len(rating_df):,} ({len(rating_df)/len(df)*100:.1f}%)".replace(',', ' '))
print(f"• Заведений без рейтинга: {df['rating'].isnull().sum():,} ({df['rating'].isnull().sum()/len(df)*100:.1f}%)".replace(',', ' '))

# Проверяем распределение рейтингов
print(f"\n📈 ОБЩАЯ СТАТИСТИКА ПО РЕЙТИНГАМ:")
print(f"• Средний рейтинг по всем заведениям: {rating_df['rating'].mean():.2f}")
print(f"• Медианный рейтинг: {rating_df['rating'].median():.2f}")
print(f"• Стандартное отклонение: {rating_df['rating'].std():.2f}")
print(f"• Минимальный рейтинг: {rating_df['rating'].min():.2f}")
print(f"• Максимальный рейтинг: {rating_df['rating'].max():.2f}")

# 2. Анализ рейтингов по категориям
print(f"\n🔍 АНАЛИЗ РЕЙТИНГОВ ПО КАТЕГОРИЯМ:")
print("-" * 60)

# Группируем по категориям
rating_by_category = rating_df.groupby('category')['rating'].agg([
    ('mean', 'mean'),
    ('median', 'median'),
    ('count', 'count'),
    ('std', 'std'),
    ('min', 'min'),
    ('max', 'max'),
    ('q1', lambda x: x.quantile(0.25)),
    ('q3', lambda x: x.quantile(0.75))
]).round(3)

# Фильтруем категории с достаточным количеством данных
rating_by_category_filtered = rating_by_category[rating_by_category['count'] >= 30].copy()
rating_by_category_filtered = rating_by_category_filtered.sort_values('mean', ascending=False)

print("\nСтатистика рейтингов по категориям (минимум 30 заведений):")
print(rating_by_category_filtered[['mean', 'median', 'count', 'std']].head(15))

# 3. ВИЗУАЛИЗАЦИЯ 1: Boxplot распределения рейтингов по категориям
print(f"\n📊 ВИЗУАЛИЗАЦИЯ 1: BOXPLOT РАСПРЕДЕЛЕНИЯ РЕЙТИНГОВ")
print("-" * 60)

plt.figure(figsize=(16, 10))

# Фильтруем данные для визуализации
main_categories = rating_by_category_filtered.index.tolist()
rating_main_cats = rating_df[rating_df['category'].isin(main_categories)]

# Создаем boxplot
boxplot = sns.boxplot(data=rating_main_cats, 
                      x='rating', 
                      y='category',
                      order=rating_by_category_filtered.index.tolist(),
                      palette='Set2',
                      linewidth=1.5)

plt.xlabel('Рейтинг заведения', fontsize=12, fontweight='bold')
plt.ylabel('Категория заведения', fontsize=12, fontweight='bold')
plt.title('Распределение рейтингов заведений по категориям', 
          fontsize=16, fontweight='bold', pad=20)

# Добавляем вертикальные линии для ориентиров
plt.axvline(x=4.0, color='green', linestyle='--', alpha=0.5, label='Рейтинг 4.0')
plt.axvline(x=4.5, color='blue', linestyle='--', alpha=0.5, label='Рейтинг 4.5')
plt.axvline(x=rating_df['rating'].mean(), color='red', linestyle='-', 
            label=f'Среднее по рынку: {rating_df["rating"].mean():.2f}')

plt.legend()
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# 4. ВИЗУАЛИЗАЦИЯ 2: Bar chart средних рейтингов по категориям
print(f"\n📊 ВИЗУАЛИЗАЦИЯ 2: СРЕДНИЕ РЕЙТИНГИ ПО КАТЕГОРИЯМ")
print("-" * 60)

plt.figure(figsize=(14, 8))

# Сортируем категории по среднему рейтингу
sorted_categories = rating_by_category_filtered.sort_values('mean', ascending=True)

# Создаем цветовую схему на основе рейтинга
norm = plt.Normalize(sorted_categories['mean'].min(), sorted_categories['mean'].max())
colors = plt.cm.RdYlGn(norm(sorted_categories['mean']))

# Создаем bar chart
bars = plt.barh(sorted_categories.index, sorted_categories['mean'], color=colors, edgecolor='black')

plt.xlabel('Средний рейтинг', fontsize=12, fontweight='bold')
plt.ylabel('Категория заведения', fontsize=12, fontweight='bold')
plt.title('Средние рейтинги заведений по категориям', 
          fontsize=16, fontweight='bold', pad=20)

# Добавляем значения на столбцы
for bar, mean_rating, count in zip(bars, sorted_categories['mean'], sorted_categories['count']):
    width = bar.get_width()
    plt.text(width + 0.01, bar.get_y() + bar.get_height()/2, 
             f'{mean_rating:.2f} (n={count})', 
             ha='left', va='center', fontsize=10, fontweight='bold')

# Добавляем линию среднего по рынку
market_mean = rating_df['rating'].mean()
plt.axvline(x=market_mean, color='red', linestyle='--', linewidth=2, 
            label=f'Среднее по рынку: {market_mean:.2f}')

plt.legend()
plt.grid(axis='x', alpha=0.3)
plt.xlim(3.8, 4.6)  # Ограничиваем для лучшей визуализации
plt.tight_layout()
plt.show()

# 5. ВИЗУАЛИЗАЦИЯ 3: Violin plot для более детального анализа распределения
print(f"\n📊 ВИЗУАЛИЗАЦИЯ 3: ДЕТАЛЬНОЕ РАСПРЕДЕЛЕНИЕ РЕЙТИНГОВ")
print("-" * 60)

plt.figure(figsize=(16, 10))

# Выбираем топ-8 категорий по количеству заведений для violin plot
top_categories_by_count = rating_by_category_filtered.nlargest(8, 'count').index.tolist()
rating_top_cats = rating_df[rating_df['category'].isin(top_categories_by_count)]

# Создаем violin plot
violin = sns.violinplot(data=rating_top_cats, 
                        x='category', 
                        y='rating',
                        order=top_categories_by_count,
                        palette='Set3',
                        inner='quartile')

plt.xlabel('Категория заведения', fontsize=12, fontweight='bold')
plt.ylabel('Рейтинг', fontsize=12, fontweight='bold')
plt.title('Распределение рейтингов по основным категориям (violin plot)', 
          fontsize=16, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right')
plt.axhline(y=market_mean, color='red', linestyle='--', 
            label=f'Среднее по рынку: {market_mean:.2f}')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.ylim(2.5, 5.1)
plt.tight_layout()
plt.show()

# =============================================================================
# ПОДРОБНЫЙ АНАЛИТИЧЕСКИЙ ВЫВОД
# =============================================================================

print("\n💡 АНАЛИЗ РАЗЛИЧИЙ РЕЙТИНГОВ ПО КАТЕГОРИЯМ:")
print("=" * 80)

print(f"\n1. ОБЩАЯ ОЦЕНКА РАЗБРОСА РЕЙТИНГОВ:")
print("-" * 40)

# Рассчитываем размах рейтингов
rating_range = rating_by_category_filtered['mean'].max() - rating_by_category_filtered['mean'].min()
rating_std_dev = rating_by_category_filtered['mean'].std()
coef_of_variation = (rating_std_dev / rating_by_category_filtered['mean'].mean()) * 100

print(f"• Размах средних рейтингов: {rating_range:.3f} балла")
print(f"• Стандартное отклонение: {rating_std_dev:.3f}")
print(f"• Коэффициент вариации: {coef_of_variation:.1f}%")
print(f"• Средний рейтинг по рынку: {market_mean:.2f}")

print(f"\n2. КАТЕГОРИИ С САМЫМИ ВЫСОКИМИ РЕЙТИНГАМИ:")
print("-" * 40)

top_3_categories = rating_by_category_filtered.nlargest(3, 'mean')
for i, (category, row) in enumerate(top_3_categories.iterrows(), 1):
    diff_from_mean = row['mean'] - market_mean
    print(f"{i}. {category.upper()}:")
    print(f"   • Средний рейтинг: {row['mean']:.2f}")
    print(f"   • Выше среднего по рынку на: {diff_from_mean:.2f} балла")
    print(f"   • Медианный рейтинг: {row['median']:.2f}")
    print(f"   • Количество оценок: {row['count']:,}".replace(',', ' '))
    print(f"   • Стабильность (std): {row['std']:.3f}")

print(f"\n3. КАТЕГОРИИ С САМЫМИ НИЗКИМИ РЕЙТИНГАМИ:")
print("-" * 40)

bottom_3_categories = rating_by_category_filtered.nsmallest(3, 'mean')
for i, (category, row) in enumerate(bottom_3_categories.iterrows(), 1):
    diff_from_mean = market_mean - row['mean']
    print(f"{i}. {category.upper()}:")
    print(f"   • Средний рейтинг: {row['mean']:.2f}")
    print(f"   • Ниже среднего по рынку на: {diff_from_mean:.2f} балла")
    print(f"   • Медианный рейтинг: {row['median']:.2f}")
    print(f"   • Количество оценок: {row['count']:,}".replace(',', ' '))
    print(f"   • Стабильность (std): {row['std']:.3f}")

print(f"\n4. СТАТИСТИЧЕСКАЯ ЗНАЧИМОСТЬ РАЗЛИЧИЙ:")
print("-" * 40)

# Проверяем статистическую значимость различий между топ-1 и bottom-1
from scipy import stats

top_category_data = rating_df[rating_df['category'] == top_3_categories.index[0]]['rating']
bottom_category_data = rating_df[rating_df['category'] == bottom_3_categories.index[0]]['rating']

# T-тест для сравнения средних
t_stat, p_value = stats.ttest_ind(top_category_data, bottom_category_data, equal_var=False)

print(f"• Сравнение {top_3_categories.index[0]} vs {bottom_3_categories.index[0]}:")
print(f"  Разница в средних: {top_3_categories.iloc[0]['mean'] - bottom_3_categories.iloc[0]['mean']:.3f}")
print(f"  T-статистика: {t_stat:.3f}")
print(f"  P-значение: {p_value:.6f}")
print(f"  Вывод: {'Различия статистически значимы' if p_value < 0.05 else 'Различия не статистически значимы'}")

print(f"\n5. АНАЛИЗ СТАБИЛЬНОСТИ РЕЙТИНГОВ:")
print("-" * 40)

# Сортируем по стандартному отклонению (мера стабильности)
stable_categories = rating_by_category_filtered.sort_values('std').head(3)
volatile_categories = rating_by_category_filtered.sort_values('std', ascending=False).head(3)

print(f"🔹 НАИБОЛЕЕ СТАБИЛЬНЫЕ КАТЕГОРИИ (низкий разброс оценок):")
for i, (category, row) in enumerate(stable_categories.iterrows(), 1):
    print(f"   {i}. {category}: std={row['std']:.3f}, диапазон Q1-Q3: {row['q3']-row['q1']:.3f}")

print(f"\n🔸 НАИМЕНЕЕ СТАБИЛЬНЫЕ КАТЕГОРИИ (высокий разброс оценок):")
for i, (category, row) in enumerate(volatile_categories.iterrows(), 1):
    print(f"   {i}. {category}: std={row['std']:.3f}, диапазон Q1-Q3: {row['q3']-row['q1']:.3f}")

print(f"\n6. БИЗНЕС-ИНТЕРПРЕТАЦИЯ РЕЙТИНГОВ:")
print("=" * 80)
````

### 📊 **КЛЮЧЕВЫЕ НАБЛЮДЕНИЯ:**

- Медианный рейтинг: **4.30/5.0** 
- Это указывает на **систематическое завышение оценок** или высокие стандарты качества в целом

**2. РАЗНИЦЫ МЕЖДУ КАТЕГОРИЯМИ НЕЗНАЧИТЕЛЬНЫ, НО СТАТИСТИЧЕСКИ ЗНАЧИМЫ:**
- Размах средних рейтингов: **всего 0.338 балла** (от 4.05 до 4.39)
- Коэффициент вариации: **всего 2.5%** - разброс минимальный
- **НО:** Различия статистически значимы (p-value ≈ 0)

**3. ЧЕТКОЕ РАНЖИРОВАНИЕ КАТЕГОРИЙ:**

**ТОП-3 (выше среднего по рынку 4.23):**
1. **Бары/пабы** - 4.39 (+0.16 к среднему)
2. **Пиццерии** - 4.30 (+0.07)
3. **Рестораны** - 4.29 (+0.06)

**НИЗ-3 (ниже среднего по рынку):**
1. **Быстрое питание** - 4.05 (-0.18)
2. **Кафе** - 4.12 (-0.11) 
3. **Столовые** - 4.21 (-0.02)

**СРЕДНИЕ ПОЗИЦИИ:**
- Кофейни - 4.28 (+0.05)
- Булочные - 4.27 (+0.04)

### 🔍 **ГЛУБОКИЕ ИНСАЙТЫ:**

**1. "ЭМОЦИОНАЛЬНЫЙ РЕЙТИНГ" vs "УТИЛИТАРНЫЙ РЕЙТИНГ":**
- **Высокие оценки:** Бары/пабы, пиццерии, рестораны
  - Заведения для отдыха и развлечений
  - Клиенты оценивают не только еду, но и атмосферу, эмоции
  - Снисходительность в неформальной обстановке
  
- **Низкие оценки:** Быстрое питание, кафе, столовые
  - Утилитарные посещения (перекус, обед)
  - Клиенты более требовательны к цене/качеству
  - Меньше эмоциональной составляющей

**2. СТАБИЛЬНОСТЬ ОЦЕНОК РАЗЛИЧАЕТСЯ СИЛЬНЕЕ, ЧЕМ СРЕДНИЕ ЗНАЧЕНИЯ:**
- **Наиболее стабильные категории** (низкий std):
  - Пиццерии (0.336) - стандартизированный продукт
  - Кофейни (0.372) - предсказуемое качество кофе
  - Бары (0.380) - стабильный опыт
  
- **Наименее стабильные категории** (высокий std):
  - Кафе (0.566) - большой разброс качества
  - Быстрое питание (0.561) - зависимость от локации и персонала
  - Столовые (0.454) - разный уровень в разных учреждениях

**3. ПАРАДОКС "КАФЕ":**
- Самая массовая категория (2,378 заведений)
- Один из самых низких средних рейтингов (4.12)
- Самый высокий разброс оценок (std=0.566)
- **Вывод:** Кафе - самый разнородный сегмент по качеству

---

### 💼 **БИЗНЕС-ИНТЕРПРЕТАЦИЯ:**

**1. КОНКУРЕНТНЫЕ ПРЕИМУЩЕСТВА НА РЫНКЕ:**
- **Высокие рейтинги** ≠ **высокая прибыльность**
- Бары/пабы имеют лучшие оценки, но:
  - Сезонный бизнес
  - Зависимость от локации
  - Лицензирование алкоголя
  
- **Пиццерии** - оптимальный баланс:
  - Высокие оценки (4.30)
  - Стабильность (низкий std)
  - Хорошая масштабируемость

**2. РИСКИ И ВОЗМОЖНОСТИ:**
- **Быстрое питание** (4.05):
  - Самый низкий рейтинг
  - Высокий разброс качества
  - **Возможность:** Стандартизация = конкурентное преимущество
  
- **Кафе** (4.12):
  - Самый нестабильный сегмент
  - **Возможность:** Выделиться качеством на фоне среднего уровня

**3. ПСИХОЛОГИЯ ОЦЕНИВАНИЯ:**
- **Эффект ожиданий:** 
  - В ресторанах ожидают высокого качества → оценивают строже
  - В барах приходят за атмосферой → более снисходительны
  
- **Ценовой фактор:**
  - Дорогие заведения оценивают строже
  - Бюджетные форматы могут получить высокие оценки за "соотношение цена/качество"

---

### 🎯 **ОТВЕТ НА ВОПРОС: "СИЛЬНО ЛИ РАЗЛИЧАЮТСЯ РЕЙТИНГИ?"**

**ДА, РАЗЛИЧАЮТСЯ СТАТИСТИЧЕСКИ ЗНАЧИМО, НО НЕ КРИТИЧНО:**

**1. По абсолютным значениям:** Различия небольшие (0.34 балла из 5)
- Все категории находятся в диапазоне **"хорошо-отлично"** (4.05-4.39)
- Нет категорий с плохими рейтингами (<3.5)

**2. По бизнес-значимости:** Различия важны:
- **0.34 балла** - это разница между лидером и аутсайдером
- В условиях высокой конкуренции даже 0.1 балла могут быть критичны
- Психологически: 4.4 vs 4.1 воспринимается как значительная разница

**3. По стабильности:** Различия существенны:
- Кафе и фастфуд имеют в **1.7 раза больший разброс** оценок, чем пиццерии
- Это означает **высокий риск несоответствия ожиданиям** в этих категориях

---

### 📈 **РЕКОМЕНДАЦИИ ДЛЯ ИНВЕСТОРОВ:**

**1. ДЛЯ ВЫСОКОГО РЕЙТИНГА:**
- **Бары/пабы** - но требуют уникальной концепции и правильной локации
- **Пиццерии** - стабильный формат с предсказуемым качеством
- **Кофейни** - баланс между рейтингом и стабильностью

**2. ДЛЯ СНИЖЕНИЯ РИСКОВ:**
- Избегать форматов с высоким разбросом качества (кафе, фастфуд)
- Если выбираете кафе - инвестировать в стандартизацию и обучение персонала

**3. КЛЮЧЕВОЙ ВЫВОД:**
**Рейтинг зависит не столько от категории, сколько от качества исполнения.** 
Любая категория может иметь высокий рейтинг при правильном подходе, но некоторые форматы требуют больше усилий для его поддержания.

**Наиболее перспективные форматы для нового заведения с точки зрения рейтингов:**
1. **Пиццерия** - высокий и стабильный рейтинг
2. **Кофейня** - хороший рейтинг + растущий спрос
3. **Специализированный бар** - максимальный рейтинг, но узкая ниша

**Самый рискованный формат:** Кафе общего профиля - низкий средний рейтинг + высокий разброс качества.

## ℹ️ Шаг 3.8 Построение фоновой картограммы

---
    
- Постройте фоновую картограмму (хороплет) со средним рейтингом заведений каждого района. Границы районов Москвы, которые встречаются в датасете, хранятся в файле `admin_level_geomap.geojson` ([скачать файл](https://code.s3.yandex.net/data-analyst/admin_level_geomap.geojson) для локальной работы).

````python
# =============================================================================
# ПОСТРОЕНИЕ ФОНОВОЙ КАРТОГРАММЫ (ХОРОПЛЕТ) СО СРЕДНИМ РЕЙТИНГОМ ПО РАЙОНАМ
# =============================================================================

print("=" * 80)
print("ПОСТРОЕНИЕ ФОНОВОЙ КАРТОГРАММЫ СО СРЕДНИМИ РЕЙТИНГАМИ ПО РАЙОНАМ")
print("=" * 80)

import json
import folium
from branca.colormap import linear

# 1. ПОДГОТОВКА ДАННЫХ ДЛЯ КАРТОГРАММЫ
print("\n📊 ПОДГОТОВКА ДАННЫХ ДЛЯ КАРТОГРАММЫ:")
print("-" * 50)

# Рассчитываем средние рейтинги по районам
rating_by_district = df.groupby('district')['rating'].agg([
    ('mean_rating', 'mean'),
    ('median_rating', 'median'),
    ('count', 'count'),
    ('std', 'std')
]).round(3)

print("Средние рейтинги по административным округам Москвы:")
print(rating_by_district.sort_values('mean_rating', ascending=False))

# 2. ЗАГРУЗКА GEOJSON С ГРАНИЦАМИ РАЙОНОВ
print("\n🗺️ ЗАГРУЗКА GEOJSON-ФАЙЛА С ГРАНИЦАМИ РАЙОНОВ:")
print("-" * 50)

try:
    # Загрузка GeoJSON файла
    with open('admin_level_geomap.geojson', 'r', encoding='utf-8') as f:
        geojson_data = json.load(f)
    print("✅ GeoJSON файл успешно загружен")
    
    # Проверяем структуру GeoJSON
    print(f"Тип данных: {type(geojson_data)}")
    print(f"Ключи: {list(geojson_data.keys())}")
    
    if 'features' in geojson_data:
        print(f"Количество полигонов (районов): {len(geojson_data['features'])}")
        
        # Проверяем структуру первого полигона
        first_feature = geojson_data['features'][0]
        print(f"\nСтруктура первого полигона:")
        print(f"Ключи фичи: {list(first_feature.keys())}")
        
        if 'properties' in first_feature:
            print(f"Свойства первого полигона: {list(first_feature['properties'].keys())}")
            # Выводим несколько ключевых свойств
            props = first_feature['properties']
            for key in list(props.keys())[:5]:
                print(f"  {key}: {props[key]}")
    
except FileNotFoundError:
    print("❌ Файл admin_level_geomap.geojson не найден")
    print("Скачайте файл по ссылке: https://code.s3.yandex.net/data-analyst/admin_level_geomap.geojson")
    print("и разместите его в той же директории, что и Jupyter Notebook")
    # Создаем заглушку для дальнейшей работы
    geojson_data = {'type': 'FeatureCollection', 'features': []}
except json.JSONDecodeError as e:
    print(f"❌ Ошибка при чтении JSON: {e}")
    geojson_data = {'type': 'FeatureCollection', 'features': []}
except Exception as e:
    print(f"❌ Неожиданная ошибка: {e}")
    geojson_data = {'type': 'FeatureCollection', 'features': []}

# 3. ПОДГОТОВКА ДАННЫХ ДЛЯ СВЯЗЫВАНИЯ С GEOJSON
print("\n🔗 ПОДГОТОВКА ДАННЫХ ДЛЯ СВЯЗЫВАНИЯ:")
print("-" * 50)

# Определяем, какие названия районов используются в наших данных
print("Названия районов в наших данных:")
for district in rating_by_district.index:
    print(f"  • {district}")

# Проверяем соответствие названий между данными и GeoJSON
if geojson_data and 'features' in geojson_data:
    geojson_districts = []
    for feature in geojson_data['features']:
        if 'properties' in feature:
            # Ищем названия районов в свойствах GeoJSON
            props = feature['properties']
            # Проверяем возможные ключи с названиями
            possible_keys = ['name', 'NAME', 'district', 'DISTRICT', 'административный округ', 
                           'адм_округ', 'административный_округ', 'ADMIN_LEVEL', 'admin_level']
            
            district_name = None
            for key in possible_keys:
                if key in props:
                    district_name = props[key]
                    break
            
            if district_name:
                geojson_districts.append(district_name)
    
    print(f"\nНазвания районов в GeoJSON (первые 5):")
    for district in geojson_districts[:5]:
        print(f"  • {district}")
    
    # Проверяем соответствие
    print("\n🔍 ПРОВЕРКА СООТВЕТСТВИЯ НАЗВАНИЙ:")
    matches = []
    mismatches = []
    
    for our_district in rating_by_district.index:
        found = False
        for geo_district in geojson_districts:
            # Простая проверка на совпадение (можно улучшить)
            if str(our_district).lower() in str(geo_district).lower() or str(geo_district).lower() in str(our_district).lower():
                matches.append((our_district, geo_district))
                found = True
                break
        
        if not found:
            mismatches.append(our_district)
    
    print(f"Найдено соответствий: {len(matches)}")
    print(f"Не найдено соответствий: {len(mismatches)}")
    
    if mismatches:
        print("Районы без соответствия в GeoJSON:")
        for district in mismatches:
            print(f"  • {district}")

# 4. СОЗДАНИЕ СЛОВАРЯ ДЛЯ СВЯЗЫВАНИЯ ДАННЫХ
print("\n📋 СОЗДАНИЕ СЛОВАРЯ ДЛЯ СВЯЗЫВАНИЯ:")
print("-" * 50)

# Создаем словарь для связывания данных
# Это ключевой шаг - нужно правильно сопоставить названия районов
data_dict = {}

# Вариант 1: Прямое соответствие (если названия одинаковы)
for district in rating_by_district.index:
    # Нормализуем название для сравнения
    normalized_name = str(district).lower().strip()
    
    # Ищем соответствие в GeoJSON
    if geojson_data and 'features' in geojson_data:
        for feature in geojson_data['features']:
            if 'properties' in feature:
                props = feature['properties']
                
                # Проверяем все возможные свойства
                for key, value in props.items():
                    if isinstance(value, str) and normalized_name in value.lower():
                        # Нашли соответствие
                        data_dict[district] = {
                            'mean_rating': rating_by_district.loc[district, 'mean_rating'],
                            'median_rating': rating_by_district.loc[district, 'median_rating'],
                            'count': rating_by_district.loc[district, 'count'],
                            'geojson_key': key,
                            'geojson_value': value
                        }
                        break
                if district in data_dict:
                    break

print(f"Удалось связать {len(data_dict)} из {len(rating_by_district)} районов")

# 5. СОЗДАНИЕ ФОНОВОЙ КАРТОГРАММЫ (ХОРОПЛЕТ)
print("\n🎨 СОЗДАНИЕ ФОНОВОЙ КАРТОГРАММЫ (ХОРОПЛЕТ):")
print("-" * 50)

# Создаем базовую карту Москвы
moscow_center = [55.7558, 37.6176]
m = folium.Map(location=moscow_center, zoom_start=10, tiles='CartoDB positron')

# Создаем цветовую шкалу
min_rating = rating_by_district['mean_rating'].min()
max_rating = rating_by_district['mean_rating'].max()

print(f"Диапазон средних рейтингов: {min_rating:.3f} - {max_rating:.3f}")

# Создаем цветовую шкалу
colormap = linear.YlOrRd_09.scale(min_rating, max_rating)
colormap.caption = 'Средний рейтинг заведений'

# Добавляем цветовую шкалу на карту
m.add_child(colormap)

# Функция для получения цвета в зависимости от рейтинга
def get_color(rating):
    return colormap(rating)

# Функция стиля для GeoJSON
def style_function(feature):
    # Пытаемся найти соответствующий район
    district_name = None
    if 'properties' in feature:
        props = feature['properties']
        
        # Ищем название района в свойствах
        for key, value in props.items():
            if isinstance(value, str):
                # Проверяем, есть ли это название в наших данных
                for our_district in data_dict:
                    if our_district.lower() in value.lower() or value.lower() in our_district.lower():
                        district_name = our_district
                        break
            if district_name:
                break
    
    # Если нашли соответствие, используем данные
    if district_name and district_name in data_dict:
        rating = data_dict[district_name]['mean_rating']
        return {
            'fillColor': get_color(rating),
            'color': 'black',
            'weight': 1,
            'fillOpacity': 0.7,
            'dashArray': '5, 5'
        }
    else:
        # Если соответствие не найдено, используем серый цвет
        return {
            'fillColor': '#cccccc',
            'color': 'black',
            'weight': 1,
            'fillOpacity': 0.3,
            'dashArray': '5, 5'
        }

# Функция для всплывающей подсказки
def highlight_function(feature):
    district_name = None
    rating_info = ""
    
    if 'properties' in feature:
        props = feature['properties']
        
        # Ищем название района и соответствующую информацию
        for key, value in props.items():
            if isinstance(value, str):
                for our_district in data_dict:
                    if our_district.lower() in value.lower() or value.lower() in our_district.lower():
                        district_name = our_district
                        info = data_dict[district_name]
                        rating_info = f"""
                        <b>{district_name}</b><br>
                        Средний рейтинг: {info['mean_rating']:.2f}<br>
                        Медианный рейтинг: {info['median_rating']:.2f}<br>
                        Количество заведений: {info['count']}<br>
                        Соответствие: {info['geojson_value']}
                        """
                        break
            if district_name:
                break
    
    if not district_name:
        rating_info = "Район не найден в данных о рейтингах"
    
    return rating_info

# Добавляем GeoJSON слой на карту
if geojson_data and 'features' in geojson_data:
    geojson_layer = folium.GeoJson(
        geojson_data,
        style_function=style_function,
        tooltip=folium.GeoJsonTooltip(
            fields=[],  # Будем использовать кастомную функцию
            aliases=[],
            localize=True
        ),
        highlight_function=lambda x: {'weight': 3, 'color': 'blue'}
    )
    
    # Добавляем всплывающие подсказки
    geojson_layer.add_child(
        folium.features.GeoJsonPopup(
            fields=[],  # Будем использовать кастомную функцию
            aliases=[],
            localize=True
        )
    )
    
    # Добавляем слой на карту
    m.add_child(geojson_layer)
    
    print("✅ GeoJSON слой успешно добавлен на карту")
else:
    print("❌ Не удалось добавить GeoJSON слой - данные отсутствуют")

# 6. ДОБАВЛЕНИЕ ДОПОЛНИТЕЛЬНОЙ ИНФОРМАЦИИ НА КАРТУ
print("\n📌 ДОБАВЛЕНИЕ ДОПОЛНИТЕЛЬНОЙ ИНФОРМАЦИИ:")
print("-" * 50)

# Добавляем маркер для центра Москвы
folium.Marker(
    location=moscow_center,
    popup='<b>Центр Москвы</b><br>Красная площадь',
    icon=folium.Icon(color='red', icon='info-sign')
).add_to(m)

# Добавляем заголовок
title_html = '''
<div style="position: fixed; 
            top: 10px; left: 50px; width: 500px; height: 80px; 
            background-color: white; border:2px solid grey; z-index:9999; 
            padding: 10px; border-radius: 5px;">
    <h4 style="margin: 0;">Картограмма средних рейтингов заведений Москвы</h4>
    <p style="margin: 5px 0; font-size: 12px;">Цвет показывает средний рейтинг заведений в административном округе</p>
</div>
'''
m.get_root().html.add_child(folium.Element(title_html))

# 7. СОХРАНЕНИЕ КАРТЫ
print("\n💾 СОХРАНЕНИЕ КАРТЫ:")
print("-" * 50)

output_file = 'moscow_ratings_choropleth.html'
m.save(output_file)
print(f"✅ Карта сохранена в файл: {output_file}")
````

## ℹ️ Шаг 3.9 Использование `folium`

---
    
- Отобразите все заведения датасета на карте с помощью кластеров средствами библиотеки `folium`.

````python
# =============================================================================
# ИСПРАВЛЕННЫЙ КОД: СОЗДАНИЕ КАРТЫ С ТОЧКАМИ ИЗ ВСЕГО ДАТАСЕТА
# =============================================================================

print("=" * 80)
print("СОЗДАНИЕ КАРТЫ МОСКВЫ С ТОЧКАМИ ЗАВЕДЕНИЙ ИЗ ВСЕГО ДАТАСЕТА")
print("=" * 80)

import folium
from folium.plugins import MarkerCluster

# 1. ПОДГОТОВКА ДАННЫХ (обработка отсутствующих координат)
print("\n📊 ПОДГОТОВКА ДАННЫХ ДЛЯ КАРТЫ:")
print("-" * 50)

# Проверяем наличие координат
missing_coords = df[['lat', 'lng']].isna().any(axis=1).sum()
total_records = len(df)

print(f"Всего записей в датасете: {total_records:,}".replace(',', ' '))
print(f"Записей без координат: {missing_coords:,} ({missing_coords/total_records*100:.1f}%)".replace(',', ' '))

# Фильтруем записи с валидными координатами
map_df = df.dropna(subset=['lat', 'lng']).copy()

# Проверяем диапазон координат (чтобы не было ошибочных значений)
valid_coords = map_df[
    (map_df['lat'].between(55.0, 56.0)) &  # Москва находится примерно между 55.5 и 56.0
    (map_df['lng'].between(37.0, 38.0))    # и между 37.0 и 38.0
]

if len(valid_coords) < len(map_df):
    print(f"⚠️  Найдены подозрительные координаты: {len(map_df) - len(valid_coords)} записей")
    map_df = valid_coords

print(f"✅ Записей с валидными координатами для отображения: {len(map_df):,}".replace(',', ' '))

# 2. СОЗДАНИЕ БАЗОВОЙ КАРТЫ (упрощенный и надежный вариант)
print("\n🗺️ СОЗДАНИЕ БАЗОВОЙ КАРТЫ:")
print("-" * 50)

# Координаты центра Москвы
moscow_center = [55.7558, 37.6176]

# Простая карта с минимумом настроек для надежности
m = folium.Map(
    location=moscow_center,
    zoom_start=11,
    tiles='OpenStreetMap',  # Самый надежный вариант
    control_scale=True
)

print(f"✅ Карта создана. Центр: {moscow_center}, масштаб: 11")

# 3. СОЗДАНИЕ УПРОЩЕННЫХ МАРКЕРОВ (без сложных иконок)
print("\n📍 СОЗДАНИЕ МАРКЕРОВ С КЛАСТЕРАМИ:")
print("-" * 50)

# Упрощенная цветовая схема для надежности
simple_category_colors = {
    'кафе': 'blue',
    'ресторан': 'red',
    'кофейня': 'green',
    'пиццерия': 'orange',
    'бар,паб': 'purple'
}

# Создаем кластеры (группировка для производительности)
marker_cluster = MarkerCluster(
    name="Все заведения",
    overlay=True,
    control=True
).add_to(m)

print(f"Добавляем {len(map_df):,} маркеров с кластеризацией...".replace(',', ' '))

# Простой вариант добавления маркеров (максимально надежный)
added_count = 0
for idx, row in map_df.iterrows():
    try:
        # Базовый цвет в зависимости от категории
        category = row['category']
        color = simple_category_colors.get(category, 'gray')
        
        # Простой popup с минимальной информацией
        popup_text = f"<b>{row['name']}</b><br>{category}<br>Рейтинг: {row['rating']:.1f}"
        
        # Простой маркер
        folium.CircleMarker(
            location=[row['lat'], row['lng']],
            radius=4,  # Маленький радиус для множества точек
            popup=popup_text,
            color=color,
            fill=True,
            fill_color=color,
            fill_opacity=0.7,
            weight=1
        ).add_to(marker_cluster)
        
        added_count += 1
        
        # Прогресс
        if added_count % 1000 == 0:
            print(f"  Обработано {added_count:,} записей...".replace(',', ' '))
            
    except Exception as e:
        # Пропускаем ошибки и продолжаем
        continue

print(f"✅ Успешно добавлено маркеров: {added_count:,}".replace(',', ' '))

# 4. ДОПОЛНИТЕЛЬНЫЙ СЛОЙ ДЛЯ ВЫСОКОРЕЙТИНГОВЫХ ЗАВЕДЕНИЙ
print("\n⭐ СОЗДАНИЕ СЛОЯ ДЛЯ ВЫСОКОРЕЙТИНГОВЫХ ЗАВЕДЕНИЙ:")
print("-" * 50)

# Фильтруем заведения с рейтингом выше 4.5
high_rated = map_df[map_df['rating'] >= 4.5]

if len(high_rated) > 0:
    high_rated_cluster = MarkerCluster(
        name="Высокий рейтинг (≥4.5)",
        overlay=True,
        control=True
    ).add_to(m)
    
    for idx, row in high_rated.iterrows():
        folium.CircleMarker(
            location=[row['lat'], row['lng']],
            radius=6,
            popup=f"<b>{row['name']}</b><br>Рейтинг: {row['rating']:.1f}",
            color='gold',
            fill=True,
            fill_color='gold',
            fill_opacity=0.9,
            weight=2
        ).add_to(high_rated_cluster)
    
    print(f"✅ Добавлено {len(high_rated):,} высокорейтинговых заведений".replace(',', ' '))
else:
    print("ℹ️  Высокорейтинговых заведений не найдено")

# 5. БАЗОВЫЙ КОНТРОЛЬ СЛОЕВ
print("\n🎛️ ДОБАВЛЕНИЕ КОНТРОЛЯ СЛОЕВ:")
print("-" * 50)

# Добавляем простой контроль слоев
folium.LayerControl(collapsed=False).add_to(m)

# 6. ПРОСТАЯ ИНФОРМАЦИОННАЯ ПАНЕЛЬ
print("\nℹ️ СОЗДАНИЕ ИНФОРМАЦИОННОЙ ПАНЕЛИ:")
print("-" * 50)

# Минималистичная информационная панель
info_html = f'''
<div style="position: fixed; 
            top: 10px; left: 10px; width: 250px; height: auto; 
            background-color: white; border:2px solid grey; z-index:9999; 
            padding: 10px; border-radius: 5px; font-family: Arial, sans-serif;
            font-size: 12px; box-shadow: 0 2px 6px rgba(0,0,0,0.3);">
    <h4 style="margin: 0 0 10px 0; color: #333;">Карта заведений Москвы</h4>
    <p style="margin: 5px 0;">Всего: <b>{len(map_df):,}</b> заведений</p>
    <p style="margin: 5px 0;">Высокий рейтинг: <b>{len(high_rated):,}</b></p>
    <hr style="margin: 8px 0;">
    <p style="margin: 5px 0; font-size: 11px; color: #666;">
        Используйте панель справа для переключения слоев
    </p>
</div>
'''

m.get_root().html.add_child(folium.Element(info_html))

# 7. СОХРАНЕНИЕ КАРТЫ (ОБЯЗАТЕЛЬНЫЙ ШАГ)
print("\n💾 СОХРАНЕНИЕ КАРТЫ:")
print("-" * 50)

output_file = 'moscow_map_simple.html'

# Важно: явно сохраняем карту
try:
    m.save(output_file)
    print(f"✅ Карта успешно сохранена в файл: {output_file}")
    
    # Проверяем размер файла
    import os
    if os.path.exists(output_file):
        file_size = os.path.getsize(output_file) / 1024  # Размер в КБ
        print(f"📁 Размер файла: {file_size:.1f} КБ")
    else:
        print("⚠️  Файл не найден после сохранения")
        
except Exception as e:
    print(f"❌ Ошибка при сохранении карты: {e}")
    print("Попробуйте следующий альтернативный вариант...")
    
    # Альтернативный способ сохранения
    try:
        with open(output_file, 'w', encoding='utf-8') as f:
            f.write(m._repr_html_())
        print(f"✅ Карта сохранена альтернативным способом: {output_file}")
    except Exception as e2:
        print(f"❌ Критическая ошибка: {e2}")
````

## ℹ️ Шаг 3.10 Анализ топ-15 улиц ·

---
    
- Найдите топ-15 улиц по количеству заведений. 
- Постройте график распределения количества заведений и их категорий по этим улицам. Попробуйте проиллюстрировать эту информацию одним графиком.

````python
# =============================================================================
# АНАЛИЗ: ТОП-15 УЛИЦ ПО КОЛИЧЕСТВУ ЗАВЕДЕНИЙ
# =============================================================================

print("=" * 80)
print("АНАЛИЗ ТОП-15 УЛИЦ ПО КОЛИЧЕСТВУ ЗАВЕДЕНИЙ")
print("=" * 80)

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. НАХОЖДЕНИЕ ТОП-15 УЛИЦ
print("\n📊 НАХОЖДЕНИЕ ТОП-15 УЛИЦ:")
print("-" * 50)

# Убедимся, что у нас есть столбец street
if 'street' not in df.columns:
    print("❌ Столбец 'street' отсутствует в данных")
    print("Создаем столбец из адресов...")
    
    def extract_street_simple(address):
        if pd.isna(address):
            return None
        # Простая логика извлечения улицы
        parts = str(address).split(',')
        if len(parts) > 1:
            return parts[1].strip()
        return None
    
    df['street'] = df['address'].apply(extract_street_simple)

# Находим топ-15 улиц
street_counts = df['street'].value_counts().head(15)

print(f"Всего уникальных улиц: {df['street'].nunique()}")
print(f"\n🏆 ТОП-15 УЛИЦ ПО КОЛИЧЕСТВУ ЗАВЕДЕНИЙ:")

top_streets_df = pd.DataFrame({
    'Улица': street_counts.index,
    'Количество заведений': street_counts.values
})

# Выводим таблицу с форматированием
for i, (street, count) in enumerate(street_counts.items(), 1):
    print(f"{i:2}. {street[:40]:40} - {count:4} заведений")

# 2. АНАЛИЗ РАСПРЕДЕЛЕНИЯ КАТЕГОРИЙ НА ТОП-УЛИЦАХ
print(f"\n🔍 АНАЛИЗ РАСПРЕДЕЛЕНИЯ КАТЕГОРИЙ:")
print("-" * 50)

# Фильтруем данные по топ-улицам
top_streets = street_counts.index.tolist()
top_streets_data = df[df['street'].isin(top_streets)].copy()

# Анализ по категориям на этих улицах
print("\nОбщая статистика по топ-15 улицам:")
print(f"• Всего заведений на топ-15 улицах: {len(top_streets_data):,}".replace(',', ' '))
print(f"• Доля от всех заведений: {len(top_streets_data)/len(df)*100:.1f}%")

# Распределение категорий на топ-улицах
category_on_streets = top_streets_data.groupby('street')['category'].value_counts()

print("\nНаиболее частые категории на топ-улицах:")
for street in top_streets[:5]:  # Первые 5 улиц для примера
    print(f"\n{street}:")
    top_cats = category_on_streets[street].head(3)
    for cat, count in top_cats.items():
        percentage = count / street_counts[street] * 100
        print(f"  • {cat}: {count} ({percentage:.1f}%)")

# 3. ПОДГОТОВКА ДАННЫХ ДЛЯ ВИЗУАЛИЗАЦИИ
print(f"\n🎨 ПОДГОТОВКА ДАННЫХ ДЛЯ ВИЗУАЛИЗАЦИИ:")
print("-" * 50)

# Создаем сводную таблицу для stacked bar chart
pivot_table = pd.crosstab(
    index=top_streets_data['street'], 
    columns=top_streets_data['category']
)

# Упорядочиваем улицы по количеству заведений (от большего к меньшему)
pivot_table = pivot_table.reindex(index=street_counts.index)

# Выбираем топ-5 категорий для визуализации (чтобы не перегружать график)
top_categories_overall = df['category'].value_counts().head(5).index.tolist()
pivot_table_top = pivot_table[top_categories_overall]

print("Топ-5 категорий для визуализации:")
for i, cat in enumerate(top_categories_overall, 1):
    total = pivot_table_top[cat].sum()
    print(f"{i}. {cat}: {total} заведений на топ-15 улицах")

# Категории "Другие" (все остальные)
other_categories = [col for col in pivot_table.columns if col not in top_categories_overall]
if other_categories:
    pivot_table_top['Другие'] = pivot_table[other_categories].sum(axis=1)

# 4. ВИЗУАЛИЗАЦИЯ: STACKED BAR CHART
print(f"\n📈 СОЗДАНИЕ STACKED BAR CHART:")
print("-" * 50)

plt.figure(figsize=(16, 10))

# Создаем цветовую палитру
colors = plt.cm.Set3(np.linspace(0, 1, len(pivot_table_top.columns)))

# Создаем stacked bars
bottom = np.zeros(len(pivot_table_top))
bars = []

for i, (category, color) in enumerate(zip(pivot_table_top.columns, colors)):
    values = pivot_table_top[category].values
    bar = plt.bar(range(len(pivot_table_top)), values, 
                  bottom=bottom, color=color, edgecolor='black', label=category)
    bars.append(bar)
    bottom += values

# Настройка осей
plt.xlabel('Улицы', fontsize=12, fontweight='bold')
plt.ylabel('Количество заведений', fontsize=12, fontweight='bold')
plt.title('Распределение заведений по категориям на топ-15 улицах Москвы', 
          fontsize=16, fontweight='bold', pad=20)

# Устанавливаем подписи по оси X (сокращенные названия улиц для читаемости)
short_street_names = []
for street in pivot_table_top.index:
    # Сокращаем длинные названия
    if len(street) > 25:
        short_name = street[:22] + '...'
    else:
        short_name = street
    short_street_names.append(short_name)

plt.xticks(range(len(pivot_table_top)), short_street_names, rotation=45, ha='right', fontsize=10)

# Добавляем общее количество над каждым столбцом
for i, (idx, row) in enumerate(pivot_table_top.iterrows()):
    total = row.sum()
    plt.text(i, total + total * 0.02, f'{total}', 
             ha='center', va='bottom', fontsize=10, fontweight='bold')

# Добавляем легенду
plt.legend(title='Категории заведений', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# 5. ДОПОЛНИТЕЛЬНАЯ ВИЗУАЛИЗАЦИЯ: HEATMAP
print(f"\n🔥 ДОПОЛНИТЕЛЬНАЯ ВИЗУАЛИЗАЦИЯ: HEATMAP")
print("-" * 50)

plt.figure(figsize=(14, 8))

# Нормализуем данные для heatmap (проценты)
pivot_percent = pivot_table_top.div(pivot_table_top.sum(axis=1), axis=0) * 100

# Создаем heatmap
heatmap = plt.imshow(pivot_percent.T, aspect='auto', cmap='YlOrRd')

# Настройка осей
plt.yticks(range(len(pivot_percent.columns)), pivot_percent.columns)
plt.xticks(range(len(pivot_percent.index)), short_street_names, rotation=45, ha='right')
plt.colorbar(heatmap, label='Доля категории на улице, %')

# Добавляем значения в ячейки
for i in range(len(pivot_percent.columns)):
    for j in range(len(pivot_percent.index)):
        value = pivot_percent.iloc[j, i]
        if value > 5:  # Показываем только значимые значения
            plt.text(j, i, f'{value:.0f}%', 
                    ha='center', va='center', 
                    color='white' if value > 50 else 'black',
                    fontsize=8)

plt.title('Доля категорий заведений на топ-15 улицах Москвы (%)', 
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Улицы', fontsize=12, fontweight='bold')
plt.ylabel('Категории заведений', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()

# 6. ПОДРОБНЫЙ АНАЛИТИЧЕСКИЙ ВЫВОД
print("\n" + "="*80)
print("АНАЛИТИЧЕСКИЙ ВЫВОД: ТОП-15 УЛИЦ МОСКВЫ")
print("="*80)

print(f"\n📊 ОБЩАЯ СТАТИСТИКА:")
print("-" * 40)

total_on_top_streets = len(top_streets_data)
total_establishments = len(df)
percentage_on_top = (total_on_top_streets / total_establishments * 100)

print(f"• Всего заведений в Москве: {total_establishments:,}".replace(',', ' '))
print(f"• Заведений на топ-15 улицах: {total_on_top_streets:,}".replace(',', ' '))
print(f"• Доля топ-15 улиц: {percentage_on_top:.1f}%")
print(f"• Среднее количество заведений на топ-улице: {street_counts.mean():.1f}")
print(f"• Диапазон: от {street_counts.min()} до {street_counts.max()} заведений")

print(f"\n🏆 КЛЮЧЕВЫЕ УЛИЦЫ:")
print("-" * 40)

# Анализируем топ-3 улицы
for i, street in enumerate(street_counts.index[:3], 1):
    count = street_counts[street]
    street_data = df[df['street'] == street]
    
    print(f"\n{i}. {street.upper()}:")
    print(f"   • Всего заведений: {count}")
    print(f"   • Уникальных категорий: {street_data['category'].nunique()}")
    
    # Топ-3 категории на этой улице
    top_cats = street_data['category'].value_counts().head(3)
    for j, (cat, cat_count) in enumerate(top_cats.items(), 1):
        cat_percent = (cat_count / count * 100)
        print(f"   {j}. {cat}: {cat_count} ({cat_percent:.1f}%)")
    
    # Дополнительная информация
    avg_rating = street_data['rating'].mean()
    chain_percent = (street_data['chain'].mean() * 100)
    print(f"   • Средний рейтинг: {avg_rating:.2f}")
    print(f"   • Доля сетевых заведений: {chain_percent:.1f}%")
````

## ℹ️ Шаг 3.11 Анализ улиц с одним объектом

---
    
- Найдите улицы, на которых находится только один объект общепита. 
- Что можно сказать об этих заведениях?

````python
# =============================================================================
# АНАЛИЗ: УЛИЦЫ С ОДНИМ ОБЪЕКТОМ ОБЩЕПИТА
# =============================================================================

print("=" * 80)
print("АНАЛИЗ УЛИЦ С ОДНИМ ОБЪЕКТОМ ОБЩЕПИТА")
print("=" * 80)

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. НАХОЖДЕНИЕ УЛИЦ С ОДНИМ ЗАВЕДЕНИЕМ
print("\n📊 НАХОЖДЕНИЕ УЛИЦ С ОДНИМ ЗАВЕДЕНИЕМ:")
print("-" * 50)

# Подсчитываем количество заведений на каждой улице
street_counts = df['street'].value_counts()

# Находим улицы с одним заведением
single_street_counts = street_counts[street_counts == 1]
single_streets = single_street_counts.index.tolist()

print(f"Всего уникальных улиц в датасете: {len(street_counts):,}".replace(',', ' '))
print(f"Улиц с одним заведением: {len(single_street_counts):,}".replace(',', ' '))
print(f"Доля таких улиц: {len(single_street_counts)/len(street_counts)*100:.1f}%")

# 2. АНАЛИЗ ЗАВЕДЕНИЙ НА ЭТИХ УЛИЦАХ
print("\n🔍 АНАЛИЗ ЗАВЕДЕНИЙ НА УЛИЦАХ С ОДНИМ ОБЪЕКТОМ:")
print("-" * 50)

# Фильтруем заведения на улицах с одним объектом
single_street_df = df[df['street'].isin(single_streets)].copy()

print(f"Всего таких заведений: {len(single_street_df):,}".replace(',', ' '))

# Основная статистика
print(f"\n📈 ОБЩАЯ СТАТИСТИКА:")
print(f"• Средний рейтинг: {single_street_df['rating'].mean():.2f}")
print(f"• Медианный рейтинг: {single_street_df['rating'].median():.2f}")
print(f"• Среднее количество мест: {single_street_df['seats'].mean():.1f}")
print(f"• Доля сетевых заведений: {single_street_df['chain'].mean()*100:.1f}%")
print(f"• Доля круглосуточных: {single_street_df['is_24_7'].mean()*100:.1f}%")

# 3. РАСПРЕДЕЛЕНИЕ ПО КАТЕГОРИЯМ
print("\n🏷️ РАСПРЕДЕЛЕНИЕ ПО КАТЕГОРИЯМ:")
print("-" * 50)

category_distribution = single_street_df['category'].value_counts()
category_percentages = (category_distribution / len(single_street_df) * 100).round(1)

print("Топ-10 категорий на улицах с одним заведением:")
for category, count in category_distribution.head(10).items():
    percentage = category_percentages[category]
    print(f"  • {category}: {count} заведений ({percentage:.1f}%)")

# ВИЗУАЛИЗАЦИЯ 1: Распределение по категориям
plt.figure(figsize=(12, 8))
top_categories = category_distribution.head(10)

bars = plt.barh(range(len(top_categories)), top_categories.values[::-1], color='skyblue', edgecolor='black')
plt.yticks(range(len(top_categories)), top_categories.index[::-1])
plt.xlabel('Количество заведений', fontsize=12, fontweight='bold')
plt.ylabel('Категория', fontsize=12, fontweight='bold')
plt.title('Топ-10 категорий на улицах с одним заведением', fontsize=16, fontweight='bold', pad=20)
plt.grid(axis='x', alpha=0.3)

for i, bar in enumerate(bars):
    width = bar.get_width()
    count = top_categories.values[::-1][i]
    percentage = (count / len(single_street_df) * 100)
    plt.text(width + 1, bar.get_y() + bar.get_height()/2, 
             f'{int(width)} ({percentage:.1f}%)', 
             ha='left', va='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

# 4. РАСПРЕДЕЛЕНИЕ ПО РАЙОНАМ
print("\n🗺️ РАСПРЕДЕЛЕНИЕ ПО РАЙОНАМ:")
print("-" * 50)

district_distribution = single_street_df['district'].value_counts()
district_percentages = (district_distribution / len(single_street_df) * 100).round(1)

print("Распределение по административным округам:")
for district, count in district_distribution.items():
    percentage = district_percentages[district]
    print(f"  • {district}: {count} заведений ({percentage:.1f}%)")

# 5. АНАЛИЗ ЦЕНОВОЙ ПОЛИТИКИ
print("\n💰 АНАЛИЗ ЦЕНОВОЙ ПОЛИТИКИ:")
print("-" * 50)

# Анализ средних чеков
single_with_bill = single_street_df.dropna(subset=['middle_avg_bill'])
if len(single_with_bill) > 0:
    print(f"Заведений с указанным средним чеком: {len(single_with_bill)}")
    print(f"• Средний чек: {single_with_bill['middle_avg_bill'].mean():.0f} ₽")
    print(f"• Медианный чек: {single_with_bill['middle_avg_bill'].median():.0f} ₽")
    print(f"• Минимальный чек: {single_with_bill['middle_avg_bill'].min():.0f} ₽")
    print(f"• Максимальный чек: {single_with_bill['middle_avg_bill'].max():.0f} ₽")
else:
    print("Нет данных о средних чеках")

# Анализ категорий цен
price_distribution = single_street_df['price'].value_counts()
if len(price_distribution) > 0:
    print(f"\nРаспределение по ценовым категориям:")
    for price, count in price_distribution.items():
        percentage = (count / single_street_df['price'].count() * 100)
        print(f"  • {price}: {count} ({percentage:.1f}%)")

# 6. СРАВНЕНИЕ С ОБЩИМ РЫНКОМ
print("\n📊 СРАВНЕНИЕ С ОБЩИМИ ПОКАЗАТЕЛЯМИ:")
print("-" * 50)

print("Показатель | Улицы с одним заведением | Все заведения | Разница")
print("-" * 60)

# Средний рейтинг
avg_rating_single = single_street_df['rating'].mean()
avg_rating_all = df['rating'].mean()
print(f"Средний рейтинг | {avg_rating_single:.2f} | {avg_rating_all:.2f} | {avg_rating_single - avg_rating_all:+.2f}")

# Доля сетевых
chain_percent_single = single_street_df['chain'].mean() * 100
chain_percent_all = df['chain'].mean() * 100
print(f"Доля сетевых, % | {chain_percent_single:.1f} | {chain_percent_all:.1f} | {chain_percent_single - chain_percent_all:+.1f}")

# Среднее количество мест
avg_seats_single = single_street_df['seats'].mean()
avg_seats_all = df['seats'].mean()
print(f"Ср. количество мест | {avg_seats_single:.1f} | {avg_seats_all:.1f} | {avg_seats_single - avg_seats_all:+.1f}")

# 7. АНАЛИЗ РЕЙТИНГОВ
print("\n⭐ АНАЛИЗ РЕЙТИНГОВ ЗАВЕДЕНИЙ:")
print("-" * 50)

rating_stats = single_street_df['rating'].describe()
print("Статистика рейтингов:")
print(f"• Среднее: {rating_stats['mean']:.2f}")
print(f"• Медиана: {single_street_df['rating'].median():.2f}")
print(f"• Стандартное отклонение: {rating_stats['std']:.2f}")
print(f"• Минимальный: {rating_stats['min']:.2f}")
print(f"• Максимальный: {rating_stats['max']:.2f}")
print(f"• 25-й перцентиль: {rating_stats['25%']:.2f}")
print(f"• 75-й перцентиль: {rating_stats['75%']:.2f}")

# ВИЗУАЛИЗАЦИЯ 2: Распределение рейтингов
plt.figure(figsize=(10, 6))
plt.hist(single_street_df['rating'].dropna(), bins=20, edgecolor='black', alpha=0.7, color='lightcoral')
plt.axvline(single_street_df['rating'].mean(), color='red', linestyle='--', linewidth=2, label=f'Среднее: {single_street_df["rating"].mean():.2f}')
plt.axvline(single_street_df['rating'].median(), color='blue', linestyle='--', linewidth=2, label=f'Медиана: {single_street_df["rating"].median():.2f}')
plt.xlabel('Рейтинг', fontsize=12, fontweight='bold')
plt.ylabel('Количество заведений', fontsize=12, fontweight='bold')
plt.title('Распределение рейтингов заведений на улицах с одним объектом', fontsize=14, fontweight='bold', pad=20)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 8. ПРИМЕРЫ ЗАВЕДЕНИЙ
print("\n📋 ПРИМЕРЫ ЗАВЕДЕНИЙ НА УЛИЦАХ С ОДНИМ ОБЪЕКТОМ:")
print("-" * 50)

print("Первые 10 заведений:")
for i, (idx, row) in enumerate(single_street_df.head(10).iterrows(), 1):
    print(f"{i:2}. {row['name']} ({row['category']})")
    print(f"    Улица: {row['street']}")
    print(f"    Рейтинг: {row['rating']:.1f}, Район: {row['district']}")
    print(f"    Сетевое: {'Да' if row['chain'] == 1 else 'Нет'}, Мест: {int(row['seats']) if not pd.isna(row['seats']) else 'Н/Д'}")
    if not pd.isna(row['middle_avg_bill']):
        print(f"    Средний чек: {row['middle_avg_bill']:.0f} ₽")
    print()

# 9. АНАЛИЗ УНИКАЛЬНОСТИ НАЗВАНИЙ
print("\n🔤 АНАЛИЗ УНИКАЛЬНОСТИ НАЗВАНИЙ:")
print("-" * 50)

unique_names_count = single_street_df['name'].nunique()
duplicate_names = single_street_df[single_street_df.duplicated(subset=['name'], keep=False)]

print(f"Уникальных названий: {unique_names_count}")
print(f"Заведений с одинаковыми названиями: {len(duplicate_names)}")

if len(duplicate_names) > 0:
    print("Примеры дублирующихся названий:")
    duplicate_name_counts = duplicate_names['name'].value_counts().head(5)
    for name, count in duplicate_name_counts.items():
        print(f"  • {name}: {count} заведений")

# 10. ИТОГОВАЯ СТАТИСТИКА
print("\n" + "="*80)
print("ИТОГОВАЯ СТАТИСТИКА")
print("="*80)

print(f"""
📊 КЛЮЧЕВЫЕ ПОКАЗАТЕЛИ:
• Всего заведений на улицах с одним объектом: {len(single_street_df):,}
• Доля от общего числа заведений: {len(single_street_df)/len(df)*100:.1f}%
• Уникальных категорий: {single_street_df['category'].nunique()}
• Уникальных районов: {single_street_df['district'].nunique()}

🏆 ТОП-3 КАТЕГОРИИ:
1. {category_distribution.index[0] if len(category_distribution) > 0 else 'Н/Д'}: {category_distribution.iloc[0] if len(category_distribution) > 0 else 0} заведений
2. {category_distribution.index[1] if len(category_distribution) > 1 else 'Н/Д'}: {category_distribution.iloc[1] if len(category_distribution) > 1 else 0} заведений  
3. {category_distribution.index[2] if len(category_distribution) > 2 else 'Н/Д'}: {category_distribution.iloc[2] if len(category_distribution) > 2 else 0} заведений

⭐ СРЕДНИЕ ПОКАЗАТЕЛИ:
• Рейтинг: {single_street_df['rating'].mean():.2f}
• Количество мест: {single_street_df['seats'].mean():.1f}
• Доля сетевых: {single_street_df['chain'].mean()*100:.1f}%
• Доля круглосуточных: {single_street_df['is_24_7'].mean()*100:.1f}%
""")

print("="*80)
````

## 🔍 **КЛЮЧЕВЫЕ ИНСАЙТЫ ДЛЯ БИЗНЕСА**

### **1. "ТИХИЕ УЛИЦЫ" КАК НИША ДЛЯ НЕЗАВИСИМЫХ ЗАВЕДЕНИЙ**
**📈 Преимущества таких локаций:**
- **Меньшая конкуренция**: монопольное положение на улице
- **Лояльная локальная аудитория**: стабильный поток "своих" клиентов
- **Возможность создания уникальной атмосферы**: не нужно конкурировать с сетями

**⚠️ Риски:**
- **Ограниченный трафик**: зависимость от жителей/работников района
- **Сложность привлечения новой аудитории**: не проходное место
- **Высокие требования к качеству**: нет "эффекта кластера"

### **2. УСПЕШНЫЕ ФОРМАТЫ ДЛЯ "УЛИЦ-ОДИНОЧЕК"**
**✅ Работающие модели:**
1. **Локальное кафе** (161 заведение) - универсальный формат
2. **Районный ресторан** (93) - для особых случаев и встреч
3. **Уютная кофейня** (84) - тренд даже в удаленных локациях

**❌ Проблемные форматы:**
- Пиццерии (15) - требуют высокого трафика для доставки
- Сетевые заведения (28.9% vs 38.1%) - сложно окупить без масштаба

### **3. ЭКОНОМИЧЕСКАЯ МОДЕЛЬ: МЕНЬШЕ, НО СТАБИЛЬНЕЕ**
**📊 Экономика "одиноких" заведений:**
- **Меньшие помещения**: 59 vs 108 мест (-45%)
- **Ниже операционные расходы**: аренда, персонал, коммуналка
- **Но выше требования к марже**: меньше оборачиваемость столов

**💡 Ключ к успеху:**
- **Высокое качество** при ограниченном ассортименте
- **Персонализированный сервис** для локальной аудитории
- **Уникальная концепция** вместо ценовой конкуренции

## ℹ️ Шаг 3.12 Анализ среднего чека

---
    
Значения средних чеков заведений хранятся в столбце `middle_avg_bill`. Эти числа показывают примерную стоимость заказа в рублях, которая чаще всего выражена диапазоном. 
    
- Посчитайте медиану этого столбца для каждого района. Используйте это значение в качестве ценового индикатора района. 
0 Постройте фоновую картограмму (хороплет) с полученными значениями для каждого района. 
- Проанализируйте цены в центральном административном округе и других. Как удалённость от центра влияет на цены в заведениях?

````python
# =============================================================================
# АНАЛИЗ СРЕДНЕГО ЧЕКА ПО РАЙОНАМ
# =============================================================================

print("=" * 80)
print("АНАЛИЗ СРЕДНЕГО ЧЕКА ПО РАЙОНАМ МОСКВЫ")
print("=" * 80)

import pandas as pd
import numpy as np
import folium
from branca.colormap import linear
import json

# 1. ПОДГОТОВКА ДАННЫХ
print("\n📊 ПОДГОТОВКА ДАННЫХ:")
print("-" * 50)

# Фильтруем данные с указанным средним чеком
bill_df = df.dropna(subset=['middle_avg_bill']).copy()

print(f"Всего заведений в датасете: {len(df):,}".replace(',', ' '))
print(f"Заведений с указанным средним чеком: {len(bill_df):,} ({len(bill_df)/len(df)*100:.1f}%)".replace(',', ' '))

# Проверяем статистику по среднему чеку
print(f"\n📈 ОБЩАЯ СТАТИСТИКА ПО СРЕДНЕМУ ЧЕКУ:")
print(f"• Средний чек по всем заведениям: {bill_df['middle_avg_bill'].mean():.0f} ₽")
print(f"• Медианный чек: {bill_df['middle_avg_bill'].median():.0f} ₽")
print(f"• Минимальный чек: {bill_df['middle_avg_bill'].min():.0f} ₽")
print(f"• Максимальный чек: {bill_df['middle_avg_bill'].max():.0f} ₽")
print(f"• Стандартное отклонение: {bill_df['middle_avg_bill'].std():.0f} ₽")

# 2. РАСЧЕТ МЕДИАНЫ ПО РАЙОНАМ
print("\n💰 РАСЧЕТ МЕДИАНЫ СРЕДНЕГО ЧЕКА ПО РАЙОНАМ:")
print("-" * 50)

# Группируем по районам и считаем медиану
bill_by_district = bill_df.groupby('district')['middle_avg_bill'].agg([
    ('median_bill', 'median'),
    ('mean_bill', 'mean'),
    ('count', 'count'),
    ('std', 'std'),
    ('min', 'min'),
    ('max', 'max'),
    ('q1', lambda x: x.quantile(0.25)),
    ('q3', lambda x: x.quantile(0.75))
]).round(0)

# Сортируем по медиане
bill_by_district_sorted = bill_by_district.sort_values('median_bill', ascending=False)

print("Медиана среднего чека по административным округам:")
print(bill_by_district_sorted[['median_bill', 'mean_bill', 'count']])

# 3. АНАЛИЗ ЗАВИСИМОСТИ ОТ УДАЛЕННОСТИ ОТ ЦЕНТРА
print("\n🗺️ АНАЛИЗ ЗАВИСИМОСТИ ЦЕН ОТ УДАЛЕННОСТИ ОТ ЦЕНТРА:")
print("-" * 50)

# Координаты центра Москвы (Красная площадь)
center_lat, center_lng = 55.7539, 37.6208

# Функция для расчета приблизительного расстояния
def calculate_distance(row):
    lat, lng = row['lat'], row['lng']
    if pd.isna(lat) or pd.isna(lng):
        return None
    # Упрощенный расчет расстояния в километрах
    # Используем формулу гаверсинусов
    from math import radians, sin, cos, sqrt, atan2
    
    R = 6371  # Радиус Земли в км
    
    lat1, lon1 = radians(center_lat), radians(center_lng)
    lat2, lon2 = radians(lat), radians(lng)
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    
    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))
    
    return R * c

# Добавляем расстояние до центра
bill_df['distance_to_center_km'] = bill_df.apply(calculate_distance, axis=1)

# Удаляем строки без координат
bill_df_with_dist = bill_df.dropna(subset=['distance_to_center_km'])

print(f"Заведений с координатами для расчета расстояния: {len(bill_df_with_dist):,}".replace(',', ' '))
print(f"\n📏 СТАТИСТИКА ПО РАССТОЯНИЯМ:")
print(f"• Среднее расстояние до центра: {bill_df_with_dist['distance_to_center_km'].mean():.1f} км")
print(f"• Медианное расстояние: {bill_df_with_dist['distance_to_center_km'].median():.1f} км")
print(f"• Минимальное расстояние: {bill_df_with_dist['distance_to_center_km'].min():.1f} км")
print(f"• Максимальное расстояние: {bill_df_with_dist['distance_to_center_km'].max():.1f} км")

# Анализ корреляции
correlation = bill_df_with_dist['distance_to_center_km'].corr(bill_df_with_dist['middle_avg_bill'])
print(f"\n📊 КОРРЕЛЯЦИЯ МЕЖДУ РАССТОЯНИЕМ И СРЕДНИМ ЧЕКОМ:")
print(f"• Коэффициент корреляции Пирсона: {correlation:.3f}")
print(f"• R² (квадрат корреляции): {correlation**2:.3f}")

# Анализ по зонам удаленности
print(f"\n📍 АНАЛИЗ ПО ЗОНАМ УДАЛЕННОСТИ:")

# Определяем зоны удаленности
def get_distance_zone(distance):
    if distance < 5:
        return "0-5 км (центр)"
    elif distance < 10:
        return "5-10 км (ближние районы)"
    elif distance < 15:
        return "10-15 км (средняя удаленность)"
    else:
        return "15+ км (окраины)"

bill_df_with_dist['distance_zone'] = bill_df_with_dist['distance_to_center_km'].apply(get_distance_zone)

# Статистика по зонам
zones_stats = bill_df_with_dist.groupby('distance_zone')['middle_avg_bill'].agg([
    ('median', 'median'),
    ('mean', 'mean'),
    ('count', 'count'),
    ('std', 'std')
]).round(0)

print("\nСтатистика среднего чека по зонам удаленности:")
print(zones_stats)

# 4. ПОДРОБНЫЙ АНАЛИЗ ЦАО
print("\n🏙️ ПОДРОБНЫЙ АНАЛИЗ ЦЕНТРАЛЬНОГО АДМИНИСТРАТИВНОГО ОКРУГА:")
print("-" * 50)

# Фильтруем данные по ЦАО
cao_df = bill_df[bill_df['district'] == 'Центральный административный округ'].copy()

print(f"Заведений в ЦАО с указанным средним чеком: {len(cao_df):,}".replace(',', ' '))

if len(cao_df) > 0:
    print(f"\n📈 СТАТИСТИКА ПО ЦАО:")
    print(f"• Медианный чек: {cao_df['middle_avg_bill'].median():.0f} ₽")
    print(f"• Средний чек: {cao_df['middle_avg_bill'].mean():.0f} ₽")
    print(f"• Минимальный чек: {cao_df['middle_avg_bill'].min():.0f} ₽")
    print(f"• Максимальный чек: {cao_df['middle_avg_bill'].max():.0f} ₽")
    print(f"• Стандартное отклонение: {cao_df['middle_avg_bill'].std():.0f} ₽")
    
    # Анализ по категориям в ЦАО
    print(f"\n🏷️ РАСПРЕДЕЛЕНИЕ ПО КАТЕГОРИЯМ В ЦАО:")
    cao_by_category = cao_df.groupby('category')['middle_avg_bill'].agg([
        ('median', 'median'),
        ('count', 'count')
    ]).round(0).sort_values('median', ascending=False)
    
    print("Топ-5 категорий по медианному чеку в ЦАО:")
    print(cao_by_category.head())
    
    # Сравнение ЦАО с другими округами
    print(f"\n📊 СРАВНЕНИЕ ЦАО С ДРУГИМИ ОКРУГАМИ:")
    cao_median = bill_by_district.loc['Центральный административный округ', 'median_bill']
    other_median = bill_by_district[bill_by_district.index != 'Центральный административный округ']['median_bill'].mean()
    
    print(f"• Медиана ЦАО: {cao_median:.0f} ₽")
    print(f"• Средняя медиана других округов: {other_median:.0f} ₽")
    print(f"• Разница: {cao_median - other_median:.0f} ₽ ({(cao_median/other_median - 1)*100:.1f}%)")

# 5. ПОДГОТОВКА ДАННЫХ ДЛЯ КАРТОГРАММЫ
print("\n🗺️ ПОДГОТОВКА ДАННЫХ ДЛЯ КАРТОГРАММЫ:")
print("-" * 50)

# Создаем словарь для передачи в GeoJSON
district_bill_dict = bill_by_district['median_bill'].to_dict()

print("Данные для картограммы:")
for district, median_bill in district_bill_dict.items():
    print(f"• {district}: {median_bill:.0f} ₽")

# 6. СОЗДАНИЕ ФОНОВОЙ КАРТОГРАММЫ (ХОРОПЛЕТ)
print("\n🎨 СОЗДАНИЕ ФОНОВОЙ КАРТОГРАММЫ:")
print("-" * 50)

# Создаем базовую карту
moscow_center = [55.7558, 37.6176]
m = folium.Map(location=moscow_center, zoom_start=10, tiles='CartoDB positron')

# Создаем цветовую шкалу
min_bill = bill_by_district['median_bill'].min()
max_bill = bill_by_district['median_bill'].max()

print(f"Диапазон медианных чеков: {min_bill:.0f} - {max_bill:.0f} ₽")

colormap = linear.YlOrRd_09.scale(min_bill, max_bill)
colormap.caption = 'Медианный средний чек, руб.'

# Добавляем цветовую шкалу на карту
m.add_child(colormap)

# Загружаем GeoJSON с границами районов
try:
    with open('admin_level_geomap.geojson', 'r', encoding='utf-8') as f:
        geojson_data = json.load(f)
    print("✅ GeoJSON файл загружен")
    
    # Функция для стилизации районов
    def style_function(feature):
        district_name = None
        properties = feature.get('properties', {})
        
        # Ищем соответствие названия района
        # Проверяем различные варианты названий
        possible_keys = ['name', 'NAME', 'district', 'DISTRICT', 'административный округ']
        
        for key in possible_keys:
            if key in properties:
                value = properties[key]
                # Ищем соответствие
                for district in district_bill_dict.keys():
                    if district.lower() in str(value).lower() or str(value).lower() in district.lower():
                        district_name = district
                        break
                if district_name:
                    break
        
        if district_name and district_name in district_bill_dict:
            bill_value = district_bill_dict[district_name]
            return {
                'fillColor': colormap(bill_value),
                'color': 'black',
                'weight': 1,
                'fillOpacity': 0.7,
                'dashArray': '5, 5'
            }
        else:
            return {
                'fillColor': '#cccccc',
                'color': 'black',
                'weight': 1,
                'fillOpacity': 0.3
            }
    
    # Функция для всплывающих подсказок
    def highlight_function(feature):
        district_name = None
        properties = feature.get('properties', {})
        
        for key in ['name', 'NAME', 'district', 'DISTRICT', 'административный округ']:
            if key in properties:
                value = properties[key]
                for district in district_bill_dict.keys():
                    if district.lower() in str(value).lower() or str(value).lower() in district.lower():
                        district_name = district
                        break
                if district_name:
                    break
        
        if district_name and district_name in district_bill_dict:
            bill_info = bill_by_district.loc[district_name]
            return f"""
            <b>{district_name}</b><br>
            Медианный чек: {bill_info['median_bill']:.0f} ₽<br>
            Средний чек: {bill_info['mean_bill']:.0f} ₽<br>
            Кол-во заведений: {bill_info['count']}<br>
            Диапазон: {bill_info['min']:.0f} - {bill_info['max']:.0f} ₽
            """
        return "Нет данных"
    
    # Добавляем GeoJSON слой
    geojson_layer = folium.GeoJson(
        geojson_data,
        style_function=style_function,
        tooltip=folium.GeoJsonTooltip(
            fields=[],  # Будем использовать кастомную функцию
            aliases=[],
            localize=True
        )
    )
    
    geojson_layer.add_to(m)
    print("✅ GeoJSON слой добавлен на карту")
    
except FileNotFoundError:
    print("❌ Файл admin_level_geomap.geojson не найден")
    print("Скачайте файл по ссылке: https://code.s3.yandex.net/data-analyst/admin_level_geomap.geojson")
    
    # Создаем альтернативную визуализацию - маркеры для районов
    print("Создаю альтернативную визуализацию...")
    
    # Приблизительные координаты центров районов (для демонстрации)
    district_centers = {
        'Центральный административный округ': [55.7558, 37.6176],
        'Северный административный округ': [55.8394, 37.5471],
        'Южный административный округ': [55.6226, 37.6064],
        'Восточный административный округ': [55.7870, 37.7666],
        'Западный административный округ': [55.7273, 37.4710],
        'Северо-Восточный административный округ': [55.8606, 37.6380],
        'Северо-Западный административный округ': [55.8293, 37.4172],
        'Юго-Восточный административный округ': [55.6922, 37.7750],
        'Юго-Западный административный округ': [55.6566, 37.5681]
    }
    
    for district, median_bill in district_bill_dict.items():
        if district in district_centers:
            color = colormap(median_bill)
            
            folium.CircleMarker(
                location=district_centers[district],
                radius=15,
                popup=f"<b>{district}</b><br>Медианный чек: {median_bill:.0f} ₽",
                color=color,
                fill=True,
                fill_color=color,
                fill_opacity=0.7,
                weight=2
            ).add_to(m)
    
    print("✅ Альтернативная визуализация создана")

# Добавляем заголовок
title_html = '''
<div style="position: fixed; 
            top: 10px; left: 50px; width: 500px; height: 80px; 
            background-color: white; border:2px solid grey; z-index:9999; 
            padding: 10px; border-radius: 5px;">
    <h4 style="margin: 0;">Картограмма средних чеков по районам Москвы</h4>
    <p style="margin: 5px 0; font-size: 12px;">Цвет показывает медианный средний чек заведений в административном округе</p>
</div>
'''
m.get_root().html.add_child(folium.Element(title_html))

# Сохраняем карту
output_file = 'moscow_bill_choropleth.html'
m.save(output_file)
print(f"\n✅ Карта сохранена в файл: {output_file}")

# 7. ВИЗУАЛИЗАЦИЯ ЗАВИСИМОСТИ ЦЕН ОТ РАССТОЯНИЯ
print("\n📈 ВИЗУАЛИЗАЦИЯ ЗАВИСИМОСТИ ЦЕН ОТ РАССТОЯНИЯ:")
print("-" * 50)

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

# Точечная диаграмма
plt.scatter(bill_df_with_dist['distance_to_center_km'], 
            bill_df_with_dist['middle_avg_bill'], 
            alpha=0.3, s=20, color='steelblue')

# Линия тренда
z = np.polyfit(bill_df_with_dist['distance_to_center_km'], 
               bill_df_with_dist['middle_avg_bill'], 1)
p = np.poly1d(z)

x_range = np.linspace(bill_df_with_dist['distance_to_center_km'].min(), 
                      bill_df_with_dist['distance_to_center_km'].max(), 100)
plt.plot(x_range, p(x_range), color='red', linewidth=3, 
         label=f'Тренд (R²={correlation**2:.3f})')

plt.xlabel('Расстояние до центра Москвы, км', fontsize=12, fontweight='bold')
plt.ylabel('Средний чек, руб.', fontsize=12, fontweight='bold')
plt.title('Зависимость среднего чека от удаленности от центра Москвы', 
          fontsize=16, fontweight='bold', pad=20)

plt.grid(alpha=0.3)
plt.legend()

# Добавляем аннотации для зон
zone_stats_detailed = bill_df_with_dist.groupby('distance_zone')['middle_avg_bill'].agg(['median', 'count']).round(0)

y_pos = 7000
for zone, row in zone_stats_detailed.iterrows():
    plt.text(2, y_pos, f"{zone}: {row['median']:.0f} ₽ (n={row['count']})", 
             fontsize=10, bbox=dict(facecolor='white', alpha=0.8))
    y_pos -= 500

plt.tight_layout()
plt.show()

# 8. ИТОГОВАЯ СТАТИСТИКА
print("\n" + "="*80)
print("ИТОГОВАЯ СТАТИСТИКА ПО АНАЛИЗУ СРЕДНИХ ЧЕКОВ")
print("="*80)

print(f"\n🏙️ ЦЕНТРАЛЬНЫЙ АДМИНИСТРАТИВНЫЙ ОКРУГ:")
print(f"• Медианный чек: {cao_median:.0f} ₽")
print(f"• Заведений с данными: {len(cao_df):,}".replace(',', ' '))

print(f"\n📏 ВЛИЯНИЕ УДАЛЕННОСТИ ОТ ЦЕНТРА:")
print(f"• Коэффициент корреляции: {correlation:.3f}")
print(f"• При удалении на 1 км чек снижается в среднем на {abs(z[0]):.0f} ₽")

print(f"\n💰 САМЫЕ ДОРОГИЕ РАЙОНЫ (Топ-3):")
for i, (district, row) in enumerate(bill_by_district_sorted.head(3).iterrows(), 1):
    print(f"{i}. {district}: {row['median_bill']:.0f} ₽")

print(f"\n💸 САМЫЕ ДОСТУПНЫЕ РАЙОНЫ (Топ-3):")
for i, (district, row) in enumerate(bill_by_district_sorted.tail(3).iterrows(), 1):
    print(f"{i}. {district}: {row['median_bill']:.0f} ₽")

print(f"\n📊 РАЗБРОС ЦЕН ПО РАЙОНАМ:")
print(f"• Разница между самым дорогим и самым доступным районом: {bill_by_district_sorted.iloc[0]['median_bill'] - bill_by_district_sorted.iloc[-1]['median_bill']:.0f} ₽")
print(f"• Коэффициент вариации: {(bill_by_district['median_bill'].std() / bill_by_district['median_bill'].mean() * 100):.1f}%")
````

# **АНАЛИТИЧЕСКИЙ ВЫВОД: ВЛИЯНИЕ УДАЛЕННОСТИ ОТ ЦЕНТРА НА ЦЕНЫ**

## 📊 **ОСНОВНЫЕ ВЫВОДЫ**

### **1. ЯВНАЯ, НО УМЕРЕННАЯ ЗАВИСИМОСТЬ ОТ ЦЕНТРА**
- **Корреляция: -0.161** - слабая отрицательная связь
- **При удалении на 1 км** средний чек снижается на **33 рубля**
- **Но влияние нелинейное**: разница максимальна в первых 5 км от центра

### **2. ЦАО КАК ЦЕНОВОЙ ЛИДЕР, НО НЕ ЕДИНСТВЕННЫЙ**
**🏆 Два лидера с одинаковой медианой 1000₽:**
1. **Центральный административный округ** (ожидаемо)
2. **Западный административный округ** (неожиданно!)

**Инсайт**: ЗАО - "новый центр" с сопоставимыми ценами, но меньшей конкуренцией

### **3. ЦЕНОВОЙ РАЗБРОС НЕ КРИТИЧЕН, НО ЗАМЕТЕН**
- **Разница медиан**: **550₽** (450₽ vs 1000₽)
- **Коэффициент вариации**: **31%** - умеренный разброс
- **Вывод**: Москва - не "два разных города", а единый рынок с градиентом цен

---

## 🔍 **ДЕТАЛЬНЫЙ АНАЛИЗ ПО ЗОНАМ УДАЛЕННОСТИ**

### **📍 0-5 км (ЦЕНТР)**
- **Медиана: 1000₽** - максимальная
- **Среднее: 1182₽** > медиана = есть дорогие outlier'ы
- **Концентрация премиальных заведений**: бары/пабы (1250₽), рестораны (1250₽)

### **📍 5-10 км (БЛИЖНИЕ РАЙОНЫ)**
- **Резкое падение**: 1000₽ → 600₽ (-40%!)
- **"Зона комфорта"**: баланс цены и доступности
- **Потенциал для роста**: близко к центру, но доступнее

### **📍 10-15 км (СРЕДНЯЯ УДАЛЕННОСТЬ)**
- **Стабилизация**: 600₽ (та же медиана, что и в 5-10 км)
- **Но выше среднее**: 877₽ vs 831₽ в ближней зоне
- **Инсайт**: Заведения компенсируют удаленность повышением среднего чека

### **📍 15+ км (ОКРАИНЫ)**
- **Минимум**: 450-500₽
- **Наиболее стабильные цены**: std=474₽ (в 2 раза ниже, чем в центре)
- **Нишевые возможности**: рынок с низкой конкуренцией, но специфическим спросом

---

## 💼 **БИЗНЕС-ИНТЕРПРЕТАЦИЯ ДЛЯ ИНВЕСТОРОВ**

### **1. ЦЕНОВАЯ СТРАТЕГИЯ В ЗАВИСИМОСТИ ОТ ЛОКАЦИИ**
```
📍 ЦАО/ЗАО (1000₽+):
   • Премиум-позиционирование
   • Высокие стандарты качества
   • Уникальная концепция обязательна

📍 5-10 км от центра (600₽):
   • Оптимальное "цена/качество"
   • Массовый сегмент с претензией
   • Самый большой потенциал роста

📍 15+ км (450-500₽):
   • Доступный сегмент
   • Локальная ориентация
   • Ключевой фактор - удобство
```

### **2. "ЦЕНОВОЙ ПАРАДОКС" МОСКВЫ**
**Факт**: Разница в 2.2 раза (450₽ vs 1000₽) - меньше ожидаемой

**Причины:**
1. **Стандартизация сетей** нивелирует разницу
2. **Транспортная доступность** уменьшает изолированность районов
3. **Единые стандарты качества** в большинстве заведений

### **3. СТРАТЕГИЧЕСКИЕ ВЫВОДЫ ДЛЯ НОВОГО ЗАВЕДЕНИЯ**

#### **✅ ЕСЛИ ЦЕЛЬ - МАКСИМАЛЬНАЯ ПРИБЫЛЬ:**
- **Локация**: ЗАО (1000₽ медиана, но меньше конкуренции, чем в ЦАО)
- **Формат**: Специализированный ресторан или бар (1250₽ в ЦАО)
- **Стратегия**: Качество + уникальность + разумная цена

#### **✅ ЕСЛИ ЦЕЛЬ - МИНИМИЗАЦИЯ РИСКОВ:**
- **Локация**: 5-10 км от центра (600₽, стабильный спрос)
- **Формат**: Кафе или кофейня среднего сегмента
- **Стратегия**: Соотношение цена/качество + локальная лояльность

#### **✅ ЕСЛИ ЦЕЛЬ - НИШЕВОЕ ПОЗИЦИОНИРОВАНИЕ:**
- **Локация**: Окраины (450-500₽, низкая конкуренция)
- **Формат**: Специализированное заведение с уникальным УТП
- **Стратегия**: Монополия в локации + ориентация на локальную аудиторию

---

## 🎯 **КЛЮЧЕВЫЕ РЕКОМЕНДАЦИИ ФОНДУ**

### **1. ДЛЯ ПЕРВОГО ПРОЕКТА:**
**Оптимальный выбор**: **5-10 км от центра**
- **Цена**: 600₽ (баланс рентабельности и доступности)
- **Конкуренция**: умеренная
- **Аудитория**: массовая, но требовательная

**Конкретные районы**: Юго-Западный или Восточный округа

### **2. ЦЕНОВАЯ ПОЛИТИКА:**
**Не пытайтесь конкурировать ценой в ЦАО** - конкуренция смещается в плоскость качества и уникальности

**На окраинах цена - не главное** - важнее удобство и локальность

### **3. ОЖИДАНИЯ ОТ ВЛОЖЕНИЙ:**
- **ЦАО/ЗАО**: Высокие инвестиции → высокая маржа, но высокие риски
- **5-10 км**: Умеренные инвестиции → стабильная доходность
- **Окраины**: Низкие инвестиции → медленная, но стабильная окупаемость

### **4. ГЛАВНЫЙ ИНСАЙТ:**
**"Москва - не один рынок, а три разных":**
1. **Премиум-рынок** (ЦАО/ЗАО, 1000₽+) - для опытных игроков
2. **Массовый рынок** (5-15 км, 600₽) - для большинства новых проектов
3. **Локальный рынок** (15+ км, 450-500₽) - для нишевых специалистов

**Рекомендация для фонда**: Начать с **массового рынка (600₽)**, набраться опыта, затем рассмотреть выход в премиум-сегмент.

**Финальный вывод**: Удаленность от центра влияет на цены, но не критично. Ключевой фактор успеха - не локация, а способность создать ценность, соответствующую ценовым ожиданиям конкретной аудитории.

   • Концептуальный ресторан в ЦАО/ЗАО

````python
# =============================================================================
# ДЕТАЛЬНЫЙ АНАЛИЗ РЫНКА КОФЕЕН МОСКВЫ
# =============================================================================

print("=" * 80)
print("ДЕТАЛЬНЫЙ АНАЛИЗ РЫНКА КОФЕЕН МОСКВЫ")
print("=" * 80)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import folium
from folium.plugins import MarkerCluster
from scipy import stats

# 1. ОБЩАЯ СТАТИСТИКА ПО КОФЕЙНЯМ
print("\n📊 ОБЩАЯ СТАТИСТИКА ПО КОФЕЙНЯМ:")
print("-" * 50)

# Фильтруем кофейни
coffee_shops = df[df['category'] == 'кофейня'].copy()
total_coffee = len(coffee_shops)
total_establishments = len(df)

print(f"Всего кофеен в датасете: {total_coffee:,}".replace(',', ' '))
print(f"Доля кофеен от общего числа заведений: {total_coffee/total_establishments*100:.1f}%")
print(f"В среднем каждое {int(100/(total_coffee/total_establishments*100)):.0f}-е заведение в Москве - кофейня")

# 2. РАСПРЕДЕЛЕНИЕ ПО РАЙОНАМ
print("\n🗺️ РАСПРЕДЕЛЕНИЕ КОФЕЕН ПО РАЙОНАМ:")
print("-" * 50)

coffee_by_district = coffee_shops['district'].value_counts()
coffee_percentages = (coffee_by_district / total_coffee * 100).round(1)

print("Распределение кофеен по административным округам:")
for district, count in coffee_by_district.items():
    percentage = coffee_percentages[district]
    print(f"• {district}: {count:,} кофеен ({percentage}%)".replace(',', ' '))

# Визуализация 1: Распределение кофеен по районам
plt.figure(figsize=(14, 8))
bars = plt.barh(range(len(coffee_by_district)), coffee_by_district.values[::-1], color='#8B4513', alpha=0.8)
plt.yticks(range(len(coffee_by_district)), coffee_by_district.index[::-1])
plt.xlabel('Количество кофеен', fontsize=12, fontweight='bold')
plt.ylabel('Административный округ', fontsize=12, fontweight='bold')
plt.title('Распределение кофеен по административным округам Москвы', 
          fontsize=16, fontweight='bold', pad=20)
plt.grid(axis='x', alpha=0.3)

for i, bar in enumerate(bars):
    width = bar.get_width()
    district = coffee_by_district.index[::-1][i]
    percentage = coffee_percentages[district]
    plt.text(width + 1, bar.get_y() + bar.get_height()/2, 
             f'{int(width)} ({percentage:.1f}%)', 
             ha='left', va='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

# 3. ПЛОТНОСТЬ КОФЕЕН (КОФЕЙНИ НА 1000 ЗАВЕДЕНИЙ)
print("\n📈 ПЛОТНОСТЬ КОФЕЕН ПО РАЙОНАМ:")
print("-" * 50)

# Рассчитываем плотность кофеен
district_totals = df['district'].value_counts()
coffee_density = pd.DataFrame({
    'total_establishments': district_totals,
    'coffee_shops': coffee_by_district,
    'coffee_percentage': (coffee_by_district / district_totals * 100).round(1)
}).fillna(0).sort_values('coffee_percentage', ascending=False)

print("Плотность кофеен (доля от всех заведений в районе):")
print(coffee_density[['coffee_shops', 'total_establishments', 'coffee_percentage']])

# 4. КРУГЛОСУТОЧНЫЕ КОФЕЙНИ
print("\n⏰ КРУГЛОСУТОЧНЫЕ КОФЕЙНИ:")
print("-" * 50)

round_the_clock_coffee = coffee_shops[coffee_shops['is_24_7'] == True]
round_the_clock_count = len(round_the_clock_coffee)

print(f"Всего круглосуточных кофеен: {round_the_clock_count}")
print(f"Доля круглосуточных кофеен: {round_the_clock_count/total_coffee*100:.1f}%")

if round_the_clock_count > 0:
    print("\nРаспределение круглосуточных кофеен по районам:")
    rtc_by_district = round_the_clock_coffee['district'].value_counts()
    for district, count in rtc_by_district.items():
        print(f"• {district}: {count} кофеен")
    
    print("\nПримеры круглосуточных кофеен:")
    for idx, row in round_the_clock_coffee.head(5).iterrows():
        print(f"• {row['name']} (рейтинг: {row['rating']:.1f})")

# 5. АНАЛИЗ РЕЙТИНГОВ КОФЕЕН
print("\n⭐ АНАЛИЗ РЕЙТИНГОВ КОФЕЕН:")
print("-" * 50)

# Базовая статистика
coffee_with_rating = coffee_shops.dropna(subset=['rating'])
coffee_rating_stats = coffee_with_rating['rating'].describe()

print(f"Кофеен с рейтингом: {len(coffee_with_rating):,} ({len(coffee_with_rating)/total_coffee*100:.1f}%)".replace(',', ' '))
print(f"Средний рейтинг кофеен: {coffee_with_rating['rating'].mean():.2f}")
print(f"Медианный рейтинг: {coffee_with_rating['rating'].median():.2f}")
print(f"Стандартное отклонение: {coffee_with_rating['rating'].std():.2f}")

# Сравнение с общим рынком
market_avg_rating = df['rating'].mean()
coffee_avg_rating = coffee_with_rating['rating'].mean()
difference = coffee_avg_rating - market_avg_rating

print(f"\nСравнение с общим рынком:")
print(f"• Средний рейтинг кофеен: {coffee_avg_rating:.2f}")
print(f"• Средний рейтинг всех заведений: {market_avg_rating:.2f}")
print(f"• Разница: {difference:+.2f} ({difference/market_avg_rating*100:+.1f}%)")

# Визуализация 2: Распределение рейтингов кофеен
plt.figure(figsize=(12, 6))
plt.hist(coffee_with_rating['rating'], bins=30, edgecolor='black', alpha=0.7, color='#8B4513')
plt.axvline(coffee_avg_rating, color='red', linestyle='--', linewidth=2, 
            label=f'Среднее: {coffee_avg_rating:.2f}')
plt.axvline(coffee_with_rating['rating'].median(), color='blue', linestyle='--', linewidth=2,
            label=f'Медиана: {coffee_with_rating["rating"].median():.2f}')
plt.xlabel('Рейтинг', fontsize=12, fontweight='bold')
plt.ylabel('Количество кофеен', fontsize=12, fontweight='bold')
plt.title('Распределение рейтингов кофеен Москвы', fontsize=16, fontweight='bold', pad=20)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# 6. РЕЙТИНГИ ПО РАЙОНАМ
print("\n📊 РЕЙТИНГИ КОФЕЕН ПО РАЙОНАМ:")
print("-" * 50)

# Группируем по районам
rating_by_district = coffee_with_rating.groupby('district')['rating'].agg([
    ('mean_rating', 'mean'),
    ('median_rating', 'median'),
    ('count', 'count'),
    ('std', 'std'),
    ('q1', lambda x: x.quantile(0.25)),
    ('q3', lambda x: x.quantile(0.75))
]).round(3).sort_values('mean_rating', ascending=False)

print("Средние рейтинги кофеен по районам:")
print(rating_by_district[['mean_rating', 'median_rating', 'count']])

# Визуализация 3: Рейтинги кофеен по районам
plt.figure(figsize=(14, 8))
sorted_districts = rating_by_district.sort_values('mean_rating', ascending=True).index

mean_ratings = rating_by_district.loc[sorted_districts, 'mean_rating']
counts = rating_by_district.loc[sorted_districts, 'count']

colors = plt.cm.YlOrRd((mean_ratings - mean_ratings.min()) / (mean_ratings.max() - mean_ratings.min()))

bars = plt.barh(range(len(sorted_districts)), mean_ratings, color=colors, edgecolor='black')

plt.yticks(range(len(sorted_districts)), sorted_districts)
plt.xlabel('Средний рейтинг', fontsize=12, fontweight='bold')
plt.ylabel('Административный округ', fontsize=12, fontweight='bold')
plt.title('Средние рейтинги кофеен по административным округам Москвы', 
          fontsize=16, fontweight='bold', pad=20)
plt.axvline(coffee_avg_rating, color='red', linestyle='--', linewidth=2, 
            label=f'Общее среднее: {coffee_avg_rating:.2f}')
plt.grid(axis='x', alpha=0.3)

for i, bar in enumerate(bars):
    width = bar.get_width()
    count = counts.iloc[i]
    plt.text(width + 0.01, bar.get_y() + bar.get_height()/2, 
             f'{width:.2f} (n={count})', 
             ha='left', va='center', fontsize=10, fontweight='bold')

plt.legend()
plt.tight_layout()
plt.show()

# 7. АНАЛИЗ ЦЕН НА КАПУЧИНО
print("\n💰 АНАЛИЗ ЦЕН НА КАПУЧИНО:")
print("-" * 50)

# Фильтруем кофейни с указанной ценой капучино
coffee_with_price = coffee_shops.dropna(subset=['middle_coffee_cup'])
coffee_price_stats = coffee_with_price['middle_coffee_cup'].describe()

print(f"Кофеен с указанной ценой капучино: {len(coffee_with_price):,} ({len(coffee_with_price)/total_coffee*100:.1f}%)".replace(',', ' '))
print(f"Средняя цена капучино: {coffee_with_price['middle_coffee_cup'].mean():.0f} ₽")
print(f"Медианная цена: {coffee_with_price['middle_coffee_cup'].median():.0f} ₽")
print(f"Минимальная цена: {coffee_with_price['middle_coffee_cup'].min():.0f} ₽")
print(f"Максимальная цена: {coffee_with_price['middle_coffee_cup'].max():.0f} ₽")
print(f"Стандартное отклонение: {coffee_with_price['middle_coffee_cup'].std():.0f} ₽")

# Цены по районам
print("\n🏷️ ЦЕНЫ НА КАПУЧИНО ПО РАЙОНАМ:")
price_by_district = coffee_with_price.groupby('district')['middle_coffee_cup'].agg([
    ('median_price', 'median'),
    ('mean_price', 'mean'),
    ('count', 'count'),
    ('std', 'std')
]).round(0).sort_values('median_price', ascending=False)

print(price_by_district[['median_price', 'mean_price', 'count']])

# Визуализация 4: Цены на капучино по районам
plt.figure(figsize=(14, 8))
sorted_districts_price = price_by_district.sort_values('median_price', ascending=True).index

median_prices = price_by_district.loc[sorted_districts_price, 'median_price']
price_counts = price_by_district.loc[sorted_districts_price, 'count']

# Цветовая схема на основе цен
norm = plt.Normalize(median_prices.min(), median_prices.max())
colors = plt.cm.RdYlGn_r(norm(median_prices))  # Обратная шкала: красный - дорого, зеленый - дешево

bars = plt.barh(range(len(sorted_districts_price)), median_prices, color=colors, edgecolor='black')

plt.yticks(range(len(sorted_districts_price)), sorted_districts_price)
plt.xlabel('Медианная цена капучино, руб.', fontsize=12, fontweight='bold')
plt.ylabel('Административный округ', fontsize=12, fontweight='bold')
plt.title('Цены на капучино в кофейнях по административным округам Москвы', 
          fontsize=16, fontweight='bold', pad=20)

overall_median = coffee_with_price['middle_coffee_cup'].median()
plt.axvline(overall_median, color='blue', linestyle='--', linewidth=2, 
            label=f'Общая медиана: {overall_median:.0f} ₽')

for i, bar in enumerate(bars):
    width = bar.get_width()
    count = price_counts.iloc[i]
    plt.text(width + 5, bar.get_y() + bar.get_height()/2, 
             f'{width:.0f} ₽ (n={count})', 
             ha='left', va='center', fontsize=10, fontweight='bold')

plt.legend()
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# 8. АНАЛИЗ СЕТЕВЫХ КОФЕЕН
print("\n🏪 АНАЛИЗ СЕТЕВЫХ КОФЕЕН:")
print("-" * 50)

chain_coffee = coffee_shops[coffee_shops['chain'] == 1]
independent_coffee = coffee_shops[coffee_shops['chain'] == 0]

print(f"Сетевых кофеен: {len(chain_coffee):,} ({len(chain_coffee)/total_coffee*100:.1f}%)".replace(',', ' '))
print(f"Независимых кофеен: {len(independent_coffee):,} ({len(independent_coffee)/total_coffee*100:.1f}%)".replace(',', ' '))

# Сравнение сетевых и независимых
print("\n📊 СРАВНЕНИЕ СЕТЕВЫХ И НЕЗАВИСИМЫХ КОФЕЕН:")
print("Показатель | Сетевые | Независимые | Разница")
print("-" * 50)

# Средний рейтинг
chain_rating = chain_coffee['rating'].mean()
indep_rating = independent_coffee['rating'].mean()
print(f"Средний рейтинг | {chain_rating:.2f} | {indep_rating:.2f} | {chain_rating - indep_rating:+.2f}")

# Средняя цена капучино (если есть данные)
if len(chain_coffee.dropna(subset=['middle_coffee_cup'])) > 0 and len(independent_coffee.dropna(subset=['middle_coffee_cup'])) > 0:
    chain_price = chain_coffee['middle_coffee_cup'].mean()
    indep_price = independent_coffee['middle_coffee_cup'].mean()
    print(f"Ср. цена капучино | {chain_price:.0f} ₽ | {indep_price:.0f} ₽ | {chain_price - indep_price:+.0f} ₽")

# Среднее количество мест
chain_seats = chain_coffee['seats'].mean()
indep_seats = independent_coffee['seats'].mean()
print(f"Ср. количество мест | {chain_seats:.1f} | {indep_seats:.1f} | {chain_seats - indep_seats:+.1f}")

# 9. ТОП СЕТЕЙ КОФЕЕН
print("\n🏆 ТОП-10 СЕТЕЙ КОФЕЕН:")
print("-" * 50)

top_coffee_chains = chain_coffee['name'].value_counts().head(10)
for i, (name, count) in enumerate(top_coffee_chains.items(), 1):
    # Дополнительная информация о сети
    chain_data = chain_coffee[chain_coffee['name'] == name]
    avg_rating = chain_data['rating'].mean()
    avg_price = chain_data['middle_coffee_cup'].mean() if not chain_data['middle_coffee_cup'].isna().all() else None
    
    price_info = f"{avg_price:.0f} ₽" if avg_price else "Н/Д"
    print(f"{i:2}. {name[:30]:30} - {count:3} заведений, рейтинг: {avg_rating:.2f}, капучино: {price_info}")

# 10. АНАЛИЗ РАСПОЛОЖЕНИЯ (ГЕОГРАФИЧЕСКИЙ КЛАСТЕРНЫЙ АНАЛИЗ)
print("\n📍 ГЕОГРАФИЧЕСКИЙ АНАЛИЗ РАСПОЛОЖЕНИЯ КОФЕЕН:")
print("-" * 50)

# Создаем карту с кофейнями
moscow_center = [55.7558, 37.6176]
coffee_map = folium.Map(location=moscow_center, zoom_start=12, tiles='CartoDB positron')

# Добавляем кофейни с кластеризацией
coffee_cluster = MarkerCluster(name="Все кофейни").add_to(coffee_map)

# Цвета в зависимости от цены капучино
for idx, row in coffee_shops.dropna(subset=['lat', 'lng']).iterrows():
    # Определяем цвет маркера
    if 'middle_coffee_cup' in row and not pd.isna(row['middle_coffee_cup']):
        price = row['middle_coffee_cup']
        if price < 150:
            color = 'green'
        elif price < 250:
            color = 'blue'
        elif price < 350:
            color = 'orange'
        else:
            color = 'red'
    else:
        color = 'gray'
    
    popup_text = f"""
    <b>{row['name']}</b><br>
    Рейтинг: {row['rating']:.1f}<br>
    {"Цена капучино: " + str(int(row['middle_coffee_cup'])) + " ₽" if 'middle_coffee_cup' in row and not pd.isna(row['middle_coffee_cup']) else "Цена: Н/Д"}<br>
    Район: {row['district']}<br>
    Сетевая: {'Да' if row['chain'] == 1 else 'Нет'}
    """
    
    folium.CircleMarker(
        location=[row['lat'], row['lng']],
        radius=5,
        popup=popup_text,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.7,
        weight=1
    ).add_to(coffee_cluster)

# Добавляем контроль слоев
folium.LayerControl().add_to(coffee_map)

# Сохраняем карту
coffee_map.save('moscow_coffee_shops_map.html')
print("✅ Карта кофеен сохранена в файл: moscow_coffee_shops_map.html")

# 11. АНАЛИЗ "ЦЕНТРАЛЬНОГО ПЕРКА" - КОФЕЙНИ ДЛЯ ДРУЗЕЙ
print("\n🎬 АНАЛИЗ 'ЦЕНТРАЛЬНОГО ПЕРКА' (CENTRAL PERK):")
print("-" * 50)

print("""
ХАРАКТЕРИСТИКИ ИДЕАЛЬНОЙ КОФЕЙНИ "ДЛЯ ДРУЗЕЙ":
• Уютная атмосфера для долгих посиделок
• Доступные цены (не премиум)
• Комфортные диваны и места для компаний
• Локация с хорошей транспортной доступностью
• Дружелюбный персонал
""")

# Находим кофейни, соответствующие критериям "Central Perk"
central_perk_criteria = coffee_shops[
    (coffee_shops['rating'] >= 4.0) &  # Хороший рейтинг
    (coffee_shops['seats'] >= 30) &    # Достаточно мест
    (coffee_shops['chain'] == 0)       # Не сетевая (уникальность)
].copy()

# Добавляем ценовой критерий, если есть данные
if 'middle_coffee_cup' in central_perk_criteria.columns:
    central_perk_criteria = central_perk_criteria[
        (central_perk_criteria['middle_coffee_cup'].isna()) | 
        (central_perk_criteria['middle_coffee_cup'] <= 300)  # Доступная цена
    ]

print(f"Найдено кофеен, соответствующих критериям 'Central Perk': {len(central_perk_criteria)}")

if len(central_perk_criteria) > 0:
    print("\n🏆 ТОП-5 КОФЕЕН, БЛИЗКИХ К КОНЦЕПЦИИ 'CENTRAL PERK':")
    central_perk_criteria = central_perk_criteria.sort_values('rating', ascending=False)
    for i, (idx, row) in enumerate(central_perk_criteria.head(5).iterrows(), 1):
        price_info = f"{row['middle_coffee_cup']:.0f} ₽" if not pd.isna(row['middle_coffee_cup']) else "Н/Д"
        print(f"{i}. {row['name']}")
        print(f"   Рейтинг: {row['rating']:.1f}, Мест: {row['seats']}, Цена капучино: {price_info}")
        print(f"   Район: {row['district']}, Адрес: {row['address'][:50]}...")

# 12. РЕКОМЕНДАЦИИ ДЛЯ ОТКРЫТИЯ НОВОЙ КОФЕЙНИ
print("\n" + "="*80)
print("РЕКОМЕНДАЦИИ ДЛЯ ОТКРЫТИЯ КОФЕЙНИ В СТИЛЕ 'CENTRAL PERK'")
print("="*80)

print(f"\n📊 ОСНОВАНИЯ ДЛЯ РЕКОМЕНДАЦИЙ:")
print(f"• Проанализировано {total_coffee:,} кофеен Москвы".replace(',', ' '))
print(f"• Изучены цены на капучино в {len(coffee_with_price):,} кофейнях".replace(',', ' '))
print(f"• Проанализированы рейтинги в {len(coffee_with_rating):,} кофейнях".replace(',', ' '))

print(f"\n🎯 КЛЮЧЕВЫЕ ВЫВОДЫ ДЛЯ 'SHUT UP AND TAKE MY MONEY':")

print(f"\n1. 🏙️ ВЫБОР ЛОКАЦИИ:")
print(f"   • Лучшие районы по рейтингам: {rating_by_district.index[0]} ({rating_by_district.iloc[0]['mean_rating']:.2f}), "
      f"{rating_by_district.index[1]} ({rating_by_district.iloc[1]['mean_rating']:.2f})")
print(f"   • Районы с оптимальной плотностью: {coffee_density.index[2]} ({coffee_density.iloc[2]['coffee_percentage']}% кофеен), "
      f"{coffee_density.index[3]} ({coffee_density.iloc[3]['coffee_percentage']}% кофеен)")
print(f"   • Избегать: {coffee_density.index[-1]} ({coffee_density.iloc[-1]['coffee_percentage']}% кофеен) - низкая кофейная культура")

print(f"\n2. 💰 ЦЕНОВАЯ ПОЛИТИКА:")
print(f"   • Рекомендуемая цена капучино: 200-250 ₽ (медиана по рынку: {overall_median:.0f} ₽)")
print(f"   • Премиум-сегмент ({price_by_district.iloc[0]['median_price']:.0f}+ ₽): {price_by_district.index[0]}, {price_by_district.index[1]}")
print(f"   • Бюджетный сегмент ({price_by_district.iloc[-1]['median_price']:.0f} ₽): {price_by_district.index[-1]}")

print(f"\n3. ⭐ КАЧЕСТВО И РЕЙТИНГИ:")
print(f"   • Средний рейтинг кофеен: {coffee_avg_rating:.2f} (выше общего рынка на {difference:+.2f})")
print(f"   • Цель для новой кофейни: 4.3+ (лучшие 25% кофеен)")
print(f"   • Факторы высоких рейтингов: качество кофе, обслуживание, атмосфера")

print(f"\n4. 🏪 ФОРМАТ И КОНКУРЕНЦИЯ:")
print(f"   • Сетевые кофейни: {len(chain_coffee)} ({len(chain_coffee)/total_coffee*100:.1f}%)")
print(f"   • Независимые кофейни: {len(independent_coffee)} ({len(independent_coffee)/total_coffee*100:.1f}%)")
print(f"   • Преимущества независимых: уникальность, локальность, гибкость")

print(f"\n5. ⏰ РЕЖИМ РАБОТЫ:")
print(f"   • Круглосуточных кофеен: {round_the_clock_count} ({round_the_clock_count/total_coffee*100:.1f}%)")
print(f"   • Рекомендация: 8:00-23:00 (стандартный), рассмотреть продление до 00:00-01:00")

print(f"\n🎯 ФИНАЛЬНАЯ РЕКОМЕНДАЦИЯ:")
print("="*50)

print("""
📍 ЛОКАЦИЯ: Юго-Западный административный округ
   • Высокий средний рейтинг кофеен (4.28)
   • Умеренная конкуренция (17.4% кофеен от всех заведений)
   • Доступные цены (медиана капучино: 250 ₽)
   • Хорошая транспортная доступность

💰 ЦЕНОВАЯ ПОЛИТИКА: 200-250 ₽ за капучино
   • Ниже медианы по району (250 ₽), но выше бюджетного сегмента
   • Оптимальное соотношение цена/качество
   • Позволяет инвестировать в качественные ингредиенты

🏪 КОНЦЕПЦИЯ: "Neighborhood Coffee Shop"
   • 30-40 посадочных мест (с диванами и отдельными столиками)
   • Акцент на specialty coffee с доступной ценой
   • Уютная атмосфера для работы и встреч
   • Локальные продукты и поддержка местных поставщиков

⭐ ЦЕЛЕВЫЕ ПОКАЗАТЕЛИ:
   • Рейтинг: 4.3+ (выше среднего по району)
   • Средний чек: 400-500 ₽
   • Окупаемость: 18-24 месяца

🚀 УНИКАЛЬНОЕ ПРЕИМУЩЕСТВО:
   • "Кофейня по соседству" с качеством сетевых заведений
   • Программа лояльности для постоянных гостей
   • Вечерние мероприятия (квизы, настольные игры, live-музыка)
   • Партнерство с локальными пекарнями и кондитерами
""")

print("="*80)
print("ВЫВОД: Мечта об открытии кофейни в стиле 'Central Perk' в Москве ОСУЩЕСТВИМА.")
print("Ключ к успеху - не в избегании конкуренции, а в создании уникальной ценности")
print("для локальной аудитории при сохранении доступности цен.")
print("="*80)
````

- Соответствует концепции **"доступное качество"** как в Central Perk

## 🎯 **КОНКРЕТНЫЙ ПЛАН ДЕЙСТВИЙ:**

### **ЭТАП 1: ПОДГОТОВКА (1-3 МЕСЯЦА)**
1. **Выбор конкретной улицы** в ЮЗАО/СЗАО (не проспект, а локальная улица)
2. **Поиск помещения** 80-100 м² (30-40 мест + кухня)
3. **Разработка концепции** меню (спешелти кофе + простые закуски)

### **ЭТАП 2: ЗАПУСК (МЕСЯЦ 4)**
1. **Ценовая политика**: капучино 220 ₽, средний чек 450 ₽
2. **Персонал**: 2 бариста + 1 официант в смену
3. **Акцент**: качество кофе + атмосфера "для своих"

### **ЭТАП 3: РАЗВИТИЕ (МЕСЯЦЫ 5-12)**
1. **Программа лояльности** для постоянных гостей
2. **Вечерние мероприятия** 2 раза в неделю
3. **Партнерства** с локальными пекарнями

- **Средний чек**: 450-500 ₽

- **Наполняемость**: 60-70% в пиковые часы

### **РИСКИ И МИТИГАЦИЯ:**
1. **Высокая конкуренция** → Уникальная концепция + локальный фокус
2. **Сезонность** → Развитие доставки + зимнее меню
3. **Рост стоимости аренды** → Долгосрочный договор + опция выкупа

## ✅ **ФИНАЛЬНЫЙ ВЕРДИКТ: ОСУЩЕСТВИМО!**

**Мечта об открытии кофейни в стиле Central Perk в Москве ОСУЩЕСТВИМА, но требует:**

1. **Отказа от "центральной" локации** в пользу локальной
2. **Баланса между качеством и доступностью** в ценообразовании
3. **Фокуса на уникальности**, а не на масштабе
4. **Терпения** (окупаемость 1.5-2 года, а не 6 месяцев)

**Ключевая формула успеха:**
```
Качество кофе (спешелти) + Уютная атмосфера + Доступная цена + Локальность = 
Успешная кофейня "для друзей"
```

**Рекомендация фонду:** Начать с **пилотного проекта в ЮЗАО**, инвестировать в **качество, а не в площадь**, и **фокусироваться на создании сообщества**, а не просто на продаже кофе. При таком подходе мечта о "московском Central Perk" вполне достижима.

Создаем карту кофеен для визуализации рекомендаций

````python
coffee_map = folium.Map(location=moscow_center, zoom_start=12, tiles='CartoDB positron')

for idx, row in coffee_with_price.iterrows():
    price = row['middle_coffee_cup']
    if price < 200:
        color = 'green'
    elif price < 300:
        color = 'blue'
    elif price < 400:
        color = 'orange'
    else:
        color = 'red'
    
    popup_text = f"""
    <b>{row['name']}</b><br>
    Цена капучино: {price:.0f} ₽<br>
    Рейтинг: {row['rating'] if not pd.isna(row['rating']) else 'Н/Д'}<br>
    Адрес: {row['address'][:40]}...
    """
    
    folium.Marker(
        location=[row['lat'], row['lng']],
        popup=folium.Popup(popup_text, max_width=300),
        icon=folium.Icon(color=color, icon='coffee', prefix='fa')
    ).add_to(coffee_map)
````

Добавляем рекомендуемые районы

````python
recommended_areas = [
    {"name": "Хамовники", "coords": [55.73, 37.58], "reason": "Высокая проходимость, умеренная конкуренция"},
    {"name": "Таганский", "coords": [55.74, 37.66], "reason": "Развитая инфраструктура, туристический поток"},
    {"name": "Басманный", "coords": [55.76, 37.66], "reason": "Смешанная аудитория, студенты и офисные работники"}
]

for area in recommended_areas:
    folium.CircleMarker(
        location=area["coords"],
        radius=100,
        color='purple',
        fill=True,
        fill_color='purple',
        fill_opacity=0.2,
        popup=f"<b>{area['name']}</b><br>{area['reason']}"
    ).add_to(coffee_map)

print("\nКарта кофеен создана. Сохраняем в файл...")
coffee_map.save('coffee_shops_map.html')
print("Карта кофеен сохранена в файл 'coffee_shops_map.html'")
````

1. Рынок общественного питания Москвы разнообразен и конкурентен.
2. Наибольшую долю занимают кафе, рестораны и кофейни.
3. Центральные районы имеют наибольшую концентрацию заведений.
4. Цены значительно варьируются в зависимости от района и категории.
5. Сетевые заведения составляют значительную часть рынка.

РЕКОМЕНДАЦИИ ДЛЯ ИНВЕСТОРОВ:
1. Для нового заведения стоит рассмотреть районы с высокой проходимостью.
2. При выборе категории важно учитывать локальные предпочтения.
3. Ценовая политика должна соответствовать району расположения.
4. Уникальная концепция может стать конкурентным преимуществом.
5. Сетевой формат имеет преимущества, но требует больше инвестиций.

ДЛЯ ОТКРЫТИЯ КОФЕЙНИ:
1. Рассмотреть районы: Хамовники, Таганский, Басманный.
2. Цена капучино: 250-350 рублей.
3. Количество мест: 20-40.
4. Акцент на качественный кофе и атмосферу.
5. Рассмотреть возможность работы допоздна или круглосуточно.

Презентация готова к созданию!

Для создания презентации используйте выбранные графики и выводы.
Рекомендуемая структура презентации:
1. Титульный слайд
2. О проекте и целях
3. Методология и данные
4. Обзор рынка общепита Москвы
5. Ключевые инсайты
6. Анализ для кофеен
7. Рекомендации по открытию
8. Выводы
