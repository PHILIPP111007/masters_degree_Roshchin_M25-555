# Предобработка данных

```python
# Проверка дубликатов
print(f"Количество полных дубликатов: {df.duplicated().sum()}")

# Проверка пропущенных значений
print("\nПРОПУЩЕННЫЕ ЗНАЧЕНИЯ:")
missing_data = df.isnull().sum()
missing_percent = (missing_data / len(df)) * 100
missing_df = pd.DataFrame({
    'Пропущено': missing_data,
    'Процент': missing_percent
})
print(missing_df[missing_df['Пропущено'] > 0])

# Анализ ключевых столбцов с пропусками
print("\nАнализ столбца middle_avg_bill:")
print(f"Заполнено: {df['middle_avg_bill'].notna().sum()} / {len(df)}")
print(f"Процент заполнения: {(df['middle_avg_bill'].notna().sum() / len(df) * 100):.2f}%")

print("\nАнализ столбца middle_coffee_cup:")
print(f"Заполнено: {df['middle_coffee_cup'].notna().sum()} / {len(df)}")
print(f"Процент заполнения: {(df['middle_coffee_cup'].notna().sum() / len(df) * 100):.2f}%")

# Создание столбца street (название улицы)
def extract_street(address):
    if pd.isna(address):
        return None
    
    # Разбиваем адрес на части
    parts = str(address).split(',')
    
    # Ищем улицу (обычно это второй элемент после номера дома)
    for part in parts:
        part = part.strip()
        # Проверяем, содержит ли часть слово "улица", "проспект", "бульвар" и т.д.
        street_keywords = ['ул.', 'улица', 'пр.', 'проспект', 'б-р', 'бульвар', 
                          'ш.', 'шоссе', 'наб.', 'набережная', 'аллея', 'пер.', 'переулок']
        
        for keyword in street_keywords:
            if keyword in part.lower():
                return part
    
    # Если не нашли, возвращаем второй элемент (обычно это улица)
    if len(parts) > 1:
        return parts[1].strip()
    
    return None

# Применяем функцию к столбцу address
df['street'] = df['address'].apply(extract_street)

# Проверяем результат
print(f"\nУдалось извлечь улицы для {df['street'].notna().sum()} / {len(df)} записей")
print("\nПримеры извлеченных улиц:")
print(df['street'].head(10))

# Создание столбца is_24_7 (круглосуточные заведения)
def check_24_7(hours):
    if pd.isna(hours):
        return False
    
    hours_str = str(hours).lower()
    
    # Проверяем различные варианты обозначения круглосуточной работы
    patterns = [
        r'24/7',
        r'круглосуточно',
        r'ежедневно.*24.*часа',
        r'24 часа',
        r'0:00.*24:00',
        r'00:00.*24:00',
        r'00:00.*00:00'
    ]
    
    for pattern in patterns:
        if re.search(pattern, hours_str):
            return True
    
    # Дополнительная проверка для формата "пн-вс 0:00-24:00"
    if re.search(r'пн.*вс.*0.*24', hours_str) or re.search(r'ежедневно.*0.*24', hours_str):
        return True
    
    return False

# Применяем функцию
df['is_24_7'] = df['hours'].apply(check_24_7)

print(f"\nКоличество круглосуточных заведений: {df['is_24_7'].sum()}")

# Преобразование типов данных
df['seats'] = pd.to_numeric(df['seats'], errors='coerce')
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

print("\nТипы данных после преобразования:")
print(df.dtypes)
```

```python
# 3.1 Распределение заведений по категориям
print("РАСПРЕДЕЛЕНИЕ ЗАВЕДЕНИЙ ПО КАТЕГОРИЯМ")
print("=" * 50)

category_counts = df['category'].value_counts()
print(f"Всего категорий: {len(category_counts)}")
print("\nТоп-10 категорий:")
print(category_counts.head(10))

# Визуализация: распределение по категориям (топ-15)
plt.figure(figsize=(14, 8))
top_categories = category_counts.head(15)
bars = plt.barh(top_categories.index[::-1], top_categories.values[::-1])
plt.xlabel('Количество заведений')
plt.title('Топ-15 категорий заведений общественного питания в Москве', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем значения на столбцы
for bar in bars:
    width = bar.get_width()
    plt.text(width + 10, bar.get_y() + bar.get_height()/2, 
             f'{int(width)}', ha='left', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# Круговой график для основных категорий
main_categories = ['кафе', 'ресторан', 'кофейня', 'бар', 'паб', 'столовая', 
                   'пиццерия', 'фастфуд', 'буфет', 'кондитерская']
main_df = df[df['category'].isin(main_categories)]

plt.figure(figsize=(10, 10))
main_counts = main_df['category'].value_counts()
plt.pie(main_counts.values, labels=main_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Распределение основных категорий заведений', fontsize=16, fontweight='bold')
plt.axis('equal')
plt.show()

# 3.2 Количество посадочных мест по категориям
print("\nАНАЛИЗ ПОСАДОЧНЫХ МЕСТ ПО КАТЕГОРИЯМ")
print("=" * 50)

# Удаляем пропуски в seats
seats_df = df.dropna(subset=['seats'])

# Группируем по категориям
seats_by_category = seats_df.groupby('category')['seats'].agg(['mean', 'median', 'count']).round(1)
seats_by_category = seats_by_category.sort_values('median', ascending=False)

print("Топ-10 категорий по медианному количеству мест:")
print(seats_by_category.head(10))

# Визуализация: boxplot по основным категориям
plt.figure(figsize=(14, 8))
main_cats_seats = seats_df[seats_df['category'].isin(main_categories)]

# Убираем выбросы для лучшей визуализации
Q1 = main_cats_seats['seats'].quantile(0.25)
Q3 = main_cats_seats['seats'].quantile(0.75)
IQR = Q3 - Q1
filtered_seats = main_cats_seats[
    (main_cats_seats['seats'] >= Q1 - 1.5*IQR) & 
    (main_cats_seats['seats'] <= Q3 + 1.5*IQR)
]

sns.boxplot(data=filtered_seats, x='seats', y='category', order=filtered_seats.groupby('category')['seats'].median().sort_values(ascending=False).index)
plt.xlabel('Количество посадочных мест')
plt.ylabel('Категория')
plt.title('Распределение посадочных мест по основным категориям', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# 3.3 Соотношение сетевых и несетевых заведений
print("\nСООТНОШЕНИЕ СЕТЕВЫХ И НЕСЕТЕВЫХ ЗАВЕДЕНИЙ")
print("=" * 50)

chain_counts = df['chain'].value_counts()
chain_percent = (chain_counts / len(df)) * 100

print(f"Сетевые заведения: {chain_counts.get(1, 0)} ({chain_percent.get(1, 0):.1f}%)")
print(f"Несетевые заведения: {chain_counts.get(0, 0)} ({chain_percent.get(0, 0):.1f}%)")

# Визуализация
plt.figure(figsize=(8, 8))
labels = ['Несетевые', 'Сетевые']
colors = ['#ff9999', '#66b3ff']
plt.pie([chain_counts.get(0, 0), chain_counts.get(1, 0)], 
        labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('Соотношение сетевых и несетевых заведений', fontsize=16, fontweight='bold')
plt.axis('equal')
plt.show()

# 3.4 Какие категории чаще являются сетевыми
print("\nКАТЕГОРИИ, ЧАЩЕ ЯВЛЯЮЩИЕСЯ СЕТЕВЫМИ")
print("=" * 50)

# Группируем по категориям и считаем долю сетевых
chain_by_category = df.groupby('category').agg(
    total=('chain', 'count'),
    chain_count=('chain', lambda x: (x == 1).sum())
)
chain_by_category['chain_percent'] = (chain_by_category['chain_count'] / chain_by_category['total'] * 100).round(1)
chain_by_category = chain_by_category.sort_values('chain_percent', ascending=False)

print("Топ-10 категорий по доле сетевых заведений:")
print(chain_by_category.head(10))

# Визуализация (только категории с достаточным количеством заведений)
filtered_chain = chain_by_category[chain_by_category['total'] >= 10].head(15)

plt.figure(figsize=(14, 8))
bars = plt.barh(filtered_chain.index[::-1], filtered_chain['chain_percent'][::-1])
plt.xlabel('Доля сетевых заведений, %')
plt.title('Категории с наибольшей долей сетевых заведений', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем значения
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height()/2, 
             f'{width:.1f}%', ha='left', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# 3.5 Топ-15 популярных сетей
print("\nТОП-15 ПОПУЛЯРНЫХ СЕТЕЙ")
print("=" * 50)

# Группируем по названиям заведений
chain_names = df[df['chain'] == 1]['name'].value_counts().head(15)

print("Топ-15 сетей по количеству заведений:")
for i, (name, count) in enumerate(chain_names.items(), 1):
    # Находим категорию для этой сети
    category = df[df['name'] == name]['category'].iloc[0] if not df[df['name'] == name].empty else 'Неизвестно'
    print(f"{i:2}. {name:40} - {count:3} заведений ({category})")

# Визуализация
plt.figure(figsize=(14, 10))
bars = plt.barh(range(len(chain_names)), chain_names.values[::-1])
plt.yticks(range(len(chain_names)), chain_names.index[::-1])
plt.xlabel('Количество заведений')
plt.title('Топ-15 популярных сетей в Москве', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем значения
for i, bar in enumerate(bars):
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height()/2, 
             f'{int(width)}', ha='left', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# 3.6 Распределение заведений по административным районам
print("\nРАСПРЕДЕЛЕНИЕ ЗАВЕДЕНИЙ ПО РАЙОНАМ")
print("=" * 50)

district_counts = df['district'].value_counts()
print(f"Всего районов: {len(district_counts)}")
print("\nТоп-10 районов по количеству заведений:")
print(district_counts.head(10))

# Визуализация: распределение по районам
plt.figure(figsize=(14, 10))
top_districts = district_counts.head(15)
bars = plt.barh(top_districts.index[::-1], top_districts.values[::-1])
plt.xlabel('Количество заведений')
plt.title('Топ-15 районов по количеству заведений общественного питания', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем значения
for bar in bars:
    width = bar.get_width()
    plt.text(width + 5, bar.get_y() + bar.get_height()/2, 
             f'{int(width)}', ha='left', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# Распределение категорий по районам (heatmap)
# Берем топ-10 районов и топ-10 категорий
top_10_districts = district_counts.head(10).index
top_10_categories = category_counts.head(10).index

# Создаем сводную таблицу
pivot_data = df[df['district'].isin(top_10_districts) & df['category'].isin(top_10_categories)]
pivot_table = pd.crosstab(pivot_data['district'], pivot_data['category'])

# Нормализуем по строкам для лучшей визуализации
pivot_table_norm = pivot_table.div(pivot_table.sum(axis=1), axis=0) * 100

plt.figure(figsize=(16, 10))
sns.heatmap(pivot_table_norm, annot=True, fmt='.1f', cmap='YlOrRd', linewidths=0.5)
plt.title('Доля категорий заведений по районам (%)', fontsize=16, fontweight='bold')
plt.xlabel('Категория')
plt.ylabel('Район')
plt.tight_layout()
plt.show()

# 3.7 Распределение средних рейтингов по категориям
print("\nРАСПРЕДЕЛЕНИЕ РЕЙТИНГОВ ПО КАТЕГОРИЯМ")
print("=" * 50)

# Удаляем пропуски в рейтингах
rating_df = df.dropna(subset=['rating'])

# Группируем по категориям
rating_by_category = rating_df.groupby('category')['rating'].agg(['mean', 'median', 'count', 'std']).round(3)
rating_by_category = rating_by_category[rating_by_category['count'] >= 10].sort_values('mean', ascending=False)

print("Топ-10 категорий по среднему рейтингу:")
print(rating_by_category.head(10))

print("\nНиз-10 категорий по среднему рейтингу:")
print(rating_by_category.tail(10))

# Визуализация: boxplot рейтингов по основным категориям
plt.figure(figsize=(14, 8))
main_cats_rating = rating_df[rating_df['category'].isin(main_categories)]

sns.boxplot(data=main_cats_rating, x='rating', y='category', 
            order=main_cats_rating.groupby('category')['rating'].median().sort_values(ascending=False).index)
plt.xlabel('Рейтинг')
plt.ylabel('Категория')
plt.title('Распределение рейтингов по основным категориям', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# 3.8 Карта с кластерами заведений
print("\nСОЗДАНИЕ КАРТЫ С КЛАСТЕРАМИ")
print("=" * 50)

# Создаем базовую карту Москвы
moscow_center = [55.7558, 37.6173]
m = folium.Map(location=moscow_center, zoom_start=11, tiles='CartoDB positron')

# Добавляем кластеры
marker_cluster = MarkerCluster().add_to(m)

# Добавляем заведения (выборочно для производительности)
sample_df = df.dropna(subset=['lat', 'lng']).sample(n=1000, random_state=42)

for idx, row in sample_df.iterrows():
    popup_text = f"""
    <b>{row['name']}</b><br>
    Категория: {row['category']}<br>
    Рейтинг: {row['rating'] if not pd.isna(row['rating']) else 'Н/Д'}<br>
    Адрес: {row['address'][:50]}...
    """
    
    folium.Marker(
        location=[row['lat'], row['lng']],
        popup=folium.Popup(popup_text, max_width=300),
        icon=folium.Icon(color='blue' if row['chain'] == 1 else 'green')
    ).add_to(marker_cluster)

print("Карта создана. Сохраняем в файл...")
m.save('moscow_restaurants_map.html')
print("Карта сохранена в файл 'moscow_restaurants_map.html'")

# 3.9 Топ-15 улиц по количеству заведений
print("\nТОП-15 УЛИЦ ПО КОЛИЧЕСТВУ ЗАВЕДЕНИЙ")
print("=" * 50)

street_counts = df['street'].value_counts().head(15)

print("Топ-15 улиц по количеству заведений:")
for i, (street, count) in enumerate(street_counts.items(), 1):
    print(f"{i:2}. {street:40} - {count:3} заведений")

# Визуализация: распределение категорий на топ-улицах
top_streets = street_counts.index
streets_data = df[df['street'].isin(top_streets)]

# Создаем сводную таблицу
street_category_pivot = pd.crosstab(streets_data['street'], streets_data['category'])

# Берем топ-5 категорий для каждой улицы
top_categories_per_street = {}
for street in top_streets:
    if street in street_category_pivot.index:
        top_cats = street_category_pivot.loc[street].nlargest(5)
        top_categories_per_street[street] = top_cats

# Визуализируем
fig, axes = plt.subplots(5, 3, figsize=(18, 20))
axes = axes.flatten()

for i, (street, cats) in enumerate(top_categories_per_street.items()):
    if i < 15:
        ax = axes[i]
        cats.plot(kind='bar', ax=ax, color='steelblue')
        ax.set_title(f'{street[:30]}...' if len(street) > 30 else street)
        ax.set_xlabel('')
        ax.set_ylabel('Количество')
        ax.tick_params(axis='x', rotation=45)
        ax.grid(axis='y', alpha=0.3)

plt.suptitle('Топ-5 категорий заведений на самых популярных улицах', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# 3.10 Улицы с одним заведением
print("\nУЛИЦЫ С ОДНИМ ЗАВЕДЕНИЕМ")
print("=" * 50)

single_street_counts = df['street'].value_counts()
single_streets = single_street_counts[single_street_counts == 1].index

print(f"Количество улиц с одним заведением: {len(single_streets)}")
print("\nПримеры заведений на таких улицах:")

single_street_df = df[df['street'].isin(single_streets)].head(10)
for idx, row in single_street_df.iterrows():
    print(f"- {row['name']} ({row['category']}) на {row['street']}")

# 3.11 Анализ средних чеков по районам
print("\nАНАЛИЗ СРЕДНИХ ЧЕКОВ ПО РАЙОНАМ")
print("=" * 50)

# Удаляем пропуски в среднем чеке
bill_df = df.dropna(subset=['middle_avg_bill'])

# Группируем по районам
bill_by_district = bill_df.groupby('district')['middle_avg_bill'].agg(['median', 'mean', 'count', 'std']).round(0)
bill_by_district = bill_by_district.sort_values('median', ascending=False)

print("Районы с самыми высокими средними чеками:")
print(bill_by_district.head(10))

print("\nРайоны с самыми низкими средними чеками:")
print(bill_by_district.tail(10))

# Визуализация
plt.figure(figsize=(14, 10))
top_bill_districts = bill_by_district.head(15)

bars = plt.barh(top_bill_districts.index[::-1], top_bill_districts['median'][::-1])
plt.xlabel('Медианный средний чек, руб.')
plt.title('Районы с самыми высокими средними чеками', fontsize=16, fontweight='bold')
plt.grid(axis='x', alpha=0.3)

# Добавляем значения
for bar in bars:
    width = bar.get_width()
    plt.text(width + 50, bar.get_y() + bar.get_height()/2, 
             f'{int(width)} ₽', ha='left', va='center', fontsize=10)

plt.tight_layout()
plt.show()

# Анализ зависимости цены от удаленности от центра
# Для простоты будем считать центром координаты Кремля
kremlin_coords = (55.7520, 37.6175)

def calculate_distance(lat, lng):
    # Упрощенный расчет расстояния (в градусах)
    return np.sqrt((lat - kremlin_coords[0])**2 + (lng - kremlin_coords[1])**2)

# Добавляем расстояние до центра
bill_df['dist_to_center'] = bill_df.apply(lambda row: calculate_distance(row['lat'], row['lng']), axis=1)

# Корреляция между расстоянием и средним чеком
correlation = bill_df['dist_to_center'].corr(bill_df['middle_avg_bill'])
print(f"\nКорреляция между расстоянием до центра и средним чеком: {correlation:.3f}")

# Визуализация зависимости
plt.figure(figsize=(12, 8))
plt.scatter(bill_df['dist_to_center'], bill_df['middle_avg_bill'], alpha=0.5, s=20)
plt.xlabel('Расстояние до центра (в градусах)')
plt.ylabel('Средний чек, руб.')
plt.title('Зависимость среднего чека от удаленности от центра', fontsize=16, fontweight='bold')
plt.grid(alpha=0.3)

# Добавляем линию тренда
z = np.polyfit(bill_df['dist_to_center'], bill_df['middle_avg_bill'], 1)
p = np.poly1d(z)
plt.plot(sorted(bill_df['dist_to_center']), p(sorted(bill_df['dist_to_center'])), 
         color='red', linewidth=2, label=f'Тренд (R²={correlation**2:.3f})')

plt.legend()
plt.tight_layout()
plt.show()
```
