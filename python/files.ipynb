```python
os.listdir(path='.')  # Список файлов и поддиректорий в указанной папке (по умолчанию текущей). Не рекурсивно.
```

```python
path = "."

os.walk(path) # Генератор для рекурсивного обхода. Возвращает кортежи (root, dirs, files) для каждой директории


for root, dirs, files in os.walk('.'):
    print(f"Директория: {root}")
    for d in dirs:
        print(f"Поддиректория: {os.path.join(root, d)}")
    for f in files:
        print(f"Файл: {os.path.join(root, f)}")
```

Копирование файлов и директорий

Модуль shutil для копирования.

1) shutil.copy(src, dst): Копирует файл (только содержимое, без метаданных).

2) shutil.copy2(src, dst): Копирует файл с метаданными (временные метки и т.д.).

```python
import os
from contextlib import contextmanager

@contextmanager
def change_dir(new_dir):
    old_dir = os.getcwd()  # Сохраняем текущую
    try:
        os.chdir(new_dir)  # Меняем
        yield  # Возвращаем контроль в блок with (без значения)
    finally:
        os.chdir(old_dir)  # Восстанавливаем в finally для гарантии

# Использование
print(os.getcwd())  # Текущая директория
with change_dir('/tmp'):
    print(os.getcwd())  # /tmp
print(os.getcwd())  # Вернулись обратно
```
