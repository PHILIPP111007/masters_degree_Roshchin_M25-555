```python
import psycopg2

conn = psycopg2.connect(dbname = 'postgres',
                        user = 'postgres', 
                        password = 'qwerty',
                        host = 'localhost')
cur = conn.cursor()
cur.execute("SELECT * FROM test")
result = cur.fetchall()
for row in result:
    print(row, row[0])
    for item in row: 
        print(item)
cur.execute("SELECT * FROM test")
r = cur.fetchone()
while r:
    for item in r:
        print(item)
    r = cur.fetchone()
conn.close()
```

```python
import psycopg2

conn = psycopg2.connect(dbname = 'postgres',
                        user = 'postgres', 
                        password = 'qwerty',
                        host = 'localhost')
cur = conn.cursor()
print("Введите 4 значения, нажимая после каждого ввода Enter:")
v1 = input()
v2 = input()
v3 = input()
v4 = input()
cur.execute("INSERT INTO test(test) VALUES(%s)", (v1,))
cur.execute("INSERT INTO test(test) VALUES(%(data)s)", {"data": v2})
cur.execute("INSERT INTO test(test) VALUES(%(v1)s), (%(v2)s)",
            {"v1": v3, "v2": v4})
conn.commit()
cur.execute("SELECT * FROM test")
result = cur.fetchall()
print(result)
conn.close()
```
