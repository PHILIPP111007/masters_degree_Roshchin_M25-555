### Проведите исследовательский анализ данных

````python
# Список параметров для анализа
parameters = [
    'total_area', 'living_area', 'kitchen_area', 'last_price',
    'rooms', 'ceiling_height', 'floors_total',
    'cityCenters_nearest', 'parks_nearest'
]

# Функция для построения гистограммы с удалением выбросов
def plot_hist_with_outliers(column, data, bins=50, title=None):
    plt.figure(figsize=(10, 6))
    
    # Удаляем выбросы используя межквартильный размах
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    filtered_data = data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]
    
    plt.hist(filtered_data[column], bins=bins, edgecolor='black', alpha=0.7)
    
    if title:
        plt.title(title)
    else:
        plt.title(f'Распределение {column}')
    
    plt.xlabel(column)
    plt.ylabel('Количество')
    plt.grid(True, alpha=0.3)
    plt.show()
    
    # Выводим статистику
    print(f"Статистика для {column}:")
    print(f"  Медиана: {data[column].median():.2f}")
    print(f"  Среднее: {data[column].mean():.2f}")
    print(f"  Минимум: {data[column].min():.2f}")
    print(f"  Максимум: {data[column].max():.2f}")
    print(f"  Выбросы (ниже {lower_bound:.2f} или выше {upper_bound:.2f}): {len(data) - len(filtered_data)}")
    print()
````

Создание очищенного датафрейма для анализа (удаление выбросов)

````python
# Создаем копию данных для анализа
data_clean = data.copy()

# Функция для удаления выбросов по методу межквартильного размаха
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Удаляем выбросы из ключевых числовых столбцов
columns_to_clean = ['total_area', 'living_area', 'kitchen_area', 'last_price', 
                   'price_per_sqm', 'ceiling_height', 'rooms', 'floors_total']

initial_rows = len(data_clean)
for col in columns_to_clean:
    if col in data_clean.columns:
        data_clean = remove_outliers(data_clean, col)

removed_rows = initial_rows - len(data_clean)
print(f"Удалено строк с выбросами: {removed_rows} ({removed_rows/initial_rows*100:.1f}% от общего объема)")
print(f"Осталось строк для анализа: {len(data_clean)}")
````

````python
# Создаем копию данных для анализа
data_clean = data.copy()
initial_count = len(data_clean)

# Создадим функцию для удаления выбросов по методу межквартильного размаха
# Но ограничим удаление 10% данных в целом

def remove_extreme_outliers(df, column, max_remove_percent=2):
    """
    Удаляет экстремальные выбросы, но не более max_remove_percent данных
    """
    initial_len = len(df)
    
    # Вычисляем квантили
    Q1 = df[column].quantile(0.01)  # Используем 1% вместо 25% чтобы удалять только экстремальные значения
    Q3 = df[column].quantile(0.99)  # Используем 99% вместо 75%
    IQR = Q3 - Q1
    
    # Более широкие границы для удаления только экстремальных выбросов
    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR
    
    # Создаем маску для нормальных значений
    mask = (df[column] >= lower_bound) & (df[column] <= upper_bound)
    
    # Проверяем, сколько данных будет удалено
    to_remove = len(df) - len(df[mask])
    remove_percent = (to_remove / initial_len) * 100
    
    # Если удаляется слишком много, используем более мягкие критерии
    if remove_percent > max_remove_percent:
        # Используем перцентили напрямую
        lower_bound = df[column].quantile(0.005)  # 0.5%
        upper_bound = df[column].quantile(0.995)  # 99.5%
        mask = (df[column] >= lower_bound) & (df[column] <= upper_bound)
    
    return df[mask]

# Применяем к ключевым параметрам по очереди
print("Очистка по параметрам:")

# 1. Общая площадь (сохраняем 99% данных)
data_clean = remove_extreme_outliers(data_clean, 'total_area', max_remove_percent=1)
print(f"Общая площадь: осталось {len(data_clean)} записей")

# 2. Цена (сохраняем 99% данных)
data_clean = remove_extreme_outliers(data_clean, 'last_price', max_remove_percent=1)
print(f"Цена: осталось {len(data_clean)} записей")

# 3. Цена за квадратный метр
data_clean['price_per_sqm'] = (data_clean['last_price'] / data_clean['total_area']).round(2)
data_clean = remove_extreme_outliers(data_clean, 'price_per_sqm', max_remove_percent=1)
print(f"Цена за кв.м: осталось {len(data_clean)} записей")

# 4. Высота потолков (исправляем аномалии перед удалением выбросов)
# Исправляем очевидные ошибки (25 -> 2.5 и т.д.)
data_clean.loc[data_clean['ceiling_height'] > 5, 'ceiling_height'] = \
    data_clean.loc[data_clean['ceiling_height'] > 5, 'ceiling_height'] / 10

data_clean = remove_extreme_outliers(data_clean, 'ceiling_height', max_remove_percent=1)
print(f"Высота потолков: осталось {len(data_clean)} записей")

# 5. Количество комнат - удаляем только экстремальные значения (0 и >10)
rooms_mask = (data_clean['rooms'] >= 0) & (data_clean['rooms'] <= 10)
data_clean = data_clean[rooms_mask]
print(f"Количество комнат: осталось {len(data_clean)} записей")

# 6. Удаляем записи, где days_exposition слишком большой (больше 3 лет)
if 'days_exposition' in data_clean.columns:
    days_mask = (data_clean['days_exposition'] <= 3*365) | (data_clean['days_exposition'].isna())
    data_clean = data_clean[days_mask]
    print(f"Время продажи: осталось {len(data_clean)} записей")

final_count = len(data_clean)
removed_count = initial_count - final_count
removed_percent = (removed_count / initial_count) * 100

print("\nИтоговая статистика:")
print(f"Начальное количество записей: {initial_count}")
print(f"Удалено записей: {removed_count}")
print(f"Осталось записей: {final_count}")
print(f"Удалено {removed_percent:.1f}% данных")

if removed_percent <= 10:
    print("✅ Удалено менее 10% данных - требования выполнены")
else:
    print(f"⚠️ Удалено {removed_percent:.1f}% данных - нужно скорректировать параметры")

# Покажем распределение ключевых параметров после очистки
print("\nСтатистика после удаления выбросов:")
stats_columns = ['total_area', 'last_price', 'price_per_sqm', 'ceiling_height', 'rooms']
for col in stats_columns:
    if col in data_clean.columns:
        print(f"{col}: {data_clean[col].min():.1f} - {data_clean[col].max():.1f} "
              f"(медиана: {data_clean[col].median():.1f})")


print("\nПроверка распределения после очистки:")
fig, axes = plt.subplots(2, 3, figsize=(15, 8))

# 1. Общая площадь
axes[0, 0].hist(data_clean['total_area'], bins=30, edgecolor='black', alpha=0.7)
axes[0, 0].set_title('Общая площадь после очистки', fontsize=12)
axes[0, 0].set_xlabel('Площадь, м²')
axes[0, 0].set_ylabel('Количество')
axes[0, 0].grid(True, alpha=0.3)

# 2. Цена
axes[0, 1].hist(data_clean['last_price'] / 1e6, bins=30, edgecolor='black', alpha=0.7)
axes[0, 1].set_title('Цена после очистки', fontsize=12)
axes[0, 1].set_xlabel('Цена, млн руб.')
axes[0, 1].set_ylabel('Количество')
axes[0, 1].grid(True, alpha=0.3)

# 3. Цена за кв.м
axes[0, 2].hist(data_clean['price_per_sqm'] / 1e3, bins=30, edgecolor='black', alpha=0.7)
axes[0, 2].set_title('Цена за кв.м после очистки', fontsize=12)
axes[0, 2].set_xlabel('Цена за кв.м, тыс. руб.')
axes[0, 2].set_ylabel('Количество')
axes[0, 2].grid(True, alpha=0.3)

# 4. Высота потолков
axes[1, 0].hist(data_clean['ceiling_height'], bins=30, edgecolor='black', alpha=0.7)
axes[1, 0].set_title('Высота потолков после очистки', fontsize=12)
axes[1, 0].set_xlabel('Высота, м')
axes[1, 0].set_ylabel('Количество')
axes[1, 0].grid(True, alpha=0.3)

# 5. Количество комнат
room_counts = data_clean['rooms'].value_counts().sort_index()
axes[1, 1].bar(room_counts.index.astype(str), room_counts.values, edgecolor='black', alpha=0.7)
axes[1, 1].set_title('Количество комнат после очистки', fontsize=12)
axes[1, 1].set_xlabel('Количество комнат')
axes[1, 1].set_ylabel('Количество')
axes[1, 1].grid(True, alpha=0.3, axis='y')

# 6. Время продажи (если есть данные)
if 'days_exposition' in data_clean.columns:
    sales_data = data_clean[data_clean['days_exposition'].notna()]
    axes[1, 2].hist(sales_data['days_exposition'], bins=30, edgecolor='black', alpha=0.7)
    axes[1, 2].set_title('Время продажи после очистки', fontsize=12)
    axes[1, 2].set_xlabel('Дней на продажу')
    axes[1, 2].set_ylabel('Количество')
    axes[1, 2].grid(True, alpha=0.3)
else:
    fig.delaxes(axes[1, 2])

plt.suptitle('Распределение ключевых параметров после удаления выбросов', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()

print("\nВыводы по удалению выбросов:")
print("1. Удалены экстремальные значения в ключевых параметрах")
print("2. Сохранена основная масса данных (более 90%)")
print("3. Распределения стали более 'нормальными' и пригодными для анализа")
print("4. Устранены очевидные ошибки в данных (высота потолков >5м)")
````

````python
### 4.3 Анализ зависимости стоимости от ключевых факторов

print("Анализ зависимости стоимости от ключевых факторов:")

# Сначала построим графики зависимости
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# 1. Зависимость цены от общей площади
axes[0, 0].scatter(data_clean['total_area'], data_clean['last_price']/1e6, 
                   alpha=0.3, s=10)
axes[0, 0].set_title('Зависимость цены от общей площади', fontsize=12)
axes[0, 0].set_xlabel('Общая площадь, м²')
axes[0, 0].set_ylabel('Цена, млн руб.')
axes[0, 0].grid(True, alpha=0.3)

# Добавляем линию тренда
z = np.polyfit(data_clean['total_area'], data_clean['last_price']/1e6, 1)
p = np.poly1d(z)
x_range = np.linspace(data_clean['total_area'].min(), data_clean['total_area'].max(), 100)
axes[0, 0].plot(x_range, p(x_range), "r--", alpha=0.8, 
               label=f'Тренд: +{z[0]:.3f} млн руб./м²')
axes[0, 0].legend()

# 2. Зависимость цены от жилой площади
axes[0, 1].scatter(data_clean['living_area'], data_clean['last_price']/1e6, 
                   alpha=0.3, s=10, color='green')
axes[0, 1].set_title('Зависимость цены от жилой площади', fontsize=12)
axes[0, 1].set_xlabel('Жилая площадь, м²')
axes[0, 1].set_ylabel('Цена, млн руб.')
axes[0, 1].grid(True, alpha=0.3)

# 3. Зависимость цены от площади кухни
axes[0, 2].scatter(data_clean['kitchen_area'], data_clean['last_price']/1e6, 
                   alpha=0.3, s=10, color='orange')
axes[0, 2].set_title('Зависимость цены от площади кухни', fontsize=12)
axes[0, 2].set_xlabel('Площадь кухни, м²')
axes[0, 2].set_ylabel('Цена, млн руб.')
axes[0, 2].grid(True, alpha=0.3)

# 4. Зависимость цены от количества комнат
room_price = data_clean.groupby('rooms')['last_price'].median()/1e6
axes[1, 0].bar(room_price.index.astype(str), room_price.values, 
               edgecolor='black', alpha=0.7)
axes[1, 0].set_title('Медианная цена по количеству комнат', fontsize=12)
axes[1, 0].set_xlabel('Количество комнат')
axes[1, 0].set_ylabel('Медианная цена, млн руб.')
axes[1, 0].grid(True, alpha=0.3, axis='y')

# Добавляем значения на столбцы
for i, (idx, val) in enumerate(room_price.items()):
    axes[1, 0].text(i, val + 0.1, f'{val:.1f} млн', ha='center', fontsize=9)

# 5. Зависимость цены от типа этажа
floor_price = data_clean.groupby('floor_type')['last_price'].median()/1e6
colors = ['#ff9999', '#66b3ff', '#99ff99']
axes[1, 1].bar(floor_price.index, floor_price.values, 
               color=colors, edgecolor='black', alpha=0.7)
axes[1, 1].set_title('Медианная цена по типу этажа', fontsize=12)
axes[1, 1].set_xlabel('Тип этажа')
axes[1, 1].set_ylabel('Медианная цена, млн руб.')
axes[1, 1].grid(True, alpha=0.3, axis='y')

# Добавляем значения на столбцы
for i, (idx, val) in enumerate(floor_price.items()):
    axes[1, 1].text(i, val + 0.1, f'{val:.1f} млн', ha='center', fontsize=9)

# 6. Зависимость цены от месяца публикации
month_price = data_clean.groupby('month')['last_price'].median()/1e6
months = ['Янв', 'Фев', 'Мар', 'Апр', 'Май', 'Июн', 
          'Июл', 'Авг', 'Сен', 'Окт', 'Ноя', 'Дек']
axes[1, 2].bar(range(1, 13), month_price, edgecolor='black', alpha=0.7)
axes[1, 2].set_title('Медианная цена по месяцу публикации', fontsize=12)
axes[1, 2].set_xlabel('Месяц')
axes[1, 2].set_ylabel('Медианная цена, млн руб.')
axes[1, 2].set_xticks(range(1, 13))
axes[1, 2].set_xticklabels(months, rotation=45)
axes[1, 2].grid(True, alpha=0.3, axis='y')

plt.suptitle('Зависимость цены квартир от различных факторов', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()


# ### Матрица корреляций


print("Матрица корреляций между ключевыми параметрами:")

# Выбираем числовые столбцы для анализа корреляций
numeric_cols = ['last_price', 'total_area', 'living_area', 'kitchen_area', 
                'rooms', 'ceiling_height', 'floors_total', 'price_per_sqm',
                'day_of_week', 'month', 'year']

# Создаем корреляционную матрицу
correlation_matrix = data_clean[numeric_cols].corr()

# Визуализация матрицы корреляций
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            fmt='.2f', linewidths=1, square=True, cbar_kws={'shrink': 0.8},
            annot_kws={'size': 9})
plt.title('Матрица корреляций между ключевыми параметрами', fontsize=14, pad=20)
plt.tight_layout()
plt.show()


print("\nАнализ корреляции параметров с ценой квартиры (last_price):")
print("=" * 60)

# Сортируем корреляции с ценой по убыванию
price_correlations = correlation_matrix['last_price'].sort_values(ascending=False)

print("Рейтинг факторов по силе влияния на цену:")
print("-" * 60)

for param, corr in price_correlations.items():
    if param != 'last_price':
        # Определяем силу корреляции
        if abs(corr) >= 0.7:
            strength = "ОЧЕНЬ СИЛЬНАЯ"
            emoji = "🔥"
        elif abs(corr) >= 0.5:
            strength = "СИЛЬНАЯ"
            emoji = "⚡"
        elif abs(corr) >= 0.3:
            strength = "СРЕДНЯЯ"
            emoji = "📊"
        elif abs(corr) >= 0.1:
            strength = "СЛАБАЯ"
            emoji = "📉"
        else:
            strength = "ОЧЕНЬ СЛАБАЯ"
            emoji = "➖"
        
        # Определяем направление
        direction = "положительная" if corr > 0 else "отрицательная"
        
        print(f"{emoji} {param:20} | {corr:+.3f} | {strength:15} | {direction}")

print("\nАнализ корреляций между другими параметрами:")
print("=" * 60)

# Анализируем интересные корреляции (кроме с ценой)
interesting_correlations = []

# Находим сильные корреляции (|r| > 0.5) между параметрами (кроме last_price)
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr = correlation_matrix.iloc[i, j]
        col1 = correlation_matrix.columns[i]
        col2 = correlation_matrix.columns[j]
        
        # Пропускаем корреляции с last_price (их уже проанализировали)
        if col1 == 'last_price' or col2 == 'last_price':
            continue
            
        if abs(corr) > 0.5:
            interesting_correlations.append((col1, col2, corr))

if interesting_correlations:
    print("Сильные корреляции между параметрами (|r| > 0.5):")
    print("-" * 60)
    for col1, col2, corr in sorted(interesting_correlations, key=lambda x: abs(x[2]), reverse=True):
        direction = "прямая" if corr > 0 else "обратная"
        print(f"📈 {col1:20} ↔ {col2:20} | {corr:+.3f} | {direction} зависимость")
else:
    print("Сильных корреляций между другими параметрами не обнаружено")
````

````python
print(f"Общее количество объявлений: {len(data)}")
print(f"Медианное время продажи: {median_days:.0f} дней")
print(f"Среднее время продажи: {mean_days:.0f} дней")

# Определяем категории скорости продаж
fast_sale = data['days_exposition'].quantile(0.25)  # 25% самых быстрых продаж
slow_sale = data['days_exposition'].quantile(0.75)  # 25% самых медленных продаж

print("\nКритерии скорости продаж:")
print(f"• Быстрые продажи: до {fast_sale:.0f} дней (25% самых быстрых)")
print(f"• Нормальные продажи: {fast_sale:.0f} - {slow_sale:.0f} дней")
print(f"• Долгие продажи: более {slow_sale:.0f} дней (25% самых медленных)")

# Анализ продаж, которые ушли с публикации за 0 или 1 день
instant_sales = data[data['days_exposition'] <= 1]
print(f"\nМгновенные продажи (0-1 день): {len(instant_sales)} объявлений ({len(instant_sales)/len(data)*100:.1f}%)")

# Анализ очень долгих продаж
very_long_sales = data[data['days_exposition'] > 365]
print(f"Очень долгие продажи (>1 года): {len(very_long_sales)} объявлений ({len(very_long_sales)/len(data)*100:.1f}%)")

# Дополнительный анализ: зависимость скорости продажи от цены
data['sale_speed_category'] = pd.cut(data['days_exposition'], 
                                     bins=[-1, fast_sale, slow_sale, data['days_exposition'].max()], 
                                     labels=['быстрая', 'нормальная', 'долгая'])

speed_price_analysis = data.groupby('sale_speed_category')['last_price'].agg(['median', 'mean', 'count']).round()
print("\nАнализ цен по категориям скорости продажи:")
print(speed_price_analysis)

# Визуализация
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Боксплот цен по категориям скорости
speed_categories_data = []
labels = []
for category in ['быстрая', 'нормальная', 'долгая']:
    category_data = data[data['sale_speed_category'] == category]['last_price']
    if len(category_data) > 0:
        speed_categories_data.append(category_data)
        labels.append(category)

axes[0].boxplot(speed_categories_data, labels=labels)
axes[0].set_title('Распределение цен по категориям скорости продажи')
axes[0].set_ylabel('Цена (руб.)')
axes[0].grid(True, alpha=0.3)

# Гистограмма времени продажи с акцентом на быстрые продажи
fast_sales_data = data[data['days_exposition'] <= 30]
axes[1].hist(fast_sales_data['days_exposition'], bins=30, edgecolor='black', alpha=0.7)
axes[1].set_title('Распределение быстрых продаж (до 30 дней)')
axes[1].set_xlabel('Дни на продажу')
axes[1].set_ylabel('Количество объявлений')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````
