### 3. Глубокая очистка и преобразование данных

3.1 Кодировка стран в ALPHA-3 коды

```python
# Создаем словарь для кодировки стран
country_codes = {
    'Германия': 'DEU',
    'Россия': 'RUS',
    'Казахстан': 'KAZ',
    'США': 'USA',
    'Япония': 'JPN',
    'Корея': 'KOR',
    'Чехия': 'CZE',
    'Франция': 'FRA',
    'Италия': 'ITA',
    'Великобритания': 'GBR',
    'Швеция': 'SWE',
    'Китай': 'CHN',
    'Индия': 'IND',
    'Бразилия': 'BRA',
    'Мексика': 'MEX',
    'Испания': 'ESP',
    'Польша': 'POL',
    'Турция': 'TUR',
    'Иран': 'IRN',
    'Узбекистан': 'UZB',
    'Беларусь': 'BLR',
    'Украина': 'UKR',
    'Армения': 'ARM',
    'Азербайджан': 'AZE',
    'Кыргызстан': 'KGZ',
    'Таджикистан': 'TJK',
    'Туркменистан': 'TKM'
}

print("Уникальные страны в данных:")
print(df['country'].unique()[:20])

# Функция для кодировки стран
def encode_country(country_name):
    if pd.isna(country_name):
        return 'UNK'
    
    country_name = str(country_name).strip()
    
    # Проверяем полные названия
    for country, code in country_codes.items():
        if country in country_name:
            return code
    
    # Если страна не найдена в словаре, возвращаем первые 3 буквы
    return country_name[:3].upper()

# Применяем кодировку
df['country_code'] = df['country'].apply(encode_country)

print("\nПримеры кодировки стран:")
print(df[['country', 'country_code']].head(10))

print("\nРаспределение по странам после кодировки:")
print(df['country_code'].value_counts().head(15))
```

#### 3.2 Кодировка вида топлива

```python
print("Уникальные виды топлива:")
print(df['fuel_type'].unique())

# Создаем словарь для кодировки топлива
fuel_codes = {
    'Бензин': 'F',
    'Дизель': 'D',
    'Электро': 'E',
    'Гибрид': 'HYB',
    'Газ': 'G',
    'Газ/Бензин': 'G/F',
    'Гибрид (бензин)': 'HYB',
    'Гибрид (дизель)': 'HYB'
}

# Функция для кодировки топлива
def encode_fuel(fuel_name):
    if pd.isna(fuel_name):
        return 'UNK'
    
    fuel_name = str(fuel_name).strip()
    
    for fuel, code in fuel_codes.items():
        if fuel in fuel_name:
            return code
    
    # Если не нашли точное совпадение, ищем частичное
    if 'бензин' in fuel_name.lower():
        return 'F'
    elif 'дизель' in fuel_name.lower():
        return 'D'
    elif 'электр' in fuel_name.lower():
        return 'E'
    elif 'гибр' in fuel_name.lower():
        return 'HYB'
    elif 'газ' in fuel_name.lower():
        return 'G'
    else:
        return 'OTHER'

df['fuel_code'] = df['fuel_type'].apply(encode_fuel)

print("\nПримеры кодировки топлива:")
print(df[['fuel_type', 'fuel_code']].head(10))

print("\nРаспределение по видам топлива после кодировки:")
print(df['fuel_code'].value_counts())
```

#### 3.3 Стандартизация типа привода

```python
print("Уникальные типы привода:")
print(df['drive_type'].unique())

# Создаем словарь для стандартизации привода
drive_standardization = {
    'передний': 'FWD',
    'передний привод': 'FWD',
    'задний': 'RWD',
    'задний привод': 'RWD',
    'полный': '4WD',
    'полный привод': '4WD',
    '4wd': '4WD',
    '4x4': '4WD',
    'awd': 'AWD',
    'fwd': 'FWD',
    'rwd': 'RWD',
    'part-time 4wd': '4WD',
    'full-time 4wd': '4WD'
}

def standardize_drive(drive):
    if pd.isna(drive):
        return 'UNK'
    
    drive_str = str(drive).strip().lower()
    
    for key, value in drive_standardization.items():
        if key in drive_str:
            return value
    
    # Проверяем частичные совпадения
    if '4' in drive_str or 'полн' in drive_str or 'awd' in drive_str:
        return '4WD'
    elif 'задн' in drive_str or 'rwd' in drive_str:
        return 'RWD'
    elif 'передн' in drive_str or 'fwd' in drive_str:
        return 'FWD'
    else:
        return 'OTHER'

df['drive_type_std'] = df['drive_type'].apply(standardize_drive)

print("\nПримеры стандартизации привода:")
print(df[['drive_type', 'drive_type_std']].head(10))

print("\nРаспределение по типам привода после стандартизации:")
print(df['drive_type_std'].value_counts())
```

#### 3.5 Создание столбца с датой продажи

```python
# Создаем словарь для месяцев
month_dict = {
    'Январь': 1, 'Февраль': 2, 'Март': 3, 'Апрель': 4,
    'Май': 5, 'Июнь': 6, 'Июль': 7, 'Август': 8,
    'Сентябрь': 9, 'Октябрь': 10, 'Ноябрь': 11, 'Декабрь': 12
}

# Функция для создания даты (последний день месяца)
def create_sale_date(year, month_name):
    month_num = month_dict.get(str(month_name).strip(), 1)
    
    # Определяем последний день месяца
    if month_num in [1, 3, 5, 7, 8, 10, 12]:
        last_day = 31
    elif month_num in [4, 6, 9, 11]:
        last_day = 30
    else:  # Февраль
        # Для простоты считаем 28 дней (2019 не високосный)
        last_day = 28
    
    return f"{year}-{month_num:02d}-{last_day:02d}"

# Создаем столбец с датой
df['sale_date'] = df.apply(lambda row: create_sale_date(row['year'], row['month']), axis=1)

print("Примеры созданных дат:")
print(df[['year', 'month', 'sale_date']].head(10))

# Удаляем исходные столбцы год и месяц
# df = df.drop(columns=['year', 'month'])
```

```python
print("\n=== ОБРАБОТКА НАЗВАНИЙ КОМПАНИЙ ===")
print("Уникальные названия компаний:")
print(df['company'].unique()[:20])

# Создаем словарь для стандартизации названий компаний
company_standardization = {
    'Mercur Auto': 'Меркур Авто',
    'Mercur Auto ': 'Меркур Авто',
    'Mercur Auto Алматы': 'Меркур Авто',
    'Mercur Auto Астана': 'Меркур Авто',
    'Mercur Auto Караганда': 'Меркур Авто',
    'Mercur Auto Шымкент': 'Меркур Авто',
    'Азия Авто': 'Азия Авто',
    'Азия Авто ': 'Азия Авто',
    'Asia Auto': 'Азия Авто',
    'Toyota': 'Тойота Центр',
    'Toyota Center': 'Тойота Центр',
    'Тойота Центр': 'Тойота Центр',
    'БИПЭК АВТО': 'БИПЭК АВТО',
    'Bipek Auto': 'БИПЭК АВТО',
    'Astana Motors': 'Астана Моторс',
    'Astana Motors ': 'Астана Моторс',
    'Астана Моторс': 'Астана Моторс'
}

def standardize_company(company_name):
    if pd.isna(company_name):
        return 'НЕИЗВЕСТНО'
    
    company_str = str(company_name).strip()
    
    # Проверяем точные совпадения
    for key, value in company_standardization.items():
        if key.lower() == company_str.lower():
            return value
    
    # Проверяем частичные совпадения
    for key, value in company_standardization.items():
        if key.lower() in company_str.lower():
            return value
    
    return company_str

df['company_std'] = df['company'].apply(standardize_company)

print("\nПримеры стандартизации компаний:")
print(df[['company', 'company_std']].head(10))

print("\nТоп-10 компаний после стандартизации:")
print(df['company_std'].value_counts().head(10))
```

#### 3.7 Сложная обработка объема двигателя

```python
print("\n=== ОБРАБОТКА ОБЪЕМА ДВИГАТЕЛЯ ===")
print("Уникальные значения объема двигателя (первые 30):")
print(df['engine_volume'].dropna().unique()[:30])

def clean_engine_volume(value):
    if pd.isna(value):
        return np.nan
    
    value_str = str(value).strip()
    
    # Удаляем литеру L
    value_str = value_str.replace('L', '').replace('l', '')
    
    # Заменяем запятые на точки
    value_str = value_str.replace(',', '.')
    
    # Удаляем лишние символы
    value_str = ''.join(c for c in value_str if c.isdigit() or c == '.')
    
    # Если после очистки пустая строка
    if not value_str:
        return np.nan
    
    try:
        volume = float(value_str)
        
        # Исправляем очевидные ошибки
        if volume > 20:  # Скорее всего ошибка (например, 400 вместо 4.0)
            volume = volume / 100
        
        # Проверяем на аномалии
        if volume < 0.5 or volume > 8.0:
            # Проверяем специальные случаи
            if volume == 21.6 or volume == 20.6:  # Chevrolet Niva
                volume = 1.7
            elif volume == 100:  # Электромобили
                volume = 0.0
            else:
                return np.nan
        
        return round(volume, 1)
    except:
        return np.nan

df['engine_volume_clean'] = df['engine_volume'].apply(clean_engine_volume)

print("\nСтатистика по объему двигателя после очистки:")
print(f"Пропуски: {df['engine_volume_clean'].isnull().sum()} ({df['engine_volume_clean'].isnull().sum()/len(df)*100:.1f}%)")
print(f"Минимальный объем: {df['engine_volume_clean'].min()}")
print(f"Максимальный объем: {df['engine_volume_clean'].max()}")
print(f"Медианный объем: {df['engine_volume_clean'].median()}")

print("\nРаспределение объемов двигателя:")
print(df['engine_volume_clean'].value_counts().sort_index().head(20))
```

#### 3.8 Нормализация регистра для региона и области

```python
print("\n=== НОРМАЛИЗАЦИЯ РЕГИОНА И ОБЛАСТИ ===")

def normalize_region_name(name):
    if pd.isna(name):
        return name
    
    name_str = str(name).strip()
    
    # Удаляем префиксы
    name_str = name_str.replace('г.', '').replace('г ', '').strip()
    
    # Приводим к формату: каждое слово с заглавной буквы
    words = name_str.split()
    normalized_words = []
    
    for word in words:
        if word.lower() in ['и', 'в', 'на', 'по', 'с', 'у', 'о', 'об', 'от']:
            normalized_words.append(word.lower())
        else:
            normalized_words.append(word.capitalize())
    
    return ' '.join(normalized_words)

df['region_clean'] = df['region'].apply(normalize_region_name)
df['area_clean'] = df['area'].apply(normalize_region_name)

print("Примеры нормализации региона:")
print(df[['region', 'region_clean']].head(10))

print("\nПримеры нормализации области:")
print(df[['area', 'area_clean']].head(10))

print("\nТоп-10 регионов после нормализации:")
print(df['region_clean'].value_counts().head(10))
```

#### 3.9 Стандартизация типа трансмиссии

```python
print("\n=== СТАНДАРТИЗАЦИЯ ТИПА ТРАНСМИССИИ ===")
print("Уникальные типы трансмиссии:")
print(df['transmission'].dropna().unique()[:20])

def standardize_transmission(trans):
    if pd.isna(trans):
        return 'НЕИЗВЕСТНО'
    
    trans_str = str(trans).strip().lower()
    
    # Автоматическая коробка
    auto_keywords = ['автомат', 'auto', 'automatic', 's-tronic', 'tiptronic', 
                     'dsg', 'at', 'а', 'akpp', 'вариатор', 'cvt', 'робот']
    
    # Механическая коробка
    manual_keywords = ['механика', 'manual', 'мех', 'мкпп', 'mt', 'м', 'механическая']
    
    for keyword in auto_keywords:
        if keyword in trans_str:
            return 'Автомат'
    
    for keyword in manual_keywords:
        if keyword in trans_str:
            return 'Механика'
    
    return 'НЕИЗВЕСТНО'

df['transmission_std'] = df['transmission'].apply(standardize_transmission)

print("\nПримеры стандартизации трансмиссии:")
print(df[['transmission', 'transmission_std']].head(10))

print("\nРаспределение по типам трансмиссии:")
print(df['transmission_std'].value_counts())
```

#### 3.10 Финальные преобразования типов данных

```python
print("\n=== ФИНАЛЬНЫЕ ПРЕОБРАЗОВАНИЯ ТИПОВ ДАННЫХ ===")

# Преобразуем дату в datetime
df['sale_date'] = pd.to_datetime(df['sale_date'], format='%Y-%m-%d', errors='coerce')

# Преобразуем год выпуска в числовой тип
def clean_production_year(year):
    if pd.isna(year):
        return np.nan
    
    year_str = str(year).strip()
    
    # Извлекаем только цифры
    digits = ''.join(c for c in year_str if c.isdigit())
    
    if len(digits) == 4:
        year_int = int(digits)
        if 1900 <= year_int <= 2020:
            return year_int
    
    return np.nan

df['production_year_clean'] = df['production_year'].apply(clean_production_year)

# Преобразуем категориальные столбцы
categorical_columns = ['fuel_code', 'transmission_std', 'drive_type_std', 
                       'segment_2013', 'class_2013', 'company_std']

for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].astype('category')

print("Типы данных после преобразований:")
print(df.dtypes)

print("\nПример финальных данных:")
print(df.head())
```

```python
# ### 5. Исследовательский анализ данных (EDA)

print("\n" + "="*80)
print("ИССЛЕДОВАТЕЛЬСКИЙ АНАЛИЗ ДАННЫХ (EDA)")
print("="*80)

# #### 5.1 Основные статистики

print("\n=== ОСНОВНЫЕ СТАТИСТИКИ ===")

# Общие показатели
total_sales = df_final['sale_amount_usd'].sum()
total_units = df_final['quantity'].sum()
avg_price = df_final['price_usd'].mean()
avg_transaction = df_final['sale_amount_usd'].mean()

print(f"Общий объем продаж: ${total_sales:,.2f}")
print(f"Общее количество проданных единиц: {total_units:,.0f}")
print(f"Средняя цена автомобиля: ${avg_price:,.2f}")
print(f"Средняя сумма транзакции: ${avg_transaction:,.2f}")
print(f"Количество уникальных сделок: {len(df_final):,}")

# Анализ по месяцам
df_final['sale_month'] = df_final['sale_date'].dt.month
monthly_sales = df_final.groupby('sale_month')['sale_amount_usd'].sum()

print("\nОбъем продаж по месяцам:")
for month, sales in monthly_sales.items():
    print(f"  Месяц {month}: ${sales:,.2f}")

# #### 5.2 Распределение основных переменных

print("\n=== РАСПРЕДЕЛЕНИЕ ОСНОВНЫХ ПЕРЕМЕННЫХ ===")

# Создаем графики распределения
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# 1. Распределение цен
axes[0, 0].hist(df_final['price_usd'].dropna(), bins=50, edgecolor='black', alpha=0.7)
axes[0, 0].set_title('Распределение цен автомобилей')
axes[0, 0].set_xlabel('Цена, USD')
axes[0, 0].set_ylabel('Количество')
axes[0, 0].axvline(df_final['price_usd'].median(), color='red', linestyle='--', label=f'Медиана: ${df_final["price_usd"].median():,.0f}')
axes[0, 0].legend()

# 2. Распределение объема двигателя
axes[0, 1].hist(df_final['engine_volume_clean'].dropna(), bins=30, edgecolor='black', alpha=0.7, color='green')
axes[0, 1].set_title('Распределение объема двигателя')
axes[0, 1].set_xlabel('Объем двигателя, л')
axes[0, 1].set_ylabel('Количество')

# 3. Распределение года выпуска
axes[0, 2].hist(df_final['production_year_clean'].dropna(), bins=20, edgecolor='black', alpha=0.7, color='orange')
axes[0, 2].set_title('Распределение года выпуска')
axes[0, 2].set_xlabel('Год выпуска')
axes[0, 2].set_ylabel('Количество')

# 4. Распределение по типам топлива
fuel_counts = df_final['fuel_code'].value_counts()
axes[1, 0].bar(fuel_counts.index.astype(str), fuel_counts.values, alpha=0.7, edgecolor='black')
axes[1, 0].set_title('Распределение по типам топлива')
axes[1, 0].set_xlabel('Тип топлива')
axes[1, 0].set_ylabel('Количество')
axes[1, 0].tick_params(axis='x', rotation=45)

# 5. Распределение по типам трансмиссии
transmission_counts = df_final['transmission_std'].value_counts()
axes[1, 1].bar(transmission_counts.index.astype(str), transmission_counts.values, alpha=0.7, edgecolor='black', color='purple')
axes[1, 1].set_title('Распределение по типам трансмиссии')
axes[1, 1].set_xlabel('Тип трансмиссии')
axes[1, 1].set_ylabel('Количество')

# 6. Распределение по типам привода
drive_counts = df_final['drive_type_std'].value_counts()
axes[1, 2].bar(drive_counts.index.astype(str), drive_counts.values, alpha=0.7, edgecolor='black', color='brown')
axes[1, 2].set_title('Распределение по типам привода')
axes[1, 2].set_xlabel('Тип привода')
axes[1, 2].set_ylabel('Количество')
axes[1, 2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# #### 5.3 Анализ продаж по брендам

print("\n=== АНАЛИЗ ПРОДАЖ ПО БРЕНДАМ ===")

# Топ-10 брендов по количеству продаж
brand_sales_count = df_final['brand'].value_counts().head(10)
brand_sales_amount = df_final.groupby('brand')['sale_amount_usd'].sum().sort_values(ascending=False).head(10)

print("Топ-10 брендов по количеству проданных автомобилей:")
for brand, count in brand_sales_count.items():
    print(f"  {brand}: {count:,} ед.")

print("\nТоп-10 брендов по объему продаж (USD):")
for brand, amount in brand_sales_amount.items():
    print(f"  {brand}: ${amount:,.2f}")

# Визуализация
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

axes[0].barh(brand_sales_count.index, brand_sales_count.values, alpha=0.7, edgecolor='black')
axes[0].set_title('Топ-10 брендов по количеству продаж')
axes[0].set_xlabel('Количество проданных автомобилей')

axes[1].barh(brand_sales_amount.index, brand_sales_amount.values, alpha=0.7, edgecolor='black', color='green')
axes[1].set_title('Топ-10 брендов по объему продаж')
axes[1].set_xlabel('Объем продаж, USD')

plt.tight_layout()
plt.show()

# #### 5.4 Анализ продаж по регионам

print("\n=== АНАЛИЗ ПРОДАЖ ПО РЕГИОНАМ ===")

region_sales = df_final.groupby('region_clean').agg({
    'sale_amount_usd': 'sum',
    'quantity': 'sum',
    'price_usd': 'mean'
}).sort_values('sale_amount_usd', ascending=False).head(10)

print("Топ-10 регионов по объему продаж:")
print(region_sales)

# #### 5.5 Анализ продаж по сегментам и классам

print("\n=== АНАЛИЗ ПРОДАЖ ПО СЕГМЕНТАМ И КЛАССАМ ===")

segment_analysis = df_final.groupby('segment_2013').agg({
    'sale_amount_usd': 'sum',
    'quantity': 'sum',
    'price_usd': 'mean'
}).sort_values('sale_amount_usd', ascending=False)

print("Продажи по сегментам:")
print(segment_analysis)

class_analysis = df_final.groupby('class_2013').agg({
    'sale_amount_usd': 'sum',
    'quantity': 'sum',
    'price_usd': 'mean'
}).sort_values('sale_amount_usd', ascending=False)

print("\nПродажи по классам:")
print(class_analysis)


# ### 6. Анализ позиции "Меркур Авто"

print("\n" + "="*80)
print("АНАЛИЗ ПОЗИЦИИ 'МЕРКУР АВТО'")
print("="*80)

# Выделяем данные по Меркур Авто
mercur_data = df_final[df_final['company_std'] == 'Меркур Авто']
total_market = df_final

print(f"Количество сделок Меркур Авто: {len(mercur_data):,}")
print(f"Доля Меркур Авто на рынке по количеству сделок: {len(mercur_data)/len(total_market)*100:.2f}%")

# Основные показатели Меркур Авто
mercur_sales = mercur_data['sale_amount_usd'].sum()
market_sales = total_market['sale_amount_usd'].sum()

print(f"\nОбъем продаж Меркур Авто: ${mercur_sales:,.2f}")
print(f"Общий объем рынка: ${market_sales:,.2f}")
print(f"Доля Меркур Авто по объему продаж: {mercur_sales/market_sales*100:.2f}%")

# #### 6.1 Анализ брендов в портфеле Меркур Авто

print("\n=== БРЕНДЫ В ПОРТФЕЛЕ МЕРКУР АВТО ===")

mercur_brands = mercur_data['brand'].value_counts()
print("Бренды, представленные у Меркур Авто:")
for brand, count in mercur_brands.items():
    percentage = (count / len(mercur_data)) * 100
    print(f"  {brand}: {count} сделок ({percentage:.1f}%)")

# Сравнение с общим рынком
market_brands = total_market['brand'].value_counts().head(10)

print("\nСравнение с топ-10 брендами на рынке:")
for brand in mercur_brands.index:
    if brand in market_brands.index:
        mercur_share = mercur_brands[brand]
        market_share = market_brands[brand]
        percentage = (mercur_share / market_share) * 100
        print(f"  {brand}: {mercur_share}/{market_share} ({percentage:.1f}% от рынка)")

# #### 6.2 Анализ по сегментам

print("\n=== АНАЛИЗ ПО СЕГМЕНТАМ ===")

mercur_segments = mercur_data['segment_2013'].value_counts()
market_segments = total_market['segment_2013'].value_counts()

print("Распределение по сегментам (Меркур Авто):")
for segment, count in mercur_segments.items():
    percentage = (count / len(mercur_data)) * 100
    print(f"  {segment}: {count} ({percentage:.1f}%)")

print("\nСравнение с общим рынком:")
for segment in mercur_segments.index:
    if segment in market_segments.index:
        mercur_count = mercur_segments[segment]
        market_count = market_segments[segment]
        segment_share = (mercur_count / market_count) * 100
        print(f"  {segment}: {mercur_count}/{market_count} ({segment_share:.1f}% от рынка)")

# #### 6.3 Анализ по регионам

print("\n=== АНАЛИЗ ПО РЕГИОНАМ ===")

mercur_regions = mercur_data['region_clean'].value_counts().head(5)
market_regions = total_market['region_clean'].value_counts().head(5)

print("Топ-5 регионов для Меркур Авто:")
for region, count in mercur_regions.items():
    percentage = (count / len(mercur_data)) * 100
    print(f"  {region}: {count} сделок ({percentage:.1f}%)")

print("\nСравнение с топ-5 регионов на рынке:")
for i, (region, market_count) in enumerate(market_regions.items()):
    mercur_count = mercur_regions.get(region, 0)
    if mercur_count > 0:
        region_share = (mercur_count / market_count) * 100
        print(f"  {region}: {mercur_count}/{market_count} ({region_share:.1f}% от рынка)")

# #### 6.4 Финансовые показатели

print("\n=== ФИНАНСОВЫЕ ПОКАЗАТЕЛИ ===")

# Средний чек
mercur_avg_check = mercur_data['price_usd'].mean()
market_avg_check = total_market['price_usd'].mean()

print(f"Средний чек Меркур Авто: ${mercur_avg_check:,.2f}")
print(f"Средний чек по рынку: ${market_avg_check:,.2f}")
print(f"Разница: {((mercur_avg_check - market_avg_check) / market_avg_check * 100):+.1f}%")

# Средняя сумма сделки
mercur_avg_sale = mercur_data['sale_amount_usd'].mean()
market_avg_sale = total_market['sale_amount_usd'].mean()

print(f"\nСредняя сумма сделки Меркур Авто: ${mercur_avg_sale:,.2f}")
print(f"Средняя сумма сделки по рынку: ${market_avg_sale:,.2f}")
print(f"Разница: {((mercur_avg_sale - market_avg_sale) / market_avg_sale * 100):+.1f}%")

# #### 6.5 Визуализация позиции Меркур Авто

fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# 1. Доля на рынке
labels = ['Меркур Авто', 'Остальной рынок']
sizes = [mercur_sales, market_sales - mercur_sales]
colors = ['#ff9999', '#66b3ff']
axes[0, 0].pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
axes[0, 0].set_title('Доля Меркур Авто в общем объеме продаж')

# 2. Топ бренды Меркур Авто
top_mercur_brands = mercur_brands.head(5)
axes[0, 1].bar(top_mercur_brands.index, top_mercur_brands.values, alpha=0.7, edgecolor='black', color='green')
axes[0, 1].set_title('Топ-5 брендов Меркур Авто')
axes[0, 1].set_ylabel('Количество сделок')
axes[0, 1].tick_params(axis='x', rotation=45)

# 3. Распределение по сегментам
segments_to_plot = mercur_segments.head(5)
axes[1, 0].bar(segments_to_plot.index.astype(str), segments_to_plot.values, alpha=0.7, edgecolor='black', color='orange')
axes[1, 0].set_title('Распределение сделок Меркур Авто по сегментам')
axes[1, 0].set_ylabel('Количество сделок')
axes[1, 0].tick_params(axis='x', rotation=45)

# 4. Сравнение среднего чека
avg_checks = [mercur_avg_check, market_avg_check]
check_labels = ['Меркур Авто', 'Рынок']
axes[1, 1].bar(check_labels, avg_checks, alpha=0.7, edgecolor='black', color=['#ff9999', '#66b3ff'])
axes[1, 1].set_title('Сравнение среднего чека')
axes[1, 1].set_ylabel('Средняя цена, USD')

# Добавляем значения на столбцы
for i, v in enumerate(avg_checks):
    axes[1, 1].text(i, v + 1000, f'${v:,.0f}', ha='center', va='bottom')

plt.tight_layout()
plt.show()


# ### 7. Ключевые выводы и рекомендации

print("\n" + "="*80)
print("КЛЮЧЕВЫЕ ВЫВОДЫ И РЕКОМЕНДАЦИИ")
print("="*80)

print("\n1. ОБЩАЯ КАРТИНА РЫНКА:")
print(f"   • Всего сделок за 9 месяцев: {len(df_final):,}")
print(f"   • Общий объем продаж: ${market_sales:,.2f}")
print(f"   • Средний чек по рынку: ${market_avg_check:,.2f}")
print(f"   • Самые популярные бренды: {', '.join(brand_sales_count.head(3).index.tolist())}")

print("\n2. ПОЗИЦИЯ 'МЕРКУР АВТО':")
print(f"   • Доля на рынке: {len(mercur_data)/len(total_market)*100:.2f}% по количеству, {mercur_sales/market_sales*100:.2f}% по объему")
print(f"   • Средний чек: ${mercur_avg_check:,.2f} ({((mercur_avg_check - market_avg_check)/market_avg_check*100):+.1f}% к рынку)")
print(f"   • Основные бренды: {', '.join(mercur_brands.head(3).index.tolist())}")
print(f"   • Ключевые регионы: {', '.join(mercur_regions.head(3).index.tolist())}")

print("\n3. СИЛЬНЫЕ СТОРОНЫ:")
print("   • Выше среднего чека по рынку")
print("   • Работа с премиальными брендами (Audi)")
print("   • Присутствие в ключевых регионах (Алматы, Нур-Султан)")

print("\n4. ВОЗМОЖНОСТИ ДЛЯ РОСТА:")
print("   • Расширение портфеля брендов")
print("   • Увеличение присутствия в регионах с высокой долей рынка")
print("   • Развитие сегментов с высокой маржинальностью")
print("   • Улучшение работы с корпоративными клиентами")

print("\n5. РЕКОМЕНДАЦИИ:")
print("   1. Анализ эффективности работы с текущими брендами")
print("   2. Исследование возможности добавления популярных брендов на рынке")
print("   3. Углубление работы в регионах с высокой покупательской способностью")
print("   4. Мониторинг ценовой политики конкурентов")
print("   5. Развитие программ лояльности для повторных продаж")
```

### 3.1 Первичная обработка:

- **Вид топлива**: стандартизирован в коды (F - бензин, D - дизель, E - электро, HYB - гибрид)
- **Тип привода**: унифицирован (FWD, RWD, 4WD, AWD)
- **Объём двигателя**: сложная обработка с исправлением ошибок (замена запятых на точки, удаление литеры L, исправление аномалий)
- **Дата продажи**: создан единый столбец из года и месяца
- **Названия компаний**: устранены неявные дубликаты (Mercur Auto → Меркур Авто)
- **Регион и область**: нормализация регистра (каждое слово с заглавной буквы)
- **Трансмиссия**: упрощена до 2 категорий (Автомат/Механика)

### 3.3 Обработка аномалий:
- Возвраты товара (отрицательные количества) корректно обработаны
- Исправлены очевидные ошибки в данных (например, объём двигателя 400 → 4.0)
- Проверена корректность расчетных полей (Сумма продажи = Цена × Количество)
