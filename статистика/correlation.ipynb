1) Быстрый расчёт r, t, p-value и Fisher CI

```python
import numpy as np
from scipy import stats

x = np.array([3.63,3.02,3.82,3.42,3.59,2.87,3.03,3.46,3.36,3.30])
y = np.array([53.1,49.7,48.4,54.2,54.9,43.7,47.2,45.2,54.4,50.4])
n = len(x)

# Pearson r и p-value (scipy)
r, p_two = stats.pearsonr(x, y)
print(f"Pearson r = {r:.3f}, two-sided p = {p_two:.3f}")

# t-статистика (вручную, проверка)
t_stat = r / np.sqrt((1 - r**2) / (n - 2))
print(f"t = {t_stat:.3f}, df = {n-2}")

# Fisher z CI
z = np.arctanh(r)
se_z = 1/np.sqrt(n-3)
z_crit = stats.norm.ppf(0.975)
lo = np.tanh(z - z_crit * se_z)
hi = np.tanh(z + z_crit * se_z)
print(f"95% Fisher CI for rho: [{lo:.3f}, {hi:.3f}]")
```

Интуиция и формулировка задачи
Корреляция Спирмена оценивает, насколько хорошо взаимосвязь между двумя переменными описывается монотонной функцией. Проще: если при увеличении X значения Y обычно увеличиваются (или уменьшаются), то Спирмен будет положительным (или отрицательным). Важные случаи, когда Спирмен предпочтительнее Пирсона:
переменные порядковые (ранги) или одна/обе измерены в ранговой шкале;
распределения не нормальны;
имеются выбросы;
связь нелинейная, но монотонная (например Y растёт как лог X — Пирсон может недооценить связь, Спирмен — нет).

Итого, Пирсон используется для линейной связи, Спирмен — для монотонной (ранговой) связи.

Кроме того, стоит помнить, что метод Спирмена более устойчив к выбросам, чем корреляция Пирсона. Поскольку он опирается на ранги, отдельные аномальные значения не оказывают значительного влияния на результат — это делает его особенно полезным для анализа данных с шумом или выбросами.

И наконец, как и в случае обычной корреляции Пирсона, следует помнить, что корреляция — это мера ассоциативности, а не причинности. Наличие значимой корреляции не означает, что одно явление вызывает другое.

a) Быстрый расчёт Spearman (scipy)

```python
import numpy as np
from scipy import stats

x = np.array([106,100,86,101,99,103,97,112,113,110])
y = np.array([7,27,2,50,28,29,20,6,12,17])

rho, pval = stats.spearmanr(x, y)
print(f"Spearman rho = {rho:.3f}, p-value = {pval:.3f}")
```
