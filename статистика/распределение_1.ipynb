```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 6))
x = np.linspace(-1, 11, 5000)
ax1.plot(x, uniform.pdf(x), 'b', label='U(0,1)')
ax1.plot(x, uniform.pdf(x, loc=2, scale=5), 'r', label='U(2,7)')
ax1.plot(x, uniform.pdf(x, loc=8, scale=2), 'g', label='U(8,10)')
ax1.set_title('Функция плотности (PDF)')
ax1.grid(True)

ax2.plot(x, uniform.cdf(x), 'b', label='U(0,1)')
ax2.plot(x, uniform.cdf(x, loc=2, scale=5), 'r', label='U(2,7)')
ax2.plot(x, uniform.cdf(x, loc=8, scale=2), 'g', label='U(8,10)')
ax2.set_title('Функция распределения (CDF)')
ax2.grid(True)
for ax in (ax1, ax2):
    ax.legend()
    ax.set_xlabel('x')
    ax.set_ylabel('f(x) / F(x)')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

plt.figure(figsize=(15, 6))
x = np.linspace(-10, 10, 1000)
plt.subplot(121)
for params in [(0,1), (5,3), (-5,0.5)]:
    plt.plot(x, norm.pdf(x, *params), label=f'μ={params[0]}, σ={params[1]}')
plt.title('Функция плотности (PDF)')
plt.grid(True)
plt.subplot(122)
for params in [(0,1), (5,3), (-5,0.5)]:
    plt.plot(x, norm.cdf(x, *params), label=f'μ={params[0]}, σ={params[1]}')
plt.title('Функция распределения (CDF)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

def determine_distribution(column_data, column_name):
    """
    Определяет наиболее вероятное распределение для данных колонки
    """
    # Основные статистики
    print(f"=== Анализ распределения для {column_name} ===")
    print(f"Количество наблюдений: {len(column_data)}")
    print(f"Среднее: {column_data.mean():.4f}")
    print(f"Стандартное отклонение: {column_data.std():.4f}")
    print(f"Минимум: {column_data.min():.4f}")
    print(f"Максимум: {column_data.max():.4f}")
    print(f"Асимметрия: {stats.skew(column_data):.4f}")
    print(f"Эксцесс: {stats.kurtosis(column_data):.4f}")
    print()
    
    # Тестируемые распределения
    distributions = [
        ('norm', stats.norm),
        ('uniform', stats.uniform),
        ('expon', stats.expon),
        ('gamma', stats.gamma),
        ('beta', stats.beta),
        ('lognorm', stats.lognorm),
        ('chi2', stats.chi2),
        ('weibull_min', stats.weibull_min)
    ]
    
    # Тест Колмогорова-Смирнова для каждого распределения
    results = []
    
    for dist_name, dist in distributions:
        try:
            # Подгоняем параметры распределения
            if dist_name in ['norm', 'uniform', 'expon']:
                params = dist.fit(column_data)
            elif dist_name == 'gamma':
                params = dist.fit(column_data, floc=0)  # фиксируем loc=0 для гамма
            elif dist_name == 'beta':
                # Бета требует данные в диапазоне (0,1), нормализуем
                data_normalized = (column_data - column_data.min()) / (column_data.max() - column_data.min())
                data_normalized = data_normalized * 0.99 + 0.005  # избегаем 0 и 1
                params = dist.fit(data_normalized, floc=0, fscale=1)
            else:
                params = dist.fit(column_data)
            
            # Тест Колмогорова-Смирнова
            D, p_value = stats.kstest(column_data, dist_name, params)
            
            results.append({
                'distribution': dist_name,
                'p_value': p_value,
                'ks_statistic': D,
                'params': params
            })
            
        except Exception as e:
            print(f"Ошибка при тестировании {dist_name}: {e}")
    
    # Сортируем по p-value (чем выше, тем лучше соответствие)
    results.sort(key=lambda x: x['p_value'], reverse=True)
    
    print("Результаты тестов распределений (отсортированы по p-value):")
    print("-" * 60)
    for result in results[:5]:  # Показываем топ-5
        print(f"{result['distribution']:12} | p-value: {result['p_value']:.6f} | KS: {result['ks_statistic']:.6f}")
    
    # Визуализация
    plt.figure(figsize=(15, 5))
    
    # Гистограмма с подобранными распределениями
    plt.subplot(1, 3, 1)
    plt.hist(column_data, bins=50, density=True, alpha=0.7, color='skyblue', edgecolor='black')
    
    # Добавляем кривые лучших распределений
    x = np.linspace(column_data.min(), column_data.max(), 1000)
    colors = ['red', 'green', 'orange']
    
    for i, result in enumerate(results[:3]):
        dist_name = result['distribution']
        params = result['params']
        dist = [d[1] for d in distributions if d[0] == dist_name][0]
        
        if dist_name == 'beta':
            # Для бета-распределения используем нормализованные данные
            x_norm = (x - column_data.min()) / (column_data.max() - column_data.min())
            x_norm = x_norm * 0.99 + 0.005
            pdf = dist.pdf(x_norm, *params)
            # Масштабируем обратно
            pdf = pdf / (0.99 * (column_data.max() - column_data.min()))
        else:
            pdf = dist.pdf(x, *params)
        
        plt.plot(x, pdf, color=colors[i], linewidth=2, label=f'{dist_name} (p={result["p_value"]:.4f})')
    
    plt.title(f'Гистограмма {column_name} с подобранными распределениями')
    plt.xlabel('Значение')
    plt.ylabel('Плотность')
    plt.legend()
    
    # Q-Q plot для нормального распределения
    plt.subplot(1, 3, 2)
    stats.probplot(column_data, dist="norm", plot=plt)
    plt.title('Q-Q plot (нормальное распределение)')
    
    # Box plot
    plt.subplot(1, 3, 3)
    plt.boxplot(column_data)
    plt.title('Box plot')
    plt.ylabel('Значение')
    
    plt.tight_layout()
    plt.show()
    
    # Возвращаем лучшее распределение
    best_fit = results[0]
    print(f"\nНаиболее вероятное распределение: {best_fit['distribution']}")
    print(f"p-value: {best_fit['p_value']:.6f}")
    
    return best_fit

# Использование функции
def analyze_column_1_distribution(file_path):
    """
    Анализирует распределение данных в первой колонке файла data.csv
    """
    try:
        # Чтение данных
        data = pd.read_csv(file_path)
        column_1 = data['column_2']
        
        # Анализ распределения
        result = determine_distribution(column_1, "column_2")
        
        return result
        
    except FileNotFoundError:
        print(f"Файл {file_path} не найден!")
        return None
    except Exception as e:
        print(f"Ошибка при анализе данных: {e}")
        return None

# Запуск анализа
if __name__ == "__main__":
    result = analyze_column_1_distribution(path)
```

```python
import pandas as pd
import numpy as np

def calculate_variance(file_path='data.csv'):
    """
    Вычисляет дисперсию для данных из колонки column_1 и округляет до целого числа
    """
    try:
        # Чтение данных
        data = pd.read_csv(file_path)
        column_1 = data['column_1']
        
        # Вычисление дисперсии
        variance = column_1.var()
        
        # Округление до целого числа
        rounded_variance = round(variance)
        
        print(f"Дисперсия column_1: {variance:.4f}")
        print(f"Дисперсия (округлено до целого): {rounded_variance}")
        
        return rounded_variance
        
    except FileNotFoundError:
        print(f"Файл {file_path} не найден!")
        return None
    except KeyError:
        print("Колонка 'column_1' не найдена в файле!")
        return None
    except Exception as e:
        print(f"Ошибка при вычислении дисперсии: {e}")
        return None

# Альтернативный вариант с использованием numpy
def calculate_variance_numpy(file_path='data.csv'):
    """
    Вычисляет дисперсию с использованием numpy
    """
    try:
        # Чтение данных
        data = pd.read_csv(file_path)
        column_1 = data['column_1'].values
        
        # Вычисление дисперсии (ddof=0 для выборочной дисперсии)
        variance = np.var(column_1, ddof=1)
        
        # Округление до целого числа
        rounded_variance = round(variance)
        
        print(f"Дисперсия column_1 (numpy): {variance:.4f}")
        print(f"Дисперсия (округлено до целого): {rounded_variance}")
        
        return rounded_variance
        
    except Exception as e:
        print(f"Ошибка: {e}")
        return None

# Проверка различных методов вычисления дисперсии
def compare_variance_methods(file_path='data.csv'):
    """
    Сравнивает различные методы вычисления дисперсии
    """
    try:
        data = pd.read_csv(file_path)
        column_1 = data['column_1']
        
        # Метод 1: pandas (выборочная дисперсия)
        var_pandas = column_1.var()
        
        # Метод 2: numpy с ddof=1 (выборочная дисперсия)
        var_numpy = np.var(column_1, ddof=1)
        
        # Метод 3: numpy с ddof=0 (генеральная совокупность)
        var_numpy_pop = np.var(column_1, ddof=0)
        
        # Метод 4: вручную
        mean_val = column_1.mean()
        var_manual = sum((x - mean_val) ** 2 for x in column_1) / (len(column_1) - 1)
        
        print("Сравнение методов вычисления дисперсии:")
        print(f"Pandas .var(): {var_pandas:.4f}")
        print(f"Numpy с ddof=1: {var_numpy:.4f}")
        print(f"Numpy с ddof=0: {var_numpy_pop:.4f}")
        print(f"Ручной расчет: {var_manual:.4f}")
        print(f"Округленная дисперсия: {round(var_pandas)}")
        
        return round(var_pandas)
        
    except Exception as e:
        print(f"Ошибка: {e}")
        return None

# Запуск вычисления
if __name__ == "__main__":
    print("=== Вычисление дисперсии для column_1 ===")
    
    # Основной расчет
    variance_result = calculate_variance(path)
    
    if variance_result is not None:
        print(f"\nОтвет: {variance_result}")
    
    # Дополнительная проверка
    print("\n" + "="*50)
    compare_variance_methods(path)
```

```python
import pandas as pd
import numpy as np

def calculate_mean(file_path='data.csv'):
    """
    Вычисляет среднее значение для данных из колонки column_3 и округляет до целого числа
    """
    try:
        # Чтение данных
        data = pd.read_csv(file_path)
        column_3 = data['column_3']
        
        # Вычисление среднего значения (математического ожидания)
        mean_value = column_3.mean()
        
        # Округление до целого числа
        rounded_mean = round(mean_value)
        
        print(f"Среднее значение column_3: {mean_value:.4f}")
        print(f"Среднее значение (округлено до целого): {rounded_mean}")
        
        return rounded_mean
        
    except FileNotFoundError:
        print(f"Файл {file_path} не найден!")
        return None
    except KeyError:
        print("Колонка 'column_3' не найдена в файле!")
        return None
    except Exception as e:
        print(f"Ошибка при вычислении среднего: {e}")
        return None

# Альтернативный вариант с использованием numpy
def calculate_mean_numpy(file_path='data.csv'):
    """
    Вычисляет среднее значение с использованием numpy
    """
    try:
        # Чтение данных
        data = pd.read_csv(file_path)
        column_3 = data['column_3'].values
        
        # Вычисление среднего значения
        mean_value = np.mean(column_3)
        
        # Округление до целого числа
        rounded_mean = round(mean_value)
        
        print(f"Среднее значение column_3 (numpy): {mean_value:.4f}")
        print(f"Среднее значение (округлено до целого): {rounded_mean}")
        
        return rounded_mean
        
    except Exception as e:
        print(f"Ошибка: {e}")
        return None

# Проверка различных методов вычисления среднего
def compare_mean_methods(file_path='data.csv'):
    """
    Сравнивает различные методы вычисления среднего значения
    """
    try:
        data = pd.read_csv(file_path)
        column_3 = data['column_6']
        
        # Метод 1: pandas
        mean_pandas = column_3.mean()
        
        # Метод 2: numpy
        mean_numpy = np.mean(column_3)
        
        # Метод 3: вручную
        mean_manual = sum(column_3) / len(column_3)
        
        # Дополнительная статистика
        median = column_3.median()
        mode = column_3.mode()[0] if not column_3.mode().empty else None
        
        print("Сравнение методов вычисления среднего:")
        print(f"Pandas .mean(): {mean_pandas:.4f}")
        print(f"Numpy mean(): {mean_numpy:.4f}")
        print(f"Ручной расчет: {mean_manual:.4f}")
        print(f"Округленное среднее: {round(mean_pandas)}")
        print(f"\nДополнительная статистика:")
        print(f"Медиана: {median:.4f}")
        print(f"Мода: {mode:.4f}" if mode is not None else "Мода: не определена")
        print(f"Минимум: {column_3.min():.4f}")
        print(f"Максимум: {column_3.max():.4f}")
        print(f"Стандартное отклонение: {column_3.std():.4f}")
        
        return round(mean_pandas)
        
    except Exception as e:
        print(f"Ошибка: {e}")
        return None

# Запуск вычисления
if __name__ == "__main__":
    print("=== Вычисление среднего значения для column_3 ===")
    
    # Основной расчет
    mean_result = calculate_mean(path)
    
    if mean_result is not None:
        print(f"\nОтвет: {mean_result}")
    
    # Дополнительная проверка
    print("\n" + "="*50)
    compare_mean_methods(path)
```

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

def analyze_column_2_distribution(file_path='data.csv'):
    """
    Анализирует распределение данных во второй колонке и определяет,
    какое из указанных распределений наиболее вероятно
    """
    try:
        # Чтение данных
        data = pd.read_csv(file_path)
        column_2 = data['column_3']
        
        print("=== Анализ распределения для column_2 ===")
        print(f"Количество наблюдений: {len(column_2)}")
        print(f"Уникальных значений: {column_2.nunique()}")
        
        # Основные статистики
        stats_summary = {
            'Среднее': column_2.mean(),
            'Медиана': column_2.median(),
            'Стандартное отклонение': column_2.std(),
            'Минимум': column_2.min(),
            'Максимум': column_2.max(),
            'Асимметрия': stats.skew(column_2),
            'Эксцесс': stats.kurtosis(column_2),
            'Дисперсия': column_2.var()
        }
        
        for stat, value in stats_summary.items():
            print(f"{stat}: {value:.4f}")
        
        # Проверяем дискретные распределения
        unique_values = sorted(column_2.unique())
        print(f"\nУникальные значения: {unique_values[:10]}..." if len(unique_values) > 10 else f"Уникальные значения: {unique_values}")
        
        # Проверка на целочисленность (для дискретных распределений)
        is_integer = all(column_2.apply(lambda x: x.is_integer()))
        print(f"Все значения целые: {is_integer}")
        
        # Анализ для каждого распределения
        distributions = {
            'Биномиальное': analyze_binomial(column_2),
            'Бернулли': analyze_bernoulli(column_2),
            'Лапласа': analyze_laplace(column_2),
            'Рэлея': analyze_rayleigh(column_2),
            'Геометрическое': analyze_geometric(column_2)
        }
        
        # Визуализация
        plot_distribution_analysis(column_2, distributions)
        
        # Вывод результатов
        print("\n" + "="*60)
        print("РЕЗУЛЬТАТЫ АНАЛИЗА:")
        print("="*60)
        
        best_fit = None
        best_score = -np.inf
        
        for dist_name, result in distributions.items():
            print(f"\n{dist_name}:")
            for key, value in result.items():
                if key != 'params':
                    print(f"  {key}: {value}")
            
            if result.get('score', -np.inf) > best_score:
                best_score = result['score']
                best_fit = dist_name
        
        print(f"\n🎯 НАИБОЛЕЕ ВЕРОЯТНОЕ РАСПРЕДЕЛЕНИЕ: {best_fit}")
        
        return best_fit
        
    except Exception as e:
        print(f"Ошибка при анализе данных: {e}")
        return None

def analyze_binomial(data):
    """Анализ на соответствие биномиальному распределению"""
    result = {'score': 0}
    
    # Биномиальное: дискретное, целые значения от 0 до n
    if all(data.apply(lambda x: x.is_integer())):
        unique_vals = sorted(data.unique())
        n_max = int(max(unique_vals))
        
        # Подбираем параметры
        p_est = data.mean() / n_max if n_max > 0 else 0
        
        # Проверяем диапазон значений
        if min(unique_vals) >= 0 and max(unique_vals) <= n_max:
            result['score'] += 2
        
        # Проверяем форму распределения
        if 0 < p_est < 1:
            result['score'] += 1
        
        result['params'] = f"n={n_max}, p={p_est:.4f}"
        result['p_estimate'] = p_est
        result['n_estimate'] = n_max
    
    return result

def analyze_bernoulli(data):
    """Анализ на соответствие распределению Бернулли"""
    result = {'score': 0}
    
    # Бернулли: только 0 и 1
    unique_vals = sorted(data.unique())
    
    if len(unique_vals) == 2 and set(unique_vals) == {0, 1}:
        result['score'] += 10  # Очень высокая уверенность
        p_est = data.mean()
        result['params'] = f"p={p_est:.4f}"
        result['p_estimate'] = p_est
    elif len(unique_vals) == 2:
        result['score'] += 5  # Два значения, но не обязательно 0 и 1
    else:
        result['score'] -= 5  # Не распределение Бернулли
    
    return result

def analyze_laplace(data):
    """Анализ на соответствие распределению Лапласа"""
    result = {'score': 0}
    
    # Лапласа: непрерывное, симметричное, тяжелые хвосты
    try:
        # Подгоняем распределение Лапласа
        loc, scale = stats.laplace.fit(data)
        
        # Тест Колмогорова-Смирнова
        D, p_value = stats.kstest(data, 'laplace', args=(loc, scale))
        
        result['score'] = p_value * 10  # Нормализуем p-value
        result['ks_p_value'] = p_value
        result['params'] = f"loc={loc:.4f}, scale={scale:.4f}"
        
    except:
        result['score'] = -1
    
    return result

def analyze_rayleigh(data):
    """Анализ на соответствие распределению Рэлея"""
    result = {'score': 0}
    
    # Рэлея: непрерывное, положительные значения
    if min(data) >= 0:
        result['score'] += 2  # Бонус за неотрицательность
        
        try:
            # Подгоняем распределение Рэлея
            loc, scale = stats.rayleigh.fit(data)
            
            # Тест Колмогорова-Смирнова
            D, p_value = stats.kstest(data, 'rayleigh', args=(loc, scale))
            
            result['score'] += p_value * 8
            result['ks_p_value'] = p_value
            result['params'] = f"loc={loc:.4f}, scale={scale:.4f}"
            
        except:
            result['score'] -= 1
    
    return result

def analyze_geometric(data):
    """Анализ на соответствие геометрическому распределению"""
    result = {'score': 0}
    
    # Геометрическое: дискретное, целые положительные значения
    if all(data.apply(lambda x: x.is_integer())) and min(data) >= 0:
        result['score'] += 2
        
        try:
            # Подгоняем геометрическое распределение
            p_est = 1 / (1 + data.mean())
            
            # Проверяем убывающую природу (типично для геометрического)
            value_counts = data.value_counts().sort_index()
            if len(value_counts) > 1:
                # Проверяем, убывает ли частота с ростом значения
                is_decreasing = all(value_counts.iloc[i] >= value_counts.iloc[i+1] 
                                  for i in range(len(value_counts)-1))
                if is_decreasing:
                    result['score'] += 3
            
            result['params'] = f"p={p_est:.4f}"
            result['p_estimate'] = p_est
            
        except:
            result['score'] -= 1
    
    return result

def plot_distribution_analysis(data, distributions):
    """Визуализация анализа распределений"""
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    
    # Гистограмма
    axes[0, 0].hist(data, bins=50, density=True, alpha=0.7, color='skyblue', edgecolor='black')
    axes[0, 0].set_title('Гистограмма column_2')
    axes[0, 0].set_xlabel('Значение')
    axes[0, 0].set_ylabel('Плотность')
    
    # Box plot
    axes[0, 1].boxplot(data)
    axes[0, 1].set_title('Box plot column_2')
    axes[0, 1].set_ylabel('Значение')
    
    # Q-Q plot (нормальное распределение для сравнения)
    stats.probplot(data, dist="norm", plot=axes[1, 0])
    axes[1, 0].set_title('Q-Q plot (нормальное распределение)')
    
    # График оценок распределений
    dist_names = list(distributions.keys())
    scores = [distributions[name]['score'] for name in dist_names]
    
    colors = ['green' if score > 0 else 'red' for score in scores]
    axes[1, 1].barh(dist_names, scores, color=colors)
    axes[1, 1].set_title('Оценки соответствия распределениям')
    axes[1, 1].set_xlabel('Score')
    
    plt.tight_layout()
    plt.show()

# Запуск анализа
if __name__ == "__main__":
    result = analyze_column_2_distribution('data.csv')
    
    if result:
        print(f"\n📊 ОКОНЧАТЕЛЬНЫЙ ВЫВОД:")
        print(f"Для column_2 наиболее вероятно распределение: {result}")
```
